```python
from __future__ import annotations
import json
from pathlib import Path
import numpy as np
import pandas as pd

V2_OUT = Path('/kaggle/working/iowa_v2')
V2_MIN_TRAIN = 6
V2_FIRST_TEST = 2012
V2_LAST_TEST = 2026


def _date(value, label):
    try:
        parsed = pd.to_datetime(value, format='%Y-%m-%d', errors='raise')
        if pd.isna(parsed):
            raise ValueError('missing date')
        return parsed.strftime('%Y-%m-%d')
    except Exception as exc:
        raise ValueError(f'Invalid {label}: {value!r}') from exc


def _year_value(value, label):
    if isinstance(value, bool) or not np.isfinite(float(value)) or float(value) != int(float(value)):
        raise ValueError(f'Invalid {label}: {value!r}')
    return int(value)


def prepare_q1(snapshot):
    data = snapshot.get('q1_revenue')
    if not isinstance(data, list) or not data:
        raise ValueError('Snapshot requires a nonempty q1_revenue list')
    q1 = pd.DataFrame(data).copy()
    required = {'year', 'q1_usd', 'q1_filed', 'q1_accession', 'q1_tag'}
    if not required.issubset(q1.columns) or q1[list(required)].isna().any().any():
        raise ValueError('Q1 revenue fields are missing')
    q1['year'] = q1['year'].map(lambda v: _year_value(v, 'Q1 year'))
    if q1.year.duplicated().any():
        raise ValueError('Duplicate Q1 revenue years')
    q1['q1_usd'] = pd.to_numeric(q1.q1_usd, errors='raise')
    if (~np.isfinite(q1.q1_usd) | (q1.q1_usd <= 0)).any():
        raise ValueError('Q1 sales must be finite and positive')
    q1['q1_filed'] = q1.q1_filed.map(lambda v: _date(v, 'Q1 filing'))
    for row in q1.itertuples():
        if row.q1_filed < f'{row.year}-03-31' or row.q1_filed > f'{row.year + 1}-05-31':
            raise ValueError(f'Q1 filing outside supported window for {row.year}')
        if not str(row.q1_accession).strip() or not str(row.q1_tag).strip():
            raise ValueError(f'Missing Q1 source identifiers for {row.year}')
    return q1.set_index('year').sort_index()


def prepare_price(snapshot):
    data = snapshot.get('price_index')
    if not data:
        return None
    if not isinstance(data, list):
        raise ValueError('price_index must be a list')
    df = pd.DataFrame(data).copy()
    required = {'year', 'current_april', 'previous_april', 'as_of',
                'current_observation', 'previous_observation', 'source', 'vintage_label'}
    if not required.issubset(df.columns) or df[list(required)].isna().any().any():
        raise ValueError('Price index requires observations, source and as-of vintage label')
    df['year'] = df.year.map(lambda v: _year_value(v, 'price year'))
    if df.year.duplicated().any():
        raise ValueError('Duplicate price-index years')
    for col in ('current_april', 'previous_april'):
        df[col] = pd.to_numeric(df[col], errors='raise')
        if (~np.isfinite(df[col]) | (df[col] <= 0)).any():
            raise ValueError('Price observations must be finite and positive')
    for row in df.itertuples():
        cutoff = f'{row.year}-05-31'
        if _date(row.as_of, 'price as_of') != cutoff or _date(row.vintage_label, 'vintage label') != cutoff:
            raise ValueError(f'Wrong price vintage cutoff in {row.year}')
        if not str(row.source).startswith('https://alfred.stlouisfed.org/'):
            raise ValueError(f'Unverified ALFRED source in {row.year}')
        if (_date(row.current_observation, 'current observation') != f'{row.year}-04-01'
                or _date(row.previous_observation, 'previous observation') != f'{row.year-1}-04-01'):
            raise ValueError(f'Wrong April observations in {row.year}')
    df['price_ratio'] = df.current_april / df.previous_april
    return df.set_index('year').sort_index()


def build_v2_data(snapshot):
    base_rev, weather, meta = read_snapshot(snapshot)
    base = make_dataset(base_rev, weather)
    q1 = prepare_q1(snapshot)
    price = prepare_price(snapshot)
    rows, excluded = [], []
    for row in base.to_dict('records'):
        year = int(row['year'])
        cutoff = row['as_of']
        if year not in q1.index or year - 1 not in q1.index:
            excluded.append((year, 'missing Q1 pair'))
            continue
        current, prev = q1.loc[year], q1.loc[year-1]
        if current.q1_filed > cutoff or prev.q1_filed > cutoff:
            excluded.append((year, 'Q1 not available by cutoff'))
            continue
        row.update(q1_usd=float(current.q1_usd), prior_q1_usd=float(prev.q1_usd),
                   q1_filed=current.q1_filed, prior_q1_filed=prev.q1_filed,
                   q1_accession=current.q1_accession, prior_q1_accession=prev.q1_accession)
        if price is not None and year in price.index:
            p = price.loc[year]
            row.update(price_ratio=float(p.price_ratio), current_april=float(p.current_april),
                       previous_april=float(p.previous_april), price_as_of=p.as_of,
                       price_vintage_label=p.vintage_label, price_source=p.source)
        rows.append(row)
    result = pd.DataFrame(rows)
    if result.empty:
        raise ValueError('No complete, point-in-time compatible annual rows')
    result = result.sort_values('year').reset_index(drop=True)
    if result.year.duplicated().any():
        raise ValueError('Duplicate model years')
    return result, excluded, price is not None, meta


def nitrogen_price_nowcast(previous_q2_usd, current_april_index, previous_april_index):
    vals = np.asarray([previous_q2_usd, current_april_index, previous_april_index], dtype=float)
    if not np.isfinite(vals).all() or (vals <= 0).any():
        raise ValueError('Sales and both nitrogen indices must be finite and positive')
    return float(vals[0] * vals[1] / vals[2])


def run_v2_predictions(data, use_price):
    rows = []
    for _, row in data.iterrows():
        year = int(row.year)
        if not V2_FIRST_TEST <= year <= V2_LAST_TEST:
            continue
        train = data[(data.year < year) & (data.target_filed <= row.as_of)]
        if len(train) < V2_MIN_TRAIN:
            continue
        financial = {
            'previous_q2': float(row.prior_q2_usd),
            'original_weather': ridge_predict(train, row, ['prior_q2_usd', 'rain_mm', 'heavy_rain_days']),
            'q1_ratio': float(row.prior_q2_usd * row.q1_usd / row.prior_q1_usd),
            'q1_ridge': ridge_predict(train, row, ['prior_q2_usd', 'q1_usd']),
        }
        collections = [('financial', financial, len(train))]
        if use_price and pd.notna(row.get('price_ratio')):
            ratio = {**financial, 'price_ratio':nitrogen_price_nowcast(row.prior_q2_usd, row.current_april, row.previous_april)}
            collections.append(('price_ratio', ratio, len(train)))
            price_train = train[train.price_ratio.notna()]
            if len(price_train) >= V2_MIN_TRAIN:
                fitted = {**ratio,
                    'q1_price_ridge':ridge_predict(price_train, row, ['prior_q2_usd', 'q1_usd', 'price_ratio']),
                    'q1_price_weather_ridge':ridge_predict(price_train, row, ['prior_q2_usd', 'q1_usd', 'price_ratio', 'rain_mm', 'heavy_rain_days'])}
                collections.append(('price_fitted', fitted, len(price_train)))
        stage = 'development_2012_2021' if year <= 2021 else 'previously_inspected_2022_2026'
        for cohort, values, price_train_n in collections:
            for model, pred in values.items():
                if not np.isfinite(pred):
                    raise ValueError(f'Non-finite prediction for {model} in {year}')
                actual = float(row.target_q2_usd)
                err = pred - actual
                rows.append({'year':year, 'cohort':cohort, 'stage':stage, 'model':model,
                    'financial_training_rows':len(train),
                    'price_training_rows':price_train_n if cohort == 'price_fitted' else None,
                    'target_usd':actual, 'prediction_usd':pred, 'signed_error_usd':err,
                    'absolute_error_usd':abs(err), 'absolute_pct_error':100*abs(err)/actual})
    pred = pd.DataFrame(rows)
    if pred.empty:
        raise ValueError('Insufficient complete past data for a rolling comparison')
    for cohort, group in pred.groupby('cohort'):
        sets = [frozenset(g.year) for _, g in group.groupby('model')]
        if len(set(sets)) != 1:
            raise AssertionError(f'Model cohorts differ: {cohort}')
    return pred


def study_report(data, pred, scores, excluded, use_price, snapshot):
    labels = {'previous_q2':'Previous Q2', 'original_weather':'Original weather ridge',
              'q1_ratio':'Q1 ratio', 'q1_ridge':'Q1 + previous Q2 ridge',
              'price_ratio':'Nitrogen price ratio', 'q1_price_ridge':'Q1 + price ridge',
              'q1_price_weather_ridge':'Q1 + price + weather ridge'}
    stages = {'development_2012_2021':'2012–2021 development',
              'previously_inspected_2022_2026':'2022–2026 previously inspected',
              'all_2012_2026':'All years in cohort'}
    lines = ['# CF Q2 revenue: follow-up investigation', '',
             '**Status:** Exploratory historical comparison. Every year through 2026 was already inspected, '
             'so no period below is an untouched final test. No candidate has been established as a reliable improvement.', '',
             '## What was tested', '',
             'Each forecast is made on May 31. The previous Q2 benchmark and original rainfall model are compared '
             'with fixed models using current Q1 company sales, and, where available with validated historical '
             'vintages, nitrogen producer prices. The ratio models scale previous Q2 sales by the year-on-year '
             'change in Q1 sales or April nitrogen prices. This is a nowcast, not a forecast made before Q2 begins.', '',
             '## Time-ordered results', '',
             'MAE is mean absolute error; MAPE is mean absolute percentage error. Lower is better. '
             'Within each cohort, all candidates use the same eligible years. Forecasts for each year use only earlier training '
             'labels filed by that year’s May 31.', '',
             ]
    cohort_names = {'financial':'Financial models, all available years',
                    'price_ratio':'Financial and price-ratio models, matched price-available years',
                    'price_fitted':'All models including fitted price models, matched later years'}
    for cohort in ('financial', 'price_ratio', 'price_fitted'):
        if cohort not in set(pred.cohort):
            continue
        lines += ['', f"### {cohort_names[cohort]}", '',
                  '| Period | Model | Years | MAE (USD m) | MAPE |', '|---|---|---:|---:|---:|']
        for stage in stages:
            for model in labels:
                score = scores[(scores.cohort == cohort) & (scores.stage == stage) & (scores.model == model)]
                if len(score) == 1:
                    r = score.iloc[0]
                    lines.append(f'| {stages[stage]} | {labels[model]} | {int(r.n_years)} | {r.mae_usd/1e6:,.1f} | {r.mape_pct:.1f}% |')
    lines += ['', '## Why the original errors were large', '',
              'The earlier baseline carried last year’s Q2 revenue into the next year. This could not '
              'anticipate abrupt company-wide shifts. CF reported that Q1 net sales rose from USD 1,048m '
              'in 2021 to USD 2,868m in 2022. In Q1 2023, CF reported net sales of USD 2,012m '
              'and its UAN segment had a 28% lower average selling price per ton than Q1 2022, reflecting higher global '
              'supply availability and lower energy costs. These company disclosures offer a plausible '
              'explanation of the large errors; this is descriptive and does not establish causation.',
              'Sources: [CF Q1 2022 filing](https://www.sec.gov/Archives/edgar/data/1324404/000132440422000016/cf-20220331.htm), '
              '[CF Q1 2023 filing](https://www.sec.gov/Archives/edgar/data/1324404/000132440423000012/cf-20230331.htm).', '',
              'Signed error below is predicted minus actual; negative means underprediction.']
    for y in (2022, 2023):
        yrows = pred[(pred.cohort == 'financial') & (pred.year == y) & pred.model.isin(['previous_q2','original_weather','q1_ratio','q1_ridge'])]
        if not yrows.empty:
            parts = [f'{labels[r.model]} {r.signed_error_usd/1e6:+,.1f}' for r in yrows.itertuples()]
            lines.append(f'- **{y} signed errors (USD m):** ' + '; '.join(parts) + '.')
    lines += ['', '## Coverage and limits', '',
              f'- Complete common feature rows: {len(data)}; time-ordered prediction years: '
              f'{min(pred.year)}–{max(pred.year)}. Omitted years/reasons: {excluded or "none"}.',
              '- Q1 is taken from an exact January–March company-wide fact filed by May 31; '
              'the company serves other regions, crops and industrial customers.',
              ('- Nitrogen producer prices use the ALFRED graph’s May 31 as-of vintage and exact current/prior April dates; '
               'individual revision dates were not exposed. The index is a national proxy, not CF’s realized prices.' if use_price else
               '- Nitrogen producer prices were excluded because a complete, validated point-in-time '
               'price snapshot was not supplied.'),
              '- Each ridge uses the existing fixed penalty and fold-only scaling; all parameters are '
              're-estimated at each historical prediction date. Changes in weather effects or Q1/Q2 '
              'seasonality can still cause errors. The price-equipped models have fewer training years; the weather-plus-price model has five predictors and can overfit badly.',
              '- Historical NASA weather may have been revised; original weather availability was not reconstructed. '
              'Small samples, structural shifts and trying several candidates limit the strength of any comparison.',
              '- The 2007 Q1 figure is the initially reported 447.7m; a later correction was unavailable at the 2007 cutoff. The 2005 Q1 value is a pre-IPO predecessor figure from a 2006 annual filing; the 2009 Q1 value is from an earnings release.\n'
              '- The simple nitrogen-price ratio is a retrospective candidate, not a proven forecasting model: its selection used these already-seen data. Freeze it before evaluating a future season. If the April index vintage is missing at the cutoff, use the previous-Q2 baseline and record the fallback.\n'
              '- The already-seen 2022–2026 outcomes are not a fresh holdout. Freeze a candidate and evaluate '
              'on a future season such as 2027 before claiming prospective improvement.',
              '- Per-year predictions and signed errors are in `predictions_private.csv`. '
              'Do not upload datasets or generated results to GitHub.', '']
    return '\n'.join(lines)


def run_followup(snapshot):
    if isinstance(snapshot, str):
        snapshot = json.loads(snapshot)
    if not isinstance(snapshot, dict):
        raise ValueError('SNAPSHOT_JSON must be a dict or JSON string')
    data, excluded, use_price, metadata = build_v2_data(snapshot)
    pred = run_v2_predictions(data, use_price)
    scores = pred.groupby(['cohort','stage','model'], as_index=False).agg(
        n_years=('year','size'), mae_usd=('absolute_error_usd','mean'),
        mape_pct=('absolute_pct_error','mean'))
    overall = pred.assign(stage='all_2012_2026').groupby(['cohort','stage','model'], as_index=False).agg(
        n_years=('year','size'), mae_usd=('absolute_error_usd','mean'),
        mape_pct=('absolute_pct_error','mean'))
    scores = pd.concat([scores, overall], ignore_index=True)
    V2_OUT.mkdir(parents=True, exist_ok=True)
    data.to_csv(V2_OUT / 'annual_features_private.csv', index=False)
    pred.to_csv(V2_OUT / 'predictions_private.csv', index=False)
    scores.to_csv(V2_OUT / 'scores_private.csv', index=False)
    notes = {'snapshot_metadata':metadata,'excluded_years':excluded,'uses_price_vintages':use_price,
             'evaluated_years':sorted(set(pred.year.astype(int))),
             'warning':'Historical outcomes to 2026 already inspected; no untouched test.',
             'fixed_models':sorted(set(pred.model)), 'minimum_training_rows':V2_MIN_TRAIN}
    (V2_OUT / 'provenance_private.json').write_text(json.dumps(notes, indent=2, default=str), encoding='utf-8')
    report = study_report(data, pred, scores, excluded, use_price, snapshot)
    (V2_OUT / 'REPORT.md').write_text(report, encoding='utf-8')
    print(f'Follow-up complete: {len(set(pred.year))} financial comparison years, {len(set(pred.model))} fixed candidates across matched cohorts. Outputs:{V2_OUT}')
    return {'data':data, 'predictions':pred, 'scores':scores, 'report':report, 'uses_price':use_price}


if __name__ == '__main__':
    if 'SNAPSHOT_JSON' not in globals():
        raise RuntimeError('Run the original two Kaggle cells first, then this follow-up cell')
    V2_RESULTS = run_followup(SNAPSHOT_JSON)
    try:
        from IPython.display import display, Markdown
        display(Markdown(V2_RESULTS['report']))
    except ImportError:
        print(V2_RESULTS['report'])
```

```python
from __future__ import annotations
import json
from pathlib import Path
import numpy as np
import pandas as pd

V3_OUTPUT = Path('/kaggle/working/iowa_v3')
V3_PRICE_SERIES = 'PCU325311325311A'
V3_FIRST_YEAR = 2016
V3_LAST_YEAR = 2026
V3_GROUPS = ('2016-2021', '2022-2026')


def _strict_date(value, label):
    try:
        parsed = pd.to_datetime(value, format='%Y-%m-%d', errors='raise')
        if pd.isna(parsed):
            raise ValueError('missing date')
        return parsed.strftime('%Y-%m-%d')
    except Exception as exc:
        raise ValueError(f'Invalid {label}: {value!r}') from exc


def _year(value):
    try:
        x = float(value)
        if isinstance(value, bool) or not np.isfinite(x) or x != int(x):
            raise ValueError
        return int(x)
    except (ValueError, TypeError, OverflowError) as exc:
        raise ValueError(f'Invalid year: {value!r}') from exc


def _positive(frame, columns):
    for column in columns:
        frame[column] = pd.to_numeric(frame[column], errors='raise')
        if (~np.isfinite(frame[column]) | (frame[column] <= 0)).any():
            raise ValueError(f'{column} must contain positive finite values')


def prepare_alt_price(extra):
    records = extra.get('alt_price', [])
    if not isinstance(records, list):
        raise ValueError('alt_price must be a list')
    if not records:
        return None
    frame = pd.DataFrame(records).copy()
    required = {'year','current_april','previous_april','as_of','vintage_label',
                'current_observation','previous_observation','source'}
    if not required.issubset(frame.columns) or frame[list(required)].isna().any().any():
        raise ValueError('Alternative price observations or provenance are missing')
    frame['year'] = frame.year.map(_year)
    if frame.year.duplicated().any():
        raise ValueError('Duplicate alternative price year')
    _positive(frame, ['current_april','previous_april'])
    for x in frame.itertuples():
        cutoff = f'{x.year}-05-31'
        if (_strict_date(x.as_of,'alt as_of') != cutoff or
            _strict_date(x.vintage_label,'alt vintage') != cutoff or
            _strict_date(x.current_observation,'alt current month') != f'{x.year}-04-01' or
            _strict_date(x.previous_observation,'alt previous month') != f'{x.year-1}-04-01'):
            raise ValueError(f'Invalid alternative price date in {x.year}')
        if (not str(x.source).startswith('https://alfred.stlouisfed.org/')
                or V3_PRICE_SERIES not in str(x.source)):
            raise ValueError(f'Unverified alternative price source or series in {x.year}')
        if hasattr(x, 'series_id') and x.series_id != V3_PRICE_SERIES:
            raise ValueError(f'Wrong alternative series in {x.year}')
    frame['alt_ratio'] = frame.current_april / frame.previous_april
    return frame.rename(columns={'current_april':'alt_current_april',
        'previous_april':'alt_previous_april','as_of':'alt_as_of',
        'source':'alt_source','vintage_label':'alt_vintage'}).set_index('year').sort_index()


def prepare_q1_volume(extra):
    records = extra.get('q1_volume', [])
    if not isinstance(records, list):
        raise ValueError('q1_volume must be a list')
    if not records:
        return None
    frame = pd.DataFrame(records).copy()
    required = {'year','current_q1_tons','previous_q1_tons','current_filed',
                'previous_filed','current_accession','previous_accession','units'}
    if not required.issubset(frame.columns) or frame[list(required)].isna().any().any():
        raise ValueError('Q1 volume observations or provenance are missing')
    frame['year'] = frame.year.map(_year)
    if frame.year.duplicated().any():
        raise ValueError('Duplicate Q1 volume year')
    _positive(frame, ['current_q1_tons','previous_q1_tons'])
    for x in frame.itertuples():
        cutoff = f'{x.year}-05-31'
        current = _strict_date(x.current_filed,'current Q1 volume disclosure')
        previous = _strict_date(x.previous_filed,'previous Q1 volume disclosure')
        if not (f'{x.year}-03-31' <= current <= cutoff):
            raise ValueError(f'Current Q1 volume unavailable by {cutoff}')
        if not (f'{x.year-1}-03-31' <= previous <= cutoff):
            raise ValueError(f'Previous Q1 volume unavailable by {cutoff}')
        if x.units not in ('product tons','thousand product tons','million product tons'):
            raise ValueError('Both volume figures must use the same product-ton units')
        if not str(x.current_accession).strip() or not str(x.previous_accession).strip():
            raise ValueError('Missing Q1 volume filing accessions')
    frame['volume_ratio'] = frame.current_q1_tons / frame.previous_q1_tons
    return frame.set_index('year').sort_index()


def evaluate_new_candidates(v2_results, extra):
    if not isinstance(v2_results, dict) or 'data' not in v2_results:
        raise ValueError('Run v2 first to create V2_RESULTS with input data')
    if not isinstance(extra, dict):
        raise ValueError('EXTRA_POINTS must be a dictionary')
    base = v2_results['data'].copy()
    needed = {'year','as_of','prior_q2_usd','target_q2_usd','price_ratio','price_as_of'}
    if not needed.issubset(base.columns):
        raise ValueError('The original price-ratio inputs are missing')
    base['year'] = base.year.map(_year)
    if base.year.duplicated().any():
        raise ValueError('Duplicate original model year')
    base = base[(base.year >= V3_FIRST_YEAR) & (base.year <= V3_LAST_YEAR) & base.price_ratio.notna()].copy()
    if base.empty:
        raise ValueError('No valid original price-ratio observations')
    _positive(base, ['prior_q2_usd','target_q2_usd','price_ratio'])
    for x in base.itertuples():
        if _strict_date(x.as_of,'base cutoff') != f'{x.year}-05-31' or _strict_date(x.price_as_of,'base vintage') != f'{x.year}-05-31':
            raise ValueError(f'Original price data lack correct as-of timing in {x.year}')
    alt, volume = prepare_alt_price(extra), prepare_q1_volume(extra)
    datasets = []
    if alt is not None:
        a = base.merge(alt[['alt_ratio','alt_as_of','alt_source','alt_vintage']],left_on='year',right_index=True,validate='one_to_one')
        datasets.append(('alternative_price', a, ['base','alt_price']))
    if volume is not None:
        b = base.merge(volume[['volume_ratio','current_filed','previous_filed','current_accession','previous_accession','units']],left_on='year',right_index=True,validate='one_to_one')
        datasets.append(('q1_volume', b, ['base','base_x_volume']))
    if alt is not None and volume is not None:
        c = base.merge(alt[['alt_ratio','alt_as_of','alt_source','alt_vintage']],left_on='year',right_index=True,validate='one_to_one').merge(
            volume[['volume_ratio','current_filed','previous_filed','current_accession','previous_accession','units']],left_on='year',right_index=True,validate='one_to_one')
        datasets.append(('both', c, ['base','alt_price','base_x_volume','alt_x_volume']))
    if not datasets or not any(not frame.empty for _,frame,_ in datasets):
        raise ValueError('No additional data overlap the established price model')
    records, context = [], []
    for cohort, frame, models in datasets:
        for x in frame.itertuples():
            prediction = {
                'base': x.prior_q2_usd*x.price_ratio,
                'alt_price': x.prior_q2_usd*getattr(x,'alt_ratio',np.nan),
                'base_x_volume': x.prior_q2_usd*x.price_ratio*getattr(x,'volume_ratio',np.nan),
                'alt_x_volume': x.prior_q2_usd*getattr(x,'alt_ratio',np.nan)*getattr(x,'volume_ratio',np.nan),
            }
            context.append({'year':x.year,'cohort':cohort,'original_price_ratio':x.price_ratio,
                'alternative_price_ratio':getattr(x,'alt_ratio',None),'q1_volume_ratio':getattr(x,'volume_ratio',None),
                'price_as_of':x.price_as_of,'alt_vintage':getattr(x,'alt_vintage',None),
                'volume_current_filed':getattr(x,'current_filed',None),'volume_previous_filed':getattr(x,'previous_filed',None),
                'volume_current_accession':getattr(x,'current_accession',None),'volume_previous_accession':getattr(x,'previous_accession',None)})
            for model in models:
                value=float(prediction[model])
                if not np.isfinite(value):
                    raise ValueError(f'Invalid {model} prediction in {x.year}')
                actual=float(x.target_q2_usd)
                records.append({'year':x.year,'cohort':cohort,'period':'2016-2021' if x.year<=2021 else '2022-2026',
                    'model':model,'actual_usd':actual,'prediction_usd':value,'signed_error_usd':value-actual,
                    'absolute_error_usd':abs(value-actual),'absolute_pct_error':100*abs(value-actual)/actual})
    predictions = pd.DataFrame(records)
    if predictions.empty:
        raise ValueError('No common years for the requested comparisons')
    for cohort, g in predictions.groupby('cohort'):
        if len({frozenset(h.year) for _,h in g.groupby('model')}) != 1:
            raise AssertionError(f'Unmatched forecast years: {cohort}')
    per = predictions.groupby(['cohort','period','model'],as_index=False).agg(
        n_years=('year','size'),mae_usd=('absolute_error_usd','mean'),mape_pct=('absolute_pct_error','mean'))
    all_years=predictions.assign(period='all').groupby(['cohort','period','model'],as_index=False).agg(
        n_years=('year','size'),mae_usd=('absolute_error_usd','mean'),mape_pct=('absolute_pct_error','mean'))
    scores=pd.concat([per,all_years],ignore_index=True)
    return predictions,scores,pd.DataFrame(context)


def render_v3_report(pred, scores, extra):
    names={'base':'Frozen nitrogen-price ratio','alt_price':'Alternative producer price ratio (PCU325311325311A)',
           'base_x_volume':'Frozen price ratio × Q1 volume ratio','alt_x_volume':'Alternative producer price ratio × Q1 volume ratio'}
    report=['# Iowa fertilizer revenue: exploratory new-data checks','',
        'All years through 2026 have already been inspected. These are post-hoc comparisons, not an untouched test; '
        'all models in each cohort use exactly the same years. The original April WPU065201 price-ratio candidate remains frozen.','',
        '## Matched historical results','','MAE is mean absolute error in USD millions. Lower is better.','']
    for cohort in ('alternative_price','q1_volume','both'):
        subset=scores[scores.cohort==cohort]
        if subset.empty:
            continue
        report += [f'### {cohort.replace("_"," ").title()}','','| Period | Model | Years | MAE (USD m) | MAPE |',
                   '|---|---|---:|---:|---:|']
        for period in (*V3_GROUPS,'all'):
            for model in names:
                row=subset[(subset.period==period)&(subset.model==model)]
                if len(row)==1:
                    s=row.iloc[0]
                    report.append(f'| {period} | {names[model]} | {int(s.n_years)} | {s.mae_usd/1e6:,.1f} | {s.mape_pct:.1f}% |')
        report += ['']
        for candidate in (x for x in subset.model.unique() if x!='base'):
            enough=True; better=True
            for period in V3_GROUPS:
                a=subset[(subset.period==period)&(subset.model=='base')]
                b=subset[(subset.period==period)&(subset.model==candidate)]
                if len(a)!=1 or len(b)!=1 or int(a.iloc[0].n_years)<3 or int(b.iloc[0].n_years)<3:
                    enough=False; break
                better &= b.iloc[0].mae_usd < a.iloc[0].mae_usd
            verdict=('lower error in both periods' if better else 'did not reduce error in both periods') if enough else 'insufficient matched years in both periods'
            report.append(f'- **{names[candidate]}:** {verdict}. This comparison is exploratory and cannot validate future performance.')
        report += ['']
    report += ['## Limitations and next steps','',
      '- Alternative price values use the same May 31 ALFRED vintage for both April values; the nitrogen index is a national proxy and is not CF’s realized price.',
      '- Q1 sales volume is for product tons, not Iowa-specific corn fertilizer sales; its use assumes Q1 volume changes are informative about Q2 volume changes.',
      '- Simple formulas have no fitted parameters, but their comparison and any selection occurred after the historical outcomes were known.',
      '- Features may cover different years. Compare only rows within the same cohort, and treat small samples and acquisitions/plant outages cautiously.',
      '- A real new test requires a future season such as 2027: archive inputs and the candidate before observing Q2 results.',
      '- Year-by-year actuals, predictions, errors and feature provenance remain in the output files.','']
    return '\n'.join(report)


def run_v3(v2_results, extra):
    pred, scores, context=evaluate_new_candidates(v2_results,extra)
    V3_OUTPUT.mkdir(parents=True,exist_ok=True)
    pred.to_csv(V3_OUTPUT/'predictions_private.csv',index=False)
    scores.to_csv(V3_OUTPUT/'scores_private.csv',index=False)
    context.to_csv(V3_OUTPUT/'provenance_private.csv',index=False)
    text=render_v3_report(pred,scores,extra)
    (V3_OUTPUT/'REPORT.md').write_text(text,encoding='utf-8')
    print(f'New-data analysis complete; outputs:{V3_OUTPUT}')
    return {'predictions':pred,'scores':scores,'context':context,'report':text}


if __name__ == '__main__':
    if 'V2_RESULTS' not in globals() or 'EXTRA_POINTS' not in globals():
        raise RuntimeError('Run the v2 notebook cells and define EXTRA_POINTS first')
    V3_RESULTS=run_v3(V2_RESULTS,EXTRA_POINTS)
    try:
        from IPython.display import display, Markdown
        display(Markdown(V3_RESULTS['report']))
    except ImportError:
        print(V3_RESULTS['report'])
```

```python
from pathlib import Path

import numpy as np
import pandas as pd

V4_OUTPUT = Path('/kaggle/working/iowa_v4')


def diagnose_volume(v3_results):
    pred = v3_results['predictions'].copy()
    context = v3_results['context'].copy()
    pred = pred[(pred.cohort == 'q1_volume') & pred.model.isin(['base', 'base_x_volume'])].copy()
    context = context[context.cohort == 'q1_volume'].copy()
    required = {'year', 'actual_usd', 'prediction_usd', 'model'}
    if pred.empty or not required.issubset(pred.columns):
        raise ValueError('Missing v3 volume cohort predictions')
    if pred.duplicated(['year', 'model']).any() or context.year.duplicated().any():
        raise ValueError('Duplicate volume cohort year')
    years = pred.groupby('model').year.agg(lambda x: frozenset(x))
    if set(years.index) != {'base', 'base_x_volume'} or years['base'] != years['base_x_volume']:
        raise ValueError('Models do not use matching years')
    if not set(pred.year).issubset(range(2016, 2027)):
        raise ValueError('Diagnostic accepts only the already observed 2016--2026 years')
    base = pred[pred.model == 'base'][['year', 'actual_usd', 'prediction_usd']].rename(
        columns={'prediction_usd': 'base_prediction_usd'})
    full = pred[pred.model == 'base_x_volume'][['year', 'actual_usd', 'prediction_usd']].rename(
        columns={'actual_usd': 'full_actual_usd', 'prediction_usd': 'full_prediction_usd'})
    d = base.merge(full, on='year', validate='one_to_one').merge(
        context[['year', 'q1_volume_ratio']], on='year', validate='one_to_one').sort_values('year')
    if len(d) != len(years['base']) or not np.allclose(d.actual_usd, d.full_actual_usd):
        raise ValueError('Missing volume context or mismatched actual sales')
    numerical = d[['actual_usd', 'base_prediction_usd', 'full_prediction_usd', 'q1_volume_ratio']]
    if not np.isfinite(numerical.to_numpy(dtype=float)).all() or (numerical <= 0).any().any():
        raise ValueError('Inputs must be positive and finite')
    if not np.allclose(d.full_prediction_usd, d.base_prediction_usd * d.q1_volume_ratio):
        raise ValueError('V3 full-volume prediction does not match its inputs')
    d = d.drop(columns='full_actual_usd')
    d['damped_prediction_usd'] = d.base_prediction_usd * np.sqrt(d.q1_volume_ratio)
    for name in ('base', 'full', 'damped'):
        d[f'{name}_signed_error_usd'] = d[f'{name}_prediction_usd'] - d.actual_usd
        d[f'{name}_absolute_error_usd'] = d[f'{name}_signed_error_usd'].abs()
    d['volume_effect_usd'] = d.full_prediction_usd - d.base_prediction_usd
    d['full_extra_absolute_error_usd'] = d.full_absolute_error_usd - d.base_absolute_error_usd
    d['period'] = np.where(d.year <= 2021, '2016-2021', '2022-2026')
    results = []
    for period, group in [('2016-2021', d[d.period == '2016-2021']),
                          ('2022-2026', d[d.period == '2022-2026']), ('all', d)]:
        if group.empty:
            continue
        for model in ('base', 'full', 'damped'):
            results.append({'period': period, 'model': model, 'years': len(group),
                            'mae_usd': group[f'{model}_absolute_error_usd'].mean(),
                            'mape_pct': (100 * group[f'{model}_absolute_error_usd'] / group.actual_usd).mean()})
    return d, pd.DataFrame(results)


def render_volume_diagnostic(d, scores):
    labels = {'base': 'Frozen price', 'full': 'Full Q1 volume adjustment',
              'damped': '50%-damped Q1 volume (experimental)'}
    text = ['# Why Q1 shipment volume changed historical errors', '',
            'All 2016–2026 outcomes were examined before this investigation. These results are exploratory, not an untouched or prospective test. '
            'The frozen price forecast remains the candidate for a future test. The damped formula was specified as a sensitivity check after the original results were known.', '',
            'The full adjustment assumes Q2 sales volume changes in the same proportion as Q1; dampening uses the square root of that Q1 ratio. '
            'Neither relationship is guaranteed. Positive signed error means predicted revenue exceeded actual revenue.', '',
            '## Matched historical comparisons', '', '| Period | Model | Years | Mean absolute error ($m) |',
            '|---|---|---:|---:|']
    for x in scores.itertuples():
        text.append(f'| {x.period} | {labels[x.model]} | {x.years} | {x.mae_usd/1e6:,.1f} |')
    early = d[d.period == '2016-2021']
    if not early.empty:
        worse = early[early.full_extra_absolute_error_usd > 0]
        better = early[early.full_extra_absolute_error_usd < 0]
        text += ['', f'The full-volume adjustment raised absolute error in {len(worse)} of {len(early)} early years and lowered it in {len(better)}. '
                 'The yearly table below shows both signed errors and the increase or decrease in absolute error.']
    if not early.empty and (early.year == 2016).any():
        change = early.full_extra_absolute_error_usd.sum()
        y16 = early.loc[early.year == 2016, 'full_extra_absolute_error_usd'].iloc[0]
        text += ['', f'The sum of full-volume minus frozen-model absolute errors over 2016–2021 was USD {change/1e6:,.1f} million. '
                 f'The 2016 contribution was USD {y16/1e6:,.1f} million' +
                 (f' ({100*y16/change:,.1f}% of that signed net difference).' if abs(change) > 1e-9 else '; the net difference was zero, so its share is undefined.')]
        text += ['A share can exceed 100% if other years partly offset 2016; this is a descriptive decomposition, not causation.']
    text += ['', '## Interpretation and source limits', '',
             '- CF reported that adverse weather delayed the 2018 spring application season and shifted some shipments from Q1 to Q2. '
             'Q1 volume declined year over year while Q2 volume rose. This Q2 filing is an after-the-fact explanation, not a forecast input: '
             'https://www.sec.gov/Archives/edgar/data/1324404/000132440418000019/cf-06302018x10q.htm',
             '- The 2015 acquisition of the remaining UK business and 2016 capacity changes complicate year-over-year comparisons; they do not prove the size or direction of any forecast error: '
             'https://www.cfindustries.com/globalassets/cf-industries/media/documents/reports/annual-reports/cfi_2016-annual-report.pdf',
             '- Q1 shipment volumes are company-wide product tons, spanning different product mixes, geographies and customers; they are not Iowa corn use. The national price index is not CF’s realized price.',
             '- The damped factor is an arbitrary sensitivity check and was considered after inspecting results. Its historical score cannot establish that it will improve future forecasts. '
             'A real prospective comparison must fix and archive the rules before future Q2 outcomes are known.',
             '- The detailed year-by-year rows and errors remain in /kaggle/working/iowa_v4.', '']
    return '\n'.join(text)


def run_volume_diagnostic(v3_results):
    details, scores = diagnose_volume(v3_results)
    report = render_volume_diagnostic(details, scores)
    V4_OUTPUT.mkdir(parents=True, exist_ok=True)
    details.to_csv(V4_OUTPUT / 'yearly_diagnostics_private.csv', index=False)
    scores.to_csv(V4_OUTPUT / 'scores_private.csv', index=False)
    (V4_OUTPUT / 'REPORT.md').write_text(report, encoding='utf-8')
    print(f'Volume diagnostic complete; outputs:{V4_OUTPUT}')
    return {'details': details, 'scores': scores, 'report': report}


if __name__ == '__main__':
    if 'V3_RESULTS' not in globals():
        raise RuntimeError('Run the earlier notebook cells to create V3_RESULTS first')
    V4_RESULTS = run_volume_diagnostic(V3_RESULTS)
    try:
        from IPython.display import display, Markdown
        display(Markdown(V4_RESULTS['report']))
        columns = ['year', 'q1_volume_ratio', 'actual_usd', 'base_prediction_usd',
                   'full_prediction_usd', 'damped_prediction_usd', 'base_signed_error_usd',
                   'full_signed_error_usd', 'full_extra_absolute_error_usd']
        display(V4_RESULTS['details'][columns].round(3))
    except ImportError:
        print(V4_RESULTS['report'])
```

```python
from __future__ import annotations

import datetime as dt
import hashlib
import json
import math
import os
from pathlib import Path
from urllib.parse import urlsplit

UTC = dt.timezone.utc
TEST_DATE = dt.date(2027, 5, 31)
FIRST_Q2_RELEASE_DATE = dt.date(2027, 7, 1)
ALFRED_SERIES = 'WPU065201'
MODEL_ID = 'cf-q2-2027-price-and-damped-volume-v1'
DEFAULT_DIR = Path('/kaggle/working/cf_2027_test')


def _date(value, name):
    if not isinstance(value, str):
        raise ValueError(f'{name} must be an ISO YYYY-MM-DD string')
    try:
        result = dt.date.fromisoformat(value)
    except ValueError as exc:
        raise ValueError(f'{name} must be an ISO YYYY-MM-DD date') from exc
    if result.isoformat() != value:
        raise ValueError(f'{name} must be an ISO YYYY-MM-DD date')
    return result


def _timestamp(value, name):
    if not isinstance(value, str) or not value.endswith('Z'):
        raise ValueError(f'{name} must be a UTC ISO timestamp ending in Z')
    try:
        parsed = dt.datetime.fromisoformat(value[:-1] + '+00:00')
    except ValueError as exc:
        raise ValueError(f'Invalid {name}') from exc
    return parsed.astimezone(UTC)


def _positive(value, name):
    if isinstance(value, bool) or not isinstance(value, (int, float)):
        raise ValueError(f'{name} must be a positive number')
    number = float(value)
    if not math.isfinite(number) or number <= 0:
        raise ValueError(f'{name} must be positive and finite')
    return number


def _mapping(value, name, allowed, required):
    if not isinstance(value, dict) or set(value) - set(allowed) or not set(required) <= set(value):
        raise ValueError(f'{name} has missing or unexpected fields')
    return value


def _https(value, name, hosts):
    if not isinstance(value, str):
        raise ValueError(f'{name} must be a source URL')
    u = urlsplit(value)
    if (u.scheme != 'https' or u.username or u.password or u.port not in (None, 443)
        or not u.hostname or not any(u.hostname == h or u.hostname.endswith('.' + h) for h in hosts)):
        raise ValueError(f'{name} must be an official HTTPS source URL')
    return value


def _disclosure(row, name, period_start, period_end, earliest, latest):
    allowed = {'value', 'units', 'period_start', 'period_end', 'published', 'source'}
    _mapping(row, name, allowed, allowed)
    value = _positive(row['value'], name + ' value')
    if row['period_start'] != period_start or row['period_end'] != period_end:
        raise ValueError(f'{name} period must be {period_start} to {period_end}')
    published = _date(row['published'], name + ' published')
    if not earliest <= published <= latest:
        raise ValueError(f'{name} publication date is outside the allowed interval')
    _https(row['source'], name + ' source', ('sec.gov', 'cfindustries.com'))
    return value


def _make_prediction(inputs, now):
    if now.tzinfo is None:
        raise ValueError('The capture clock must be timezone-aware')
    now = now.astimezone(UTC)
    if now.date() != TEST_DATE:
        raise ValueError('Prospective capture is allowed only on 2027-05-31 UTC')
    allowed = {'prior_q2', 'price', 'q1_volume', 'q1_customer_advances'}
    _mapping(inputs, 'inputs', allowed, {'prior_q2', 'price', 'q1_volume'})
    q2 = inputs['prior_q2']
    revenue = _disclosure(q2, '2026 Q2 revenue', '2026-04-01', '2026-06-30',
                          dt.date(2026, 7, 1), TEST_DATE)
    if q2['units'] != 'USD':
        raise ValueError('Prior Q2 net sales units must be USD')

    price = inputs['price']
    if not isinstance(price, dict) or not isinstance(price.get('available'), bool):
        raise ValueError('price.available must be true or false')
    if price['available']:
        keys = {'available', 'series', 'vintage', 'current_observation', 'previous_observation',
                'current_value', 'previous_value', 'current_first_published', 'source', 'captured_at'}
        _mapping(price, 'price', keys, keys)
        if (price['series'] != ALFRED_SERIES or price['vintage'] != '2027-05-31'
            or price['current_observation'] != '2027-04-01'
            or price['previous_observation'] != '2026-04-01'):
            raise ValueError('Price must use both April observations from the May 31 WPU065201 vintage')
        if not dt.date(2027, 4, 1) <= _date(price['current_first_published'], 'price publication') <= TEST_DATE:
            raise ValueError('Current price must first have been published by the cutoff')
        source = _https(price['source'], 'price source', ('alfred.stlouisfed.org',))
        if ALFRED_SERIES not in source:
            raise ValueError('Price source must identify WPU065201')
        captured = _timestamp(price['captured_at'], 'price captured_at')
        if captured.date() != TEST_DATE or captured > now:
            raise ValueError('Price vintage must be captured on May 31 no later than the forecast')
        price_ratio = _positive(price['current_value'], 'current index') / _positive(price['previous_value'], 'previous index')
        main = revenue * price_ratio
        main_status = 'price_ratio'
    else:
        keys = {'available', 'reason', 'checked_at', 'source'}
        _mapping(price, 'unavailable price', keys, keys)
        if not isinstance(price['reason'], str) or len(price['reason'].strip()) < 12:
            raise ValueError('Explain why the April price index is unavailable')
        checked = _timestamp(price['checked_at'], 'price checked_at')
        if checked.date() != TEST_DATE or checked > now:
            raise ValueError('Unavailability must be checked on May 31 no later than forecast')
        source = _https(price['source'], 'price source', ('alfred.stlouisfed.org',))
        if ALFRED_SERIES not in source:
            raise ValueError('Price source must identify WPU065201')
        price_ratio = None
        main = revenue
        main_status = 'previous_q2_fallback'

    volume = inputs['q1_volume']
    if not isinstance(volume, dict) or not isinstance(volume.get('available'), bool):
        raise ValueError('q1_volume.available must be true or false')
    if volume['available']:
        keys = {'available', 'current', 'previous'}
        _mapping(volume, 'q1_volume', keys, keys)
        current = _disclosure(volume['current'], '2027 Q1 volume', '2027-01-01', '2027-03-31',
                              dt.date(2027, 4, 1), TEST_DATE)
        previous = _disclosure(volume['previous'], '2026 Q1 volume', '2026-01-01', '2026-03-31',
                               dt.date(2026, 4, 1), dt.date(2026, 5, 31))
        if volume['current']['units'] != 'thousand product tons' or volume['previous']['units'] != 'thousand product tons':
            raise ValueError('Q1 volumes must both use thousand product tons')
        volume_ratio = current / previous
        sensitivity = main * math.sqrt(volume_ratio) if price['available'] else None
        sensitivity_status = 'damped_volume_on_price' if price['available'] else 'unavailable_missing_price'
    else:
        _mapping(volume, 'unavailable q1_volume', {'available', 'reason'}, {'available', 'reason'})
        if not isinstance(volume['reason'], str) or len(volume['reason'].strip()) < 12:
            raise ValueError('Explain why Q1 volume is unavailable')
        volume_ratio, sensitivity, sensitivity_status = None, None, 'unavailable'
    advances = inputs.get('q1_customer_advances')
    advances_value = None
    advances_note = 'Not supplied'
    if isinstance(advances, dict) and advances.get('available') is False:
        _mapping(advances, 'unavailable customer advances', {'available', 'reason'}, {'available', 'reason'})
        if not isinstance(advances['reason'], str) or len(advances['reason'].strip()) < 12:
            raise ValueError('Explain why customer advances are unavailable')
        advances_note = advances['reason'].strip()
    elif advances is not None:
        fields = {'value', 'units', 'balance_date', 'published', 'source'}
        _mapping(advances, 'Q1 customer advances', fields, fields)
        if advances['units'] != 'USD' or advances['balance_date'] != '2027-03-31':
            raise ValueError('Customer advances must be USD at 2027-03-31')
        amount = advances['value']
        if isinstance(amount, bool) or not isinstance(amount, (int, float)) or not math.isfinite(amount) or amount < 0:
            raise ValueError('Customer advances must be a nonnegative finite number')
        if not dt.date(2027, 4, 1) <= _date(advances['published'], 'customer advances publication') <= TEST_DATE:
            raise ValueError('Customer advances must have been published by the cutoff')
        _https(advances['source'], 'customer advances source', ('sec.gov', 'cfindustries.com'))
        advances_value = float(amount)
        advances_note = 'Partial, company-wide deposits; context only'
    return {'model_id': MODEL_ID, 'target': 'CF consolidated Q2 2027 net sales',
            'target_units': 'USD', 'prediction_cutoff_utc': '2027-05-31T23:59:59.999999Z',
            'captured_at_utc': now.isoformat().replace('+00:00', 'Z'), 'inputs': inputs,
            'price_ratio': price_ratio, 'q1_volume_ratio': volume_ratio,
            'baseline_prediction_usd': revenue, 'main_status': main_status, 'main_prediction_usd': main,
            'sensitivity_status': sensitivity_status, 'sensitivity_prediction_usd': sensitivity,
            'context_customer_advances_usd': advances_value, 'context_customer_advances_note': advances_note,
            'warning': 'Inputs and dates are manually supplied and unverified. Archive a Kaggle version before Q2 disclosure.'}


def _canonical(obj):
    return json.dumps(obj, sort_keys=True, separators=(',', ':'), allow_nan=False, ensure_ascii=False).encode('utf-8')


def _exclusive_write(path, obj):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    fd = os.open(path, os.O_WRONLY | os.O_CREAT | os.O_EXCL, 0o600)
    with os.fdopen(fd, 'w', encoding='utf-8') as f:
        json.dump(obj, f, indent=2, sort_keys=True, allow_nan=False, ensure_ascii=False)
        f.write('\n')


def capture_2027(inputs, output_path=DEFAULT_DIR / 'prediction_2027.json'):
    result = _make_prediction(inputs, dt.datetime.now(UTC))
    envelope = {'prediction': result, 'sha256': hashlib.sha256(_canonical(result)).hexdigest()}
    _exclusive_write(output_path, envelope)
    print('Prediction recorded. Save a Kaggle version before the Q2 announcement; a working file is temporary.')
    return envelope


def _score_prediction(envelope, actual, now):
    if now.tzinfo is None:
        raise ValueError('The scoring clock must be timezone-aware')
    now = now.astimezone(UTC)
    _mapping(envelope, 'forecast envelope', {'prediction', 'sha256'}, {'prediction', 'sha256'})
    prediction = envelope['prediction']
    if not isinstance(prediction, dict) or prediction.get('model_id') != MODEL_ID:
        raise ValueError('Unknown forecast model')
    if hashlib.sha256(_canonical(prediction)).hexdigest() != envelope['sha256']:
        raise ValueError('Forecast digest mismatch; the forecast may have changed')
    captured = _timestamp(prediction.get('captured_at_utc'), 'forecast capture')
    if captured.date() != TEST_DATE or captured >= now:
        raise ValueError('Forecast must be captured on May 31 and before scoring')
    expected = _make_prediction(prediction['inputs'], captured)
    if expected != prediction:
        raise ValueError('Forecast content does not match the fixed formulas')
    if now.date() < FIRST_Q2_RELEASE_DATE:
        raise ValueError('Scoring is only permitted after Q2 ends')
    value = _disclosure(actual, '2027 Q2 actual', '2027-04-01', '2027-06-30',
                        FIRST_Q2_RELEASE_DATE, now.date())
    if actual['units'] != 'USD':
        raise ValueError('Actual Q2 net sales units must be USD')
    rows = []
    for label, field, status_field in [('baseline', 'baseline_prediction_usd', None),
                                       ('main', 'main_prediction_usd', 'main_status'),
                                       ('sensitivity', 'sensitivity_prediction_usd', 'sensitivity_status')]:
        forecast = prediction[field]
        status = prediction[status_field] if status_field else 'previous_q2'
        if forecast is None:
            rows.append({'model': label, 'status': status, 'prediction_usd': None,
                         'absolute_error_usd': None, 'absolute_pct_error': None})
        else:
            error = abs(forecast - value)
            rows.append({'model': label, 'status': status, 'prediction_usd': forecast,
                         'absolute_error_usd': error, 'absolute_pct_error': 100 * error / value})
    return {'model_id': MODEL_ID, 'forecast_sha256': envelope['sha256'], 'actual': actual,
            'scored_at_utc': now.isoformat().replace('+00:00', 'Z'), 'results': rows,
            'warning': 'One year is not enough to establish reliable improvement; verify the independent archive timestamp.'}


def score_2027(forecast_path, actual, output_path=DEFAULT_DIR / 'score_2027.json'):
    with open(forecast_path, encoding='utf-8') as f:
        envelope = json.load(f)
    result = _score_prediction(envelope, actual, dt.datetime.now(UTC))
    _exclusive_write(output_path, result)
    print('Score recorded. Save the output in a new Kaggle version.')
    return result



print("2027 Q2 test prepared; no forecast has been created.")
```

```python
import datetime as dt
import json
import math
import time
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlencode, urlsplit, parse_qs
from urllib.request import Request, urlopen

import numpy as np
import pandas as pd

F_OUTPUT = Path('/kaggle/working/iowa_forecast')
F_CACHE = F_OUTPUT / 'station_year_cache.json'
F_ENDPOINT = 'https://mesonet.agron.iastate.edu/cgi-bin/request/mos.py'
F_STATIONS = ('KAMW', 'KCID', 'KSUX')
F_YEARS = tuple(range(2016, 2027))
F_MIN_DAYS = 40
F_MIN_TRAIN = 6
F_RIDGE = 1.0
F_CONFIG = 'iem-gfs-mos-00z-plus24-p12-apr01-may20-v1'


def _expected_runtimes(year):
    return pd.date_range(f'{year}-04-01', f'{year}-05-20', freq='D', tz='UTC')


def _as_utc(value):
    try:
        value = pd.Timestamp(value)
        if pd.isna(value):
            return None
        return value.tz_localize('UTC') if value.tzinfo is None else value.tz_convert('UTC')
    except (TypeError, ValueError, OverflowError):
        return None


def summarize_station_year(records, station, year, source_url=''):
    if station not in F_STATIONS or year not in F_YEARS:
        raise ValueError('Unexpected station or year')
    if not isinstance(records, list):
        raise ValueError('IEM response is not a JSON row array')
    expected = set(_expected_runtimes(year))
    values, duplicates, invalid = {}, 0, 0
    for row in records:
        if not isinstance(row, dict):
            invalid += 1
            continue
        runtime = _as_utc(row.get('runtime'))
        valid = _as_utc(row.get('ftime'))
        if runtime not in expected or valid is None or valid != runtime + pd.Timedelta(hours=24):
            continue
        if row.get('station') != station or row.get('model') != 'GFS':
            invalid += 1
            continue
        value = row.get('p12')
        if value is None or value in ('', 'M', 'null'):
            continue
        if isinstance(value, bool):
            invalid += 1
            continue
        try:
            value = float(value)
        except (ValueError, TypeError):
            invalid += 1
            continue
        if not np.isfinite(value) or not 0 <= value <= 100:
            invalid += 1
            continue
        if runtime in values:
            if value in values[runtime]:
                duplicates += 1
            values[runtime].add(value)
        else:
            values[runtime] = {value}
    conflicting_days = sum(len(items) > 1 for items in values.values())
    clean_values = [next(iter(items)) for items in values.values() if len(items) == 1]
    count = len(clean_values)
    usable = count >= F_MIN_DAYS and invalid == 0
    return {'year': int(year), 'station': station, 'expected_days': len(expected),
            'valid_days': count, 'coverage_fraction': count / len(expected),
            'duplicate_rows': duplicates, 'conflicting_days': conflicting_days, 'invalid_rows': invalid,
            'station_mean_p12_pct': float(np.mean(clean_values)) if clean_values else None,
            'usable': usable, 'source_url': source_url, 'status': 'ok' if usable else 'incomplete_or_invalid'}


def _query_url(station, year):
    query = {'station': station, 'model': 'GFS', 'sts': f'{year}-04-01T00:00Z',
             'ets': f'{year}-05-21T00:00Z', 'format': 'json'}
    return F_ENDPOINT + '?' + urlencode(query)


def _get_json(url, attempts=3, timeout=30):
    last = None
    for attempt in range(attempts):
        try:
            request = Request(url, headers={'User-Agent': 'IowaForecastResearch/1.0 contact: public research',
                                            'Accept': 'application/json'})
            with urlopen(request, timeout=timeout) as response:
                obj = json.load(response)
            if not isinstance(obj, list):
                raise ValueError('IEM returned an unexpected JSON structure')
            return obj
        except HTTPError as e:
            if e.code == 404:
                return []
            last = e
            if e.code not in (429, 500, 502, 503, 504):
                break
        except (URLError, TimeoutError, OSError, ValueError) as e:
            last = e
        if attempt + 1 < attempts:
            time.sleep(min(2 ** attempt, 4))
    raise RuntimeError(f'IEM request failed after retries: {type(last).__name__}: {last}')


def _load_cache():
    if not F_CACHE.exists():
        return {}
    data = json.loads(F_CACHE.read_text(encoding='utf-8'))
    if data.get('config') != F_CONFIG or not isinstance(data.get('station_years'), dict):
        raise ValueError('Existing station-year cache has an incompatible configuration')
    return data['station_years']


def _write_cache(cache):
    F_OUTPUT.mkdir(parents=True, exist_ok=True)
    tmp = F_OUTPUT / 'station_year_cache.tmp'
    tmp.write_text(json.dumps({'config': F_CONFIG, 'station_years': cache}, indent=2, allow_nan=False), encoding='utf-8')
    tmp.replace(F_CACHE)


def snapshot_forecasts(snapshot):
    if not isinstance(snapshot, list):
        raise ValueError('FORECAST_SNAPSHOT must be a list of station-year summaries')
    req = {'year', 'station', 'expected_days', 'valid_days', 'duplicate_rows', 'conflicting_days',
           'invalid_rows', 'station_mean_p12_pct', 'source_url', 'status'}
    coverage = []
    seen = set()
    for item in snapshot:
        if not isinstance(item, dict) or not req.issubset(item):
            raise ValueError('Missing snapshot summary fields')
        year, station = item['year'], item['station']
        if isinstance(year, bool) or not isinstance(year, int) or year not in F_YEARS or station not in F_STATIONS:
            raise ValueError('Unexpected snapshot year or station')
        if (year, station) in seen:
            raise ValueError('Duplicate snapshot station-year')
        seen.add((year, station))
        for col in ('expected_days', 'valid_days', 'duplicate_rows', 'conflicting_days', 'invalid_rows'):
            if isinstance(item[col], bool) or not isinstance(item[col], int) or item[col] < 0:
                raise ValueError(f'Invalid {col} count')
        if item['expected_days'] != 50 or item['valid_days'] + item['conflicting_days'] > 50:
            raise ValueError('Expected 50 daily 00Z runs per station-year')
        url = urlsplit(item['source_url'])
        params = parse_qs(url.query)
        if (url.scheme != 'https' or url.hostname != 'mesonet.agron.iastate.edu'
                or url.path != '/cgi-bin/request/mos.py'
                or params.get('station') != [station] or params.get('model') != ['GFS']
                or params.get('sts') != [f'{year}-04-01T00:00Z']
                or params.get('ets') != [f'{year}-05-21T00:00Z']
                or params.get('format') != ['json']):
            raise ValueError(f'Invalid source parameters for {station} {year}')
        status = item['status']
        if not isinstance(status, str) or not status.strip():
            raise ValueError('Snapshot status is required')
        mean = item['station_mean_p12_pct']
        if mean is not None:
            if isinstance(mean, bool) or not isinstance(mean, (int, float)) or not np.isfinite(mean) or not 0 <= mean <= 100:
                raise ValueError('Invalid p12 probability mean')
            mean = float(mean)
        if item['valid_days'] > 0 and mean is None:
            raise ValueError('Nonempty station-year must include a mean')
        if item['valid_days'] == 0 and mean is not None:
            raise ValueError('Empty station-year cannot have a mean')
        usable = (status == 'ok' and item['valid_days'] >= F_MIN_DAYS and
                  item['invalid_rows'] == 0 and mean is not None)
        coverage.append({'year': year, 'station': station, 'expected_days': 50,
            'valid_days': item['valid_days'], 'coverage_fraction': item['valid_days']/50,
            'duplicate_rows': item['duplicate_rows'], 'conflicting_days': item['conflicting_days'], 'invalid_rows': item['invalid_rows'],
            'station_mean_p12_pct': mean, 'usable': usable,
            'source_url': item['source_url'], 'status': status})
    expected = {(year, station) for year in F_YEARS for station in F_STATIONS}
    if seen != expected:
        raise ValueError(f'Snapshot must explicitly cover all 33 station-years; missing {len(expected-seen)}')
    cov = pd.DataFrame(coverage).sort_values(['year','station']).reset_index(drop=True)
    annual = []
    for year, group in cov.groupby('year', sort=True):
        complete = len(group) == 3 and set(group.station) == set(F_STATIONS) and group.usable.all()
        annual.append({'year': int(year), 'forecast_mean_p12_pct':
                       float(group.station_mean_p12_pct.mean()) if complete else np.nan,
                       'complete': bool(complete), 'stations_used': int(group.usable.sum())})
    annual = pd.DataFrame(annual)
    F_OUTPUT.mkdir(parents=True, exist_ok=True)
    (F_OUTPUT/'forecast_snapshot_private.json').write_text(
        json.dumps({'config':F_CONFIG, 'method': 'Browser-derived station-year aggregates; raw records not stored',
                    'station_years':snapshot}, indent=2, allow_nan=False), encoding='utf-8')
    cov.to_csv(F_OUTPUT/'coverage_private.csv', index=False)
    annual.to_csv(F_OUTPUT/'annual_forecast_private.csv', index=False)
    print(f'Forecast snapshot: {int(annual.complete.sum())}/11 complete years. '
          'Input summaries are manually derived; raw records and release times were not independently verified.')
    return annual, cov


def collect_forecasts():
    F_OUTPUT.mkdir(parents=True, exist_ok=True)
    cache, coverage = _load_cache(), []
    for year in F_YEARS:
        for station in F_STATIONS:
            key = f'{year}:{station}'
            url = _query_url(station, year)
            if key in cache:
                item = cache[key]
            else:
                try:
                    records = _get_json(url)
                    item = summarize_station_year(records, station, year, url)
                    del records
                    cache[key] = item
                    _write_cache(cache)
                except Exception as exc:
                    item = {'year': year, 'station': station, 'expected_days': 50, 'valid_days': 0,
                            'coverage_fraction': 0.0, 'duplicate_rows': 0, 'conflicting_days': 0, 'invalid_rows': 0,
                            'station_mean_p12_pct': None, 'usable': False, 'source_url': url,
                            'status': 'fetch_error: ' + str(exc)[:180]}
                time.sleep(0.15)
            coverage.append(item)
    cov = pd.DataFrame(coverage)
    annual = []
    for year, group in cov.groupby('year', sort=True):
        complete = len(group) == 3 and set(group.station) == set(F_STATIONS) and group.usable.all()
        annual.append({'year': int(year), 'forecast_mean_p12_pct':
                       float(group.station_mean_p12_pct.mean()) if complete else np.nan,
                       'complete': bool(complete), 'stations_used': int(group.usable.sum())})
    annual = pd.DataFrame(annual)
    cov.to_csv(F_OUTPUT / 'coverage_private.csv', index=False)
    annual.to_csv(F_OUTPUT / 'annual_forecast_private.csv', index=False)
    print(f'IEM forecast summaries: {int(annual.complete.sum())}/11 complete years; '
          f'{int(cov.status.astype(str).str.startswith("fetch_error").sum())} failed station-year requests. '
          f'Outputs:{F_OUTPUT}')
    return annual, cov


def evaluate_forecasts(v2_results, annual, coverage):
    if not isinstance(v2_results, dict) or 'data' not in v2_results:
        raise ValueError('Run the prior follow-up cells to create V2_RESULTS first')
    base = v2_results['data'].copy()
    req = {'year', 'as_of', 'price_as_of', 'price_ratio', 'prior_q2_usd', 'target_q2_usd', 'target_filed'}
    if not req.issubset(base.columns):
        raise ValueError('V2 data is missing time, target or price provenance')
    if not {'year', 'complete', 'forecast_mean_p12_pct'}.issubset(annual.columns):
        raise ValueError('Annual forecast summary fields are missing')
    if base.year.duplicated().any() or annual.year.duplicated().any():
        raise ValueError('Duplicate annual year')
    base = base[base.year.isin(F_YEARS) & base.price_ratio.notna()].copy()
    for x in base.itertuples():
        if str(x.as_of)[:10] != f'{int(x.year)}-05-31' or str(x.price_as_of)[:10] != f'{int(x.year)}-05-31':
            raise ValueError(f'Price vintage/cutoff mismatch in {x.year}')
        if str(x.target_filed)[:10] <= str(x.as_of)[:10]:
            raise ValueError(f'Q2 target was reported by the forecast date in {x.year}')
    annual = annual[annual.complete.eq(True)].copy()
    df = base.merge(annual[['year', 'forecast_mean_p12_pct']], on='year', validate='one_to_one').sort_values('year')
    for col in ('price_ratio', 'prior_q2_usd', 'target_q2_usd'):
        df[col] = pd.to_numeric(df[col], errors='raise')
        if (~np.isfinite(df[col]) | (df[col] <= 0)).any():
            raise ValueError(f'Invalid {col}')
    df['baseline'] = df.prior_q2_usd * df.price_ratio
    df['forecast_mean_p12_pct'] = pd.to_numeric(df.forecast_mean_p12_pct, errors='raise')
    if (~np.isfinite(df.forecast_mean_p12_pct) | ~df.forecast_mean_p12_pct.between(0, 100)).any():
        raise ValueError('Invalid annual forecast feature')
    results, skipped = [], []
    for _, row in df.iterrows():
        year = int(row.year)
        train = df[(df.year < year) & (df.target_filed.astype(str).str[:10] <= str(row.as_of)[:10])]
        if len(train) < F_MIN_TRAIN:
            skipped.append({'year': year, 'reason': 'fewer than six earlier price and forecast years', 'train_years': len(train)})
            continue
        x = train.forecast_mean_p12_pct.to_numpy(dtype=float)
        mean, sd = float(x.mean()), float(x.std(ddof=0))
        if not np.isfinite(sd) or sd <= 1e-12:
            skipped.append({'year': year, 'reason': 'earlier feature has no variation', 'train_years': len(train)})
            continue
        z = (x - mean) / sd
        residual = np.log(train.target_q2_usd.to_numpy(dtype=float) / train.baseline.to_numpy(dtype=float))
        beta = float(np.dot(z, residual) / (np.dot(z, z) + F_RIDGE))
        ztest = (float(row.forecast_mean_p12_pct) - mean) / sd
        log_factor = beta * ztest
        if not np.isfinite(log_factor) or abs(log_factor) > 700:
            skipped.append({'year': year, 'reason': 'non-finite prediction', 'train_years': len(train)})
            continue
        baseline, actual = float(row.baseline), float(row.target_q2_usd)
        forecast = baseline * math.exp(log_factor)
        results.append({'year': year, 'train_years': len(train), 'last_training_year': int(train.year.max()),
                        'feature_mean_p12_pct': float(row.forecast_mean_p12_pct),
                        'train_feature_mean_pct': mean, 'train_feature_sd_pct': sd, 'coefficient': beta,
                        'actual_usd': actual, 'price_only_usd': baseline, 'price_forecast_usd': forecast,
                        'price_abs_error_usd': abs(baseline-actual), 'forecast_abs_error_usd': abs(forecast-actual),
                        'forecast_minus_price_abs_error_usd': abs(forecast-actual)-abs(baseline-actual)})
    pred = pd.DataFrame(results)
    skipped_df = pd.DataFrame(skipped, columns=['year','reason','train_years'])
    lines = ['# Exploratory Iowa archived forecast test', '',
        'Historical results through 2026 were already inspected. These are retrospective, not an untouched test; the frozen 2027 forecast remains unchanged.', '',
        'Feature: equal-station mean GFS MOS p12 probability (%) for the 12 hours ending 24 hours after each 00Z model initialization from April 1 through May 20, in Iowa. Each station-year needs 40 of 50 days; identical duplicate values are counted once and days with conflicting values are excluded. Runtime is initialization, not verified product publication or proof a farmer saw the forecast.', '',
        'A one-coefficient ridge model is fitted on earlier years only, on log(actual / price-only forecast). A training-only mean and standard deviation and fixed ridge penalty 1 are used; the intercept is fixed at zero. Minimum six earlier complete years. Actual outcomes filed after the forecast date are never used for training that forecast.', '',
        f'Complete forecast years: {len(annual)} of 11; matched evaluated years: {len(pred)}.', '']
    if len(pred):
        price_mae = pred.price_abs_error_usd.mean()/1e6
        weather_mae = pred.forecast_abs_error_usd.mean()/1e6
        wins = int((pred.forecast_abs_error_usd < pred.price_abs_error_usd).sum())
        lines += ['| Model | Matched years | MAE (USD millions) |', '|---|---:|---:|',
                  f'| Existing price only | {len(pred)} | {price_mae:,.1f} |',
                  f'| Price + forecast feature | {len(pred)} | {weather_mae:,.1f} |', '',
                  f'Forecast feature had lower error in {wins} of {len(pred)} matched years.', '',
                  'Per-year predictions and training counts are in predictions_private.csv. The small sample, prior inspection, model and feature choices, missing coverage, and potentially revised archives limit interpretation. A lower error here does not establish future improvement or causation.', '']
    else:
        lines += ['No score can be calculated: insufficient complete, temporally eligible forecast history. Check coverage_private.csv for errors and missing years.', '']
    lines += ['IEM documentation: https://mesonet.agron.iastate.edu/mos/ ; original model details: https://vlab.noaa.gov/web/mdl/mav-card .',
              'P12 is probability of measurable precipitation, not expected rainfall amount; product issuance/public availability timing and archive revisions require verification. The compact station-year inputs were summarized via the browser; the raw records and calculations are not independently reverified by this notebook. No rain forecasts after May 20 initialization enter the feature. Station-averaged forecasts do not indicate which forecast a farmer saw or whether any order was placed.',
              'Compare only years present for both models and use a future, pre-recorded test before promoting any candidate.', '']
    F_OUTPUT.mkdir(parents=True, exist_ok=True)
    pred.to_csv(F_OUTPUT / 'predictions_private.csv', index=False)
    skipped_df.to_csv(F_OUTPUT / 'skipped_private.csv', index=False)
    (F_OUTPUT / 'REPORT.md').write_text('\n'.join(lines), encoding='utf-8')
    print('\n'.join(lines))
    return {'predictions': pred, 'skipped': skipped_df, 'data': df, 'report': '\n'.join(lines), 'coverage': coverage}
```

```python
from pathlib import Path
import math
import numpy as np
import pandas as pd

def investigate_forecast_2025(forecast_results):
    p = forecast_results['predictions'].copy().sort_values('year')
    data = forecast_results['data'].copy()
    if p.empty or p.year.duplicated().any() or not (p.year == 2025).any():
        raise ValueError('Need unique matched forecast predictions including 2025')
    required = ['price_only_usd', 'price_forecast_usd', 'actual_usd', 'feature_mean_p12_pct',
                'train_feature_mean_pct', 'train_feature_sd_pct', 'coefficient']
    if not set(required).issubset(p.columns) or not np.isfinite(p[required].to_numpy(float)).all():
        raise ValueError('Missing or nonfinite model quantities')
    if (p[['price_only_usd', 'price_forecast_usd', 'actual_usd']] <= 0).any().any():
        raise ValueError('Predictions and actual values must be positive')
    p['half_forecast_usd'] = np.sqrt(p.price_only_usd * p.price_forecast_usd)
    cap_predictions, prior_min, prior_max = [], [], []
    for row in p.itertuples():
        train = data[(data.year < row.year) & (data.target_filed.astype(str).str[:10] <= str(row.year)+'-05-31')]
        if len(train) != row.train_years:
            raise ValueError('Training cohort differs from the original forecast')
        low, high = float(train.forecast_mean_p12_pct.min()), float(train.forecast_mean_p12_pct.max())
        x = np.clip(row.feature_mean_p12_pct, low, high)
        z = (x - row.train_feature_mean_pct) / row.train_feature_sd_pct
        cap_predictions.append(row.price_only_usd * math.exp(row.coefficient * z))
        prior_min.append(low); prior_max.append(high)
    p['capped_forecast_usd'] = cap_predictions
    p['prior_feature_min_pct'], p['prior_feature_max_pct'] = prior_min, prior_max
    p['feature_z'] = (p.feature_mean_p12_pct-p.train_feature_mean_pct)/p.train_feature_sd_pct
    models = {'price_only':'price_only_usd','full':'price_forecast_usd',
              'half':'half_forecast_usd','capped':'capped_forecast_usd'}
    for label, col in models.items():
        p[label+'_error_usd'] = (p[col]-p.actual_usd).abs()
    scores=[]
    for omitted in [None, 2023, 2026]:
        group = p if omitted is None else p[p.year != omitted]
        if omitted is not None and len(group) != len(p)-1:
            continue
        for label in models:
            scores.append({'omitted_year': 'none' if omitted is None else str(omitted),
                           'model':label,'years':len(group),'mae_musd':group[label+'_error_usd'].mean()/1e6,
                           'wins_vs_price':int((group[label+'_error_usd'] < group.price_only_error_usd).sum())})
    score = pd.DataFrame(scores)
    out = Path('/kaggle/working/iowa_forecast_diagnostic')
    out.mkdir(parents=True, exist_ok=True)
    p.to_csv(out/'yearly_diagnostic_private.csv', index=False)
    score.to_csv(out/'scores_private.csv', index=False)
    r = p[p.year == 2025].iloc[0]
    report = f'''# 2025 forecast error: exploratory investigation

The price-only prediction was ${r.price_only_usd/1e6:,.1f}m against actual revenue ${r.actual_usd/1e6:,.1f}m. The annual forecast feature was {r.feature_mean_p12_pct:.2f}%, versus prior-years' mean {r.train_feature_mean_pct:.2f}% (z={r.feature_z:.2f}, prior range {r.prior_feature_min_pct:.2f}–{r.prior_feature_max_pct:.2f}%). Its fitted coefficient was {r.coefficient:.4f}; the full model multiplied the price prediction by {r.price_forecast_usd/r.price_only_usd:.3f}, moving it to ${r.price_forecast_usd/1e6:,.1f}m and worsening absolute error from ${r.price_only_error_usd/1e6:,.1f}m to ${r.full_error_usd/1e6:,.1f}m.

Two post-hoc sensitivities are compared with the same price-only and full predictions in the same five rolling years: half the log adjustment and, separately, clipping the forecast feature to the range of the training years. They use the original coefficients and only earlier-year information, but were selected after inspecting these results. They are not independent tests or proven improvements. See scores_private.csv for MAE, wins, and sensitivity to excluding a test year; see yearly_diagnostic_private.csv for annual results and training-range diagnostics. The frozen 2027 protocol is unchanged.

The model used precipitation probability at three Iowa stations, not rain amount or actual farmer orders. This statistical diagnosis does not identify the cause of company revenue or establish that an individual farmer used the forecast. Q2 company disclosures are after the May 31 prediction cutoff and are explanations only, never inputs.

Sources: https://www.sec.gov/Archives/edgar/data/1324404/000132440425000024/cf-20250630.htm ; https://www.nass.usda.gov/Statistics_by_State/Iowa/Publications/Crop_Progress_%26_Condition/2025/IA-Crop-Progress-05-12-25.pdf ; https://www.sec.gov/Archives/edgar/data/1324404/000132440425000013/cf-05072025_ex991xearnings.htm
'''
    (out/'REPORT.md').write_text(report, encoding='utf8')
    print(report)
    display(score.round(2))
    display((p[['year','price_only_error_usd','full_error_usd','half_error_usd','capped_error_usd']].set_index('year')/1e6).round(1))
    return {'predictions':p, 'scores':score, 'report':report}

DIAGNOSTIC_2025 = investigate_forecast_2025(FORECAST_RESULTS)
```

```python
import datetime as dt
import math
from pathlib import Path
from urllib.parse import urlsplit

import numpy as np
import pandas as pd

FW_YEARS = range(2016, 2027)
FW_MIN_TRAIN = 6
FW_RIDGE = 1.0
FW_OUT = Path('/kaggle/working/iowa_fieldwork')


def _fw_date(value):
    if not isinstance(value, str):
        raise ValueError('Use ISO YYYY-MM-DD date strings')
    result = dt.date.fromisoformat(value)
    if result.isoformat() != value:
        raise ValueError('Use ISO YYYY-MM-DD date strings')
    return result


def _fw_expected(year, weeks):
    if weeks not in (1, 6):
        raise ValueError('Only the predefined one-week or six-week scopes are supported')
    may20 = dt.date(year, 5, 20)
    latest_sunday = may20 - dt.timedelta(days=(may20.weekday() + 1) % 7)
    return [latest_sunday - dt.timedelta(days=7*i) for i in range(weeks-1, -1, -1)]


def summarize_fieldwork(snapshot, weeks=6):
    if weeks not in (1, 6) or not isinstance(snapshot, list):
        raise ValueError('Need a list of one-week or six-week summaries')
    records = {}
    for item in snapshot:
        if not isinstance(item, dict) or not {'year','week_end','release_date','days_suitable','source_url'}.issubset(item):
            raise ValueError('Weekly snapshot missing required fields')
        year = item['year']
        if isinstance(year, bool) or not isinstance(year, int) or year not in FW_YEARS:
            raise ValueError('Unexpected fieldwork year')
        end = _fw_date(item['week_end'])
        if end not in _fw_expected(year, weeks):
            raise ValueError('Week end does not match fixed annual window')
        key = (year, end)
        if key in records:
            raise ValueError('Duplicate weekly report')
        val, release, url = item['days_suitable'], item['release_date'], item['source_url']
        if val is None:
            if release is not None or url is not None:
                raise ValueError('An unavailable report needs null date and source')
            records[key] = None
            continue
        if isinstance(val, bool) or not isinstance(val, (int,float)) or not np.isfinite(val) or not 0 <= val <= 7:
            raise ValueError('Weekly days suitable must be finite and in [0,7]')
        release = _fw_date(release)
        if not end < release <= dt.date(year,5,31):
            raise ValueError('Report date must be after week end and by cutoff')
        parsed = urlsplit(url)
        if parsed.scheme != 'https' or not parsed.hostname or not (parsed.hostname == 'usda.gov' or parsed.hostname.endswith('.usda.gov') or parsed.hostname in ('iowaagriculture.gov', 'publications.iowa.gov', 'k923.fm', 'whoradio.iheart.com')):
            raise ValueError('Use an official USDA/Iowa or explicitly recognized secondary HTTPS source URL')
        records[key] = {'year':year, 'week_end':end.isoformat(), 'release_date':release.isoformat(),
                        'days_suitable':float(val), 'source_url':url}
    coverage, annual = [], []
    for year in FW_YEARS:
        expected = _fw_expected(year,weeks)
        selected = [records.get((year,end)) for end in expected]
        ok = all(x is not None for x in selected)
        for end,item in zip(expected,selected):
            coverage.append(item if item else {'year':year,'week_end':end.isoformat(),
                 'release_date':None,'days_suitable':None,'source_url':None})
        annual.append({'year':year, 'fieldwork_complete':ok, 'weeks_expected':weeks,
                       'weeks_present':sum(x is not None for x in selected),
                       'fieldwork_mean_days':float(np.mean([x['days_suitable'] for x in selected])) if ok else np.nan,
                       'first_week_end':expected[0].isoformat(),'last_week_end':expected[-1].isoformat()})
    return pd.DataFrame(annual), pd.DataFrame(coverage)


def run_fieldwork_research(forecast_results, snapshot, weeks=6):
    annual, coverage = summarize_fieldwork(snapshot, weeks=weeks)
    if not isinstance(forecast_results,dict) or 'data' not in forecast_results:
        raise ValueError('Run the existing forecast notebook cells first')
    base = forecast_results['data'].copy()
    fields = {'year','as_of','price_as_of','target_filed','target_q2_usd','prior_q2_usd',
              'price_ratio','forecast_mean_p12_pct'}
    if not fields.issubset(base.columns) or base.year.duplicated().any():
        raise ValueError('Expected unique forecast research base rows')
    for row in base.itertuples():
        cutoff = f'{int(row.year)}-05-31'
        if str(row.as_of)[:10] != cutoff or str(row.price_as_of)[:10] != cutoff or str(row.target_filed)[:10] <= cutoff:
            raise ValueError('Cutoff, price vintage, or target release inconsistency')
    data = base.merge(annual[annual.fieldwork_complete], on='year', how='inner', validate='one_to_one').sort_values('year')
    required = ['target_q2_usd','prior_q2_usd','price_ratio','forecast_mean_p12_pct','fieldwork_mean_days']
    for field in required:
        data[field] = pd.to_numeric(data[field], errors='raise')
        if not np.isfinite(data[field]).all():
            raise ValueError(f'Nonfinite data for {field}')
    if (data[['target_q2_usd','prior_q2_usd','price_ratio']] <= 0).any().any():
        raise ValueError('Sales and price inputs must be positive')
    data['baseline'] = data.prior_q2_usd * data.price_ratio
    models = {'price_only':[], 'price_forecast':['forecast_mean_p12_pct'],
              'price_fieldwork':['fieldwork_mean_days'],
              'price_forecast_fieldwork':['forecast_mean_p12_pct','fieldwork_mean_days']}
    results, skipped = [], []
    for row in data.itertuples():
        train = data[(data.year < row.year) & (data.target_filed.astype(str).str[:10] <= str(row.as_of)[:10])]
        if len(train) < FW_MIN_TRAIN:
            skipped.append({'year':int(row.year), 'reason':'fewer than six earlier complete years','train_years':len(train)})
            continue
        feature_names = ['forecast_mean_p12_pct','fieldwork_mean_days']
        x = train[feature_names].to_numpy(float)
        mu, sd = x.mean(axis=0), x.std(axis=0,ddof=0)
        if (sd <= 1e-12).any() or not np.isfinite(sd).all():
            skipped.append({'year':int(row.year), 'reason':'no training feature variation','train_years':len(train)})
            continue
        z = (x-mu)/sd
        z_test = (np.array([getattr(row,col) for col in feature_names],float)-mu)/sd
        response = np.log(train.target_q2_usd.to_numpy(float)/train.baseline.to_numpy(float))
        for label, cols in models.items():
            if not cols:
                pred = float(row.baseline)
            else:
                indexes = [feature_names.index(col) for col in cols]
                xx = z[:,indexes]
                beta = np.linalg.solve(xx.T@xx + FW_RIDGE*np.eye(len(cols)), xx.T@response)
                exponent = float(z_test[indexes]@beta)
                if not np.isfinite(exponent) or abs(exponent) > 700:
                    raise ValueError('Nonfinite or extreme forecast')
                pred = float(row.baseline)*math.exp(exponent)
            results.append({'year':int(row.year),'model':label,'train_years':len(train),
                            'prediction_usd':pred,'actual_usd':float(row.target_q2_usd),
                            'absolute_error_usd':abs(pred-float(row.target_q2_usd))})
    predictions = pd.DataFrame(results, columns=['year','model','train_years','prediction_usd','actual_usd','absolute_error_usd'])
    skipped = pd.DataFrame(skipped,columns=['year','reason','train_years'])
    if not predictions.empty:
        counts = predictions.groupby('model').year.apply(lambda x:tuple(sorted(x)))
        if len(set(counts)) != 1:
            raise AssertionError('Candidate test years differ')
        scores = predictions.groupby('model',as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    else:
        scores = pd.DataFrame(columns=['model','years','mae_usd'])
    fw_window = 'six-week average' if weeks == 6 else 'latest eligible single week'
    report = f'''# Iowa fieldwork: exploratory comparison

Fieldwork feature: {fw_window} of USDA weekly Iowa days suitable for fieldwork. The eligible Sunday week endings are on or before May 20; every source must be dated by May 31. All expected weeks must be present. Two inputs (2016 and 2018) rely on contemporary secondary reproductions explicitly attributed to USDA; their recorded dates refer to those reproductions, not independently verified dates of the original reports. A weekly opportunity for fieldwork does not establish fertilizer purchases. Source values and dates were entered from browser-visible sources and not independently verified by this notebook; original report availability and possible revisions remain uncertain.

The price-only baseline and three alternatives (forecast, fieldwork, and both) use the same eligible years. Each model uses the same earlier complete training cohort, earlier filed targets, training-only scaling, a fixed ridge penalty of one, and an intercept fixed at zero for log(actual/baseline). At least six earlier years are required. The 2-feature model may overfit such a small sample. All outcomes through 2026 were already inspected; the feature and model choices are exploratory and selected after seeing prior results. None is a new independent holdout. The frozen 2027 protocol is unchanged.

Complete fieldwork years: {int(annual.fieldwork_complete.sum())}/11. Joined forecast, price and fieldwork years: {len(data)}. Evaluation years: {predictions.year.nunique() if not predictions.empty else 0}. See coverage_private.csv, annual_private.csv, predictions_private.csv, scores_private.csv and skipped_private.csv.
'''
    FW_OUT.mkdir(parents=True,exist_ok=True)
    coverage.to_csv(FW_OUT/'coverage_private.csv',index=False)
    annual.to_csv(FW_OUT/'annual_private.csv',index=False)
    predictions.to_csv(FW_OUT/'predictions_private.csv',index=False)
    scores.to_csv(FW_OUT/'scores_private.csv',index=False)
    skipped.to_csv(FW_OUT/'skipped_private.csv',index=False)
    (FW_OUT/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    return {'annual':annual,'coverage':coverage,'data':data,'predictions':predictions,'scores':scores,'skipped':skipped,'report':report}
```

```python
import datetime as dt
import math
from pathlib import Path
from urllib.parse import urlsplit

import numpy as np
import pandas as pd

FW_YEARS = range(2016, 2027)
FW_MIN_TRAIN = 6
FW_RIDGE = 1.0
FW_OUT = Path('/kaggle/working/iowa_fieldwork')


def _fw_date(value):
    if not isinstance(value, str):
        raise ValueError('Use ISO YYYY-MM-DD date strings')
    result = dt.date.fromisoformat(value)
    if result.isoformat() != value:
        raise ValueError('Use ISO YYYY-MM-DD date strings')
    return result


def _fw_expected(year, weeks):
    if weeks not in (1, 6):
        raise ValueError('Only the predefined one-week or six-week scopes are supported')
    may20 = dt.date(year, 5, 20)
    latest_sunday = may20 - dt.timedelta(days=(may20.weekday() + 1) % 7)
    return [latest_sunday - dt.timedelta(days=7*i) for i in range(weeks-1, -1, -1)]


def summarize_fieldwork(snapshot, weeks=6):
    if weeks not in (1, 6) or not isinstance(snapshot, list):
        raise ValueError('Need a list of one-week or six-week summaries')
    records = {}
    for item in snapshot:
        if not isinstance(item, dict) or not {'year','week_end','release_date','days_suitable','source_url'}.issubset(item):
            raise ValueError('Weekly snapshot missing required fields')
        year = item['year']
        if isinstance(year, bool) or not isinstance(year, int) or year not in FW_YEARS:
            raise ValueError('Unexpected fieldwork year')
        end = _fw_date(item['week_end'])
        if end not in _fw_expected(year, weeks):
            raise ValueError('Week end does not match fixed annual window')
        key = (year, end)
        if key in records:
            raise ValueError('Duplicate weekly report')
        val, release, url = item['days_suitable'], item['release_date'], item['source_url']
        if val is None:
            if release is not None or url is not None:
                raise ValueError('An unavailable report needs null date and source')
            records[key] = None
            continue
        if isinstance(val, bool) or not isinstance(val, (int,float)) or not np.isfinite(val) or not 0 <= val <= 7:
            raise ValueError('Weekly days suitable must be finite and in [0,7]')
        release = _fw_date(release)
        if not end < release <= dt.date(year,5,31):
            raise ValueError('Report date must be after week end and by cutoff')
        parsed = urlsplit(url)
        if parsed.scheme != 'https' or not parsed.hostname or not (parsed.hostname == 'usda.gov' or parsed.hostname.endswith('.usda.gov') or parsed.hostname in ('iowaagriculture.gov', 'publications.iowa.gov', 'k923.fm', 'whoradio.iheart.com', 'news.utcrops.com')):
            raise ValueError('Use an official USDA/Iowa or explicitly recognized secondary HTTPS source URL')
        records[key] = {'year':year, 'week_end':end.isoformat(), 'release_date':release.isoformat(),
                        'days_suitable':float(val), 'source_url':url, 'basis':item.get('basis', 'not_stated')}
    coverage, annual = [], []
    for year in FW_YEARS:
        expected = _fw_expected(year,weeks)
        selected = [records.get((year,end)) for end in expected]
        ok = all(x is not None for x in selected)
        for end,item in zip(expected,selected):
            coverage.append(item if item else {'year':year,'week_end':end.isoformat(),
                 'release_date':None,'days_suitable':None,'source_url':None})
        annual.append({'year':year, 'fieldwork_complete':ok, 'weeks_expected':weeks,
                       'weeks_present':sum(x is not None for x in selected),
                       'fieldwork_mean_days':float(np.mean([x['days_suitable'] for x in selected])) if ok else np.nan,
                       'first_week_end':expected[0].isoformat(),'last_week_end':expected[-1].isoformat()})
    return pd.DataFrame(annual), pd.DataFrame(coverage)


def run_fieldwork_research(forecast_results, snapshot, weeks=6):
    annual, coverage = summarize_fieldwork(snapshot, weeks=weeks)
    if not isinstance(forecast_results,dict) or 'data' not in forecast_results:
        raise ValueError('Run the existing forecast notebook cells first')
    base = forecast_results['data'].copy()
    fields = {'year','as_of','price_as_of','target_filed','target_q2_usd','prior_q2_usd',
              'price_ratio','forecast_mean_p12_pct'}
    if not fields.issubset(base.columns) or base.year.duplicated().any():
        raise ValueError('Expected unique forecast research base rows')
    for row in base.itertuples():
        cutoff = f'{int(row.year)}-05-31'
        if str(row.as_of)[:10] != cutoff or str(row.price_as_of)[:10] != cutoff or str(row.target_filed)[:10] <= cutoff:
            raise ValueError('Cutoff, price vintage, or target release inconsistency')
    data = base.merge(annual[annual.fieldwork_complete], on='year', how='inner', validate='one_to_one').sort_values('year')
    required = ['target_q2_usd','prior_q2_usd','price_ratio','forecast_mean_p12_pct','fieldwork_mean_days']
    for field in required:
        data[field] = pd.to_numeric(data[field], errors='raise')
        if not np.isfinite(data[field]).all():
            raise ValueError(f'Nonfinite data for {field}')
    if (data[['target_q2_usd','prior_q2_usd','price_ratio']] <= 0).any().any():
        raise ValueError('Sales and price inputs must be positive')
    data['baseline'] = data.prior_q2_usd * data.price_ratio
    models = {'price_only':[], 'price_forecast':['forecast_mean_p12_pct'],
              'price_fieldwork':['fieldwork_mean_days'],
              'price_forecast_fieldwork':['forecast_mean_p12_pct','fieldwork_mean_days']}
    results, skipped = [], []
    for row in data.itertuples():
        train = data[(data.year < row.year) & (data.target_filed.astype(str).str[:10] <= str(row.as_of)[:10])]
        if len(train) < FW_MIN_TRAIN:
            skipped.append({'year':int(row.year), 'reason':'fewer than six earlier complete years','train_years':len(train)})
            continue
        feature_names = ['forecast_mean_p12_pct','fieldwork_mean_days']
        x = train[feature_names].to_numpy(float)
        mu, sd = x.mean(axis=0), x.std(axis=0,ddof=0)
        if (sd <= 1e-12).any() or not np.isfinite(sd).all():
            skipped.append({'year':int(row.year), 'reason':'no training feature variation','train_years':len(train)})
            continue
        z = (x-mu)/sd
        z_test = (np.array([getattr(row,col) for col in feature_names],float)-mu)/sd
        response = np.log(train.target_q2_usd.to_numpy(float)/train.baseline.to_numpy(float))
        for label, cols in models.items():
            if not cols:
                pred = float(row.baseline)
            else:
                indexes = [feature_names.index(col) for col in cols]
                xx = z[:,indexes]
                beta = np.linalg.solve(xx.T@xx + FW_RIDGE*np.eye(len(cols)), xx.T@response)
                exponent = float(z_test[indexes]@beta)
                if not np.isfinite(exponent) or abs(exponent) > 700:
                    raise ValueError('Nonfinite or extreme forecast')
                pred = float(row.baseline)*math.exp(exponent)
            results.append({'year':int(row.year),'model':label,'train_years':len(train),
                            'prediction_usd':pred,'actual_usd':float(row.target_q2_usd),
                            'absolute_error_usd':abs(pred-float(row.target_q2_usd))})
    predictions = pd.DataFrame(results, columns=['year','model','train_years','prediction_usd','actual_usd','absolute_error_usd'])
    skipped = pd.DataFrame(skipped,columns=['year','reason','train_years'])
    if not predictions.empty:
        counts = predictions.groupby('model').year.apply(lambda x:tuple(sorted(x)))
        if len(set(counts)) != 1:
            raise AssertionError('Candidate test years differ')
        scores = predictions.groupby('model',as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    else:
        scores = pd.DataFrame(columns=['model','years','mae_usd'])
    fw_window = 'six-week average' if weeks == 6 else 'latest eligible single week'
    report = f'''# Iowa fieldwork: exploratory comparison

Fieldwork feature: {fw_window} of USDA weekly Iowa days suitable for fieldwork. The eligible Sunday week endings are on or before May 20; every source must be dated by May 31. All expected weeks must be present. Some input values may come from secondary reproductions or from the preceding-week column of a later report; source dates refer to the specific document used and previous-week entries may have been revised. See coverage_private.csv for provenance. A weekly opportunity for fieldwork does not establish fertilizer purchases. Source values and dates were entered from browser-visible sources and not independently verified by this notebook; original report availability and possible revisions remain uncertain.

The price-only baseline and three alternatives (forecast, fieldwork, and both) use the same eligible years. Each model uses the same earlier complete training cohort, earlier filed targets, training-only scaling, a fixed ridge penalty of one, and an intercept fixed at zero for log(actual/baseline). At least six earlier years are required. The 2-feature model may overfit such a small sample. All outcomes through 2026 were already inspected; the feature and model choices are exploratory and selected after seeing prior results. None is a new independent holdout. The frozen 2027 protocol is unchanged.

Complete fieldwork years: {int(annual.fieldwork_complete.sum())}/11. Joined forecast, price and fieldwork years: {len(data)}. Evaluation years: {predictions.year.nunique() if not predictions.empty else 0}. See coverage_private.csv, annual_private.csv, predictions_private.csv, scores_private.csv and skipped_private.csv.
'''
    FW_OUT.mkdir(parents=True,exist_ok=True)
    coverage.to_csv(FW_OUT/'coverage_private.csv',index=False)
    annual.to_csv(FW_OUT/'annual_private.csv',index=False)
    predictions.to_csv(FW_OUT/'predictions_private.csv',index=False)
    scores.to_csv(FW_OUT/'scores_private.csv',index=False)
    skipped.to_csv(FW_OUT/'skipped_private.csv',index=False)
    (FW_OUT/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    return {'annual':annual,'coverage':coverage,'data':data,'predictions':predictions,'scores':scores,'skipped':skipped,'report':report}
```

```python
from pathlib import Path
import numpy as np
import pandas as pd

def compare_fieldwork_windows(forecast_results, six_week_snapshot):
    global FW_OUT
    previous_out = FW_OUT
    root = Path('/kaggle/working/iowa_fieldwork_six')
    try:
        FW_OUT = root/'six_weeks'
        six = run_fieldwork_research(forecast_results, six_week_snapshot, weeks=6)
        complete = set(six['annual'].loc[six['annual'].fieldwork_complete, 'year'].astype(int))
        one_snapshot = [x for x in six_week_snapshot if x['year'] in complete
                        and _fw_date(x['week_end']) == _fw_expected(x['year'], 1)[0]]
        FW_OUT = root/'one_week_matched'
        one = run_fieldwork_research(forecast_results, one_snapshot, weeks=1)
    finally:
        FW_OUT = previous_out
    left = one['predictions'].copy()
    right = six['predictions'].copy()
    columns = ['year','model','train_years','actual_usd']
    if left.empty or right.empty:
        print('Insufficient matched complete years for a comparison.')
        return {'one':one, 'six':six, 'comparison':pd.DataFrame(), 'scores':pd.DataFrame()}
    joined=left.merge(right,on=columns,how='outer',suffixes=('_one','_six'),indicator=True,validate='one_to_one')
    if not joined['_merge'].eq('both').all():
        raise ValueError('One- and six-week folds are not identical')
    for label in ('price_only','price_forecast'):
        q=joined[joined.model.eq(label)]
        if not np.allclose(q.prediction_usd_one,q.prediction_usd_six,rtol=1e-10,atol=0):
            raise ValueError('Matched baseline predictions disagree')
    combo=joined[joined.model.eq('price_forecast_fieldwork')].copy()
    combo['six_minus_one_abs_error_usd']=combo.absolute_error_usd_six-combo.absolute_error_usd_one
    summaries=[]
    for name, dataset, window in [('price_only',one,'none'),('price_forecast',one,'none'),
        ('price_fieldwork',one,'one'),('price_fieldwork',six,'six'),
        ('price_forecast_fieldwork',one,'one'),('price_forecast_fieldwork',six,'six')]:
        q=dataset['predictions'].query('model == @name')
        summaries.append({'model':name,'fieldwork_window':window,'years':len(q),
                          'mae_musd':q.absolute_error_usd.mean()/1e6})
    scores=pd.DataFrame(summaries)
    root.mkdir(parents=True,exist_ok=True)
    joined.to_csv(root/'matched_predictions_private.csv',index=False)
    scores.to_csv(root/'matched_scores_private.csv',index=False)
    n=len(combo)
    wins=int((combo.six_minus_one_abs_error_usd<0).sum())
    report=f'''# Six-week Iowa fieldwork comparison

The six-week feature averages statewide days suitable for fieldwork over the six consecutive weeks ending on or before May 20. One-week features use the final week of the same six. Reports must be dated by May 31. Some entries are previous-week values from a later dated report and can incorporate revisions available at that time. Some reports are hosted by a university or reproduced by news outlets; the stored source date is that of the document actually used. Report contents and release dates were manually transcribed and not independently reverified by the notebook.

Only years with all six reports and matching forecast and price data enter either model; both windows use identical expanding training folds, training-only scaling, six prior observations minimum and fixed ridge penalty. Joined years: {len(six['data'])}; comparable test years: {n}. The six-week price+forecast+fieldwork model had lower absolute error than the matched one-week model in {wins} of {n} years. Full results are stored alongside source dates, coverage, and missing weeks.

This is a retrospective research comparison after prior years were inspected, with very few annual examples. It does not establish causal effects or prove future improvement. Fieldwork suitability is not farmer orders and covers all farm activities. The frozen 2027 forecast is unchanged.
'''
    (root/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(scores.round(2))
    display((combo[['year','absolute_error_usd_one','absolute_error_usd_six','six_minus_one_abs_error_usd']].set_index('year')/1e6).round(1))
    return {'one':one,'six':six,'comparison':joined,'scores':scores,'report':report}

FIELDWORK_WINDOW_RESULTS = compare_fieldwork_windows(FORECAST_RESULTS, FIELDWORK_SIX_WEEKS)
```

```python
from pathlib import Path
import math
import numpy as np
import pandas as pd

FIELDWORK_2022_OUT = Path('/kaggle/working/iowa_fieldwork_2022')

def investigate_fieldwork_2022(window_results):
    six = window_results['six']
    one = window_results['one']
    data = six['data'].copy().sort_values('year')
    if data.year.duplicated().any() or 2022 not in set(data.year):
        raise ValueError('Expected one complete 2022 annual feature row')
    current = data.loc[data.year.eq(2022)].iloc[0]
    cutoff = str(current.as_of)[:10]
    train = data[(data.year < 2022) & (data.target_filed.astype(str).str[:10] <= cutoff)].copy()
    if len(train) < FW_MIN_TRAIN:
        raise ValueError('2022 lacks the required earlier complete training years')
    features=['forecast_mean_p12_pct', 'fieldwork_mean_days']
    raw=train[features].to_numpy(dtype=float)
    means=raw.mean(axis=0); sd=raw.std(axis=0, ddof=0)
    if not np.isfinite(raw).all() or (sd <= 1e-12).any():
        raise ValueError('Invalid or constant training features')
    z=(raw-means)/sd
    z_test=(current[features].to_numpy(dtype=float)-means)/sd
    if (train[['target_q2_usd','baseline']] <= 0).any().any():
        raise ValueError('Invalid revenue or baseline')
    residual=np.log(train.target_q2_usd.to_numpy(float)/train.baseline.to_numpy(float))
    coefficients=np.linalg.solve(z.T@z+FW_RIDGE*np.eye(2), z.T@residual)
    terms=coefficients*z_test
    recon=float(current.baseline)*math.exp(float(terms.sum()))
    predictions=six['predictions'].copy()
    for label in ('price_only','price_forecast','price_forecast_fieldwork'):
        if predictions.query('model == @label').year.duplicated().any():
            raise ValueError('Duplicate prediction years')
    p=predictions.pivot(index='year',columns='model',values='prediction_usd').reset_index()
    actual=predictions[['year','actual_usd']].drop_duplicates()
    if actual.year.duplicated().any():
        raise ValueError('Conflicting actuals')
    p=p.merge(actual,on='year',validate='one_to_one')
    original=one['predictions'].query("model == 'price_forecast_fieldwork'")[['year','prediction_usd']].rename(columns={'prediction_usd':'one_week'})
    p=p.merge(original,on='year',validate='one_to_one').sort_values('year')
    if not np.isclose(recon,float(p.loc[p.year.eq(2022),'price_forecast_fieldwork'].iloc[0]),rtol=1e-10):
        raise AssertionError('Reconstructed 2022 prediction differs from original')
    if (p[['price_only','price_forecast','price_forecast_fieldwork']] <= 0).any().any():
        raise ValueError('Predictions must be positive for the geometric blend')
    p['half_increment']=np.sqrt(p.price_forecast*p.price_forecast_fieldwork)
    labels={'price_only':'price_only','forecast_only':'price_forecast',
            'one_week':'one_week','full_six_week':'price_forecast_fieldwork','half_increment':'half_increment'}
    for label,col in labels.items():
        p[label+'_absolute_error_usd']=(p[col]-p.actual_usd).abs()
    out=[]
    for omit in [None]+p.year.astype(int).tolist():
        subset=p if omit is None else p[p.year.ne(omit)]
        for label in labels:
            out.append({'omitted_year':'none' if omit is None else str(omit),
                        'model':label,'years':len(subset),
                        'mae_usd':float(subset[label+'_absolute_error_usd'].mean()),
                        'wins_vs_forecast_only':int((subset[label+'_absolute_error_usd']<subset.forecast_only_absolute_error_usd).sum())})
    scores=pd.DataFrame(out)
    corr=float(np.corrcoef(z[:,0],z[:,1])[0,1])
    diag=pd.DataFrame({'feature':features,'train_mean':means,'train_sd':sd,
        'train_min':raw.min(axis=0),'train_max':raw.max(axis=0),
        'value_2022':current[features].to_numpy(dtype=float),'z_2022':z_test,
        'coefficient':coefficients,'log_contribution':terms})
    r=p[p.year.eq(2022)].iloc[0]
    report=f'''# Exploratory 2022 six-week fieldwork error investigation

The price-only prediction was ${r.price_only/1e6:,.1f}m against actual ${r.actual_usd/1e6:,.1f}m. The six-week fieldwork mean was {float(current.fieldwork_mean_days):.2f} days/week, versus training mean {means[1]:.2f} (SD {sd[1]:.2f}, z={z_test[1]:.2f}). The fitted fieldwork coefficient {coefficients[1]:+.4f} contributed {terms[1]:+.4f} to the log adjustment; the rain-forecast contribution was {terms[0]:+.4f}. The combined factor was {math.exp(float(terms.sum())):.4f}, producing ${recon/1e6:,.1f}m, absolute error ${r.full_six_week_absolute_error_usd/1e6:,.1f}m. Earlier train years: {', '.join(map(str,train.year.astype(int)))}. The two training feature correlations were {corr:.2f}; six training observations cannot support strong claims about a stable relation.

The one-week comparison predicted ${r.one_week/1e6:,.1f}m, error ${r.one_week_absolute_error_usd/1e6:,.1f}m. Forecast-only error was ${r.forecast_only_absolute_error_usd/1e6:,.1f}m. Halving the incremental log adjustment is an exploratory sensitivity, not an untouched test: in 2022 its error was ${r.half_increment_absolute_error_usd/1e6:,.1f}m. Annual errors, matched MAE, wins against forecast-only and leave-one-test-year-out sensitivity are retained in the tables. The adjustment was chosen after inspecting history; no new prospective claim follows. The frozen 2027 protocol is unchanged.

This reconstructs how the model made its prediction; it does not establish why company revenue changed or why a farmer bought fertilizer. Fieldwork is a statewide opportunity measure for all activities, not farmer orders. Some source observations were recorded from later previous-week releases or contemporaneous reproductions and may have been revised; release dates and sources are in the earlier coverage table. CF's later Q2 disclosures are contextual only and must not be used in a May 31 prediction: https://www.sec.gov/Archives/edgar/data/1324404/000132440422000022/cf-20220630.htm
'''
    FIELDWORK_2022_OUT.mkdir(parents=True,exist_ok=True)
    p.to_csv(FIELDWORK_2022_OUT/'annual_private.csv',index=False)
    scores.to_csv(FIELDWORK_2022_OUT/'scores_private.csv',index=False)
    diag.to_csv(FIELDWORK_2022_OUT/'diagnostic_private.csv',index=False)
    (FIELDWORK_2022_OUT/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(diag.round(4))
    display(scores[scores.omitted_year.eq('none')].assign(mae_musd=lambda d:d.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display((p.set_index('year')[[f'{n}_absolute_error_usd' for n in labels]]/1e6).round(1))
    return {'diagnostic':diag,'scores':scores,'predictions':p,'report':report}

FIELDWORK_2022_RESULTS = investigate_fieldwork_2022(FIELDWORK_WINDOW_RESULTS)
```

```python
from pathlib import Path
from urllib.parse import urlsplit, parse_qs
import datetime as dt
import json
import math
import numpy as np
import pandas as pd

LH_YEARS = tuple(range(2010, 2016))
LH_STATIONS = ('KAMW', 'KCID', 'KSUX')
LH_OUTPUT = Path('/kaggle/working/iowa_long_history')

def lh_date(value):
    if not isinstance(value, str):
        raise ValueError('Date must be ISO YYYY-MM-DD')
    date=dt.date.fromisoformat(value)
    if date.isoformat()!=value: raise ValueError('Invalid ISO date')
    return date

def lh_expected(year):
    end=dt.date(year,5,20)
    sunday=end-dt.timedelta(days=(end.weekday()+1)%7)
    return [sunday-dt.timedelta(days=7*i) for i in range(5,-1,-1)]

def lh_url(url, group):
    u=urlsplit(url if isinstance(url,str) else '')
    host=u.hostname or ''
    if u.scheme!='https': raise ValueError('Require HTTPS source')
    if group=='price':
        allowed=host=='bls.gov' or host.endswith('.bls.gov')
    elif group=='fieldwork':
        allowed=(host=='usda.gov' or host.endswith('.usda.gov') or host in (
            'iowaagriculture.gov','publications.iowa.gov','news.utcrops.com',
            'k923.fm','whoradio.iheart.com','thegrundyregister.com','www.thegrundyregister.com','northiowatoday.com',
            'downloads.usda.library.cornell.edu','www.farmprogress.com','files01.core.ac.uk'))
    else:
        allowed=host=='mesonet.agron.iastate.edu' and u.path=='/cgi-bin/request/mos.py'
    if not allowed: raise ValueError(f'Unexpected source host for {group}: {host}')
    return u

def lh_forecasts(records):
    expected={(y,s) for y in LH_YEARS for s in LH_STATIONS}; found={}; coverage=[]
    required={'year','station','expected_days','valid_days','duplicate_rows','conflicting_days',
              'invalid_rows','station_mean_p12_pct','source_url','status'}
    for r in records:
        if not isinstance(r,dict) or not required.issubset(r): raise ValueError('Incomplete station summary')
        y,s=r['year'],r['station']
        if (y,s) not in expected or (y,s) in found: raise ValueError('Unexpected or duplicate station-year')
        for k in ('expected_days','valid_days','duplicate_rows','conflicting_days','invalid_rows'):
            if isinstance(r[k],bool) or not isinstance(r[k],int) or r[k]<0: raise ValueError('Invalid coverage count')
        if r['expected_days']!=50 or r['valid_days']+r['conflicting_days']>50: raise ValueError('Invalid 50-day window')
        u=lh_url(r['source_url'],'forecast'); q=parse_qs(u.query)
        wanted={'station':s,'model':'GFS','sts':f'{y}-04-01T00:00Z','ets':f'{y}-05-21T00:00Z','format':'json'}
        if any(q.get(k)!=[v] for k,v in wanted.items()): raise ValueError('Wrong forecast source parameters')
        mean=r['station_mean_p12_pct']
        if mean is not None and (isinstance(mean,bool) or not isinstance(mean,(int,float)) or not np.isfinite(mean) or not 0<=mean<=100): raise ValueError('Invalid forecast probability')
        if (r['valid_days']==0)!=(mean is None): raise ValueError('Forecast mean/count inconsistency')
        usable=(r['status']=='ok' and r['valid_days']>=40 and r['invalid_rows']==0 and mean is not None)
        found[(y,s)]={**r,'usable':usable}
    for y,s in sorted(expected): coverage.append(found.get((y,s),{'year':y,'station':s,'status':'missing','usable':False}))
    annual=[]
    for y in LH_YEARS:
        three=[found.get((y,s)) for s in LH_STATIONS]
        ok=all(x is not None and x['usable'] for x in three)
        annual.append({'year':y,'forecast_complete':ok,'forecast_mean_p12_pct':float(np.mean([x['station_mean_p12_pct'] for x in three])) if ok else np.nan})
    return pd.DataFrame(annual),pd.DataFrame(coverage)

def lh_fieldwork(records):
    found={}; cov=[]
    for r in records:
        needed={'year','week_end','release_date','days_suitable','source_url'}
        if not isinstance(r,dict) or not needed.issubset(r): raise ValueError('Incomplete weekly report')
        y=r['year']; end=lh_date(r['week_end'])
        if y not in LH_YEARS or end not in lh_expected(y) or (y,end) in found: raise ValueError('Unexpected or duplicate week')
        value=r['days_suitable']; pub=lh_date(r['release_date'])
        if not end<pub<=dt.date(y,5,31): raise ValueError('Fieldwork document after cutoff or before week ended')
        if isinstance(value,bool) or not isinstance(value,(int,float)) or not np.isfinite(value) or not 0<=value<=7: raise ValueError('Invalid days suitable')
        lh_url(r['source_url'],'fieldwork'); found[(y,end)]={**r,'usable':True}
    annual=[]
    for y in LH_YEARS:
        expected=lh_expected(y); selected=[found.get((y,d)) for d in expected]
        for d,item in zip(expected,selected): cov.append(item if item is not None else {'year':y,'week_end':d.isoformat(),'usable':False,'status':'missing'})
        ok=all(x is not None for x in selected)
        annual.append({'year':y,'fieldwork_complete':ok,'weeks_present':sum(x is not None for x in selected),
                       'fieldwork_mean_days':float(np.mean([x['days_suitable'] for x in selected])) if ok else np.nan})
    return pd.DataFrame(annual),pd.DataFrame(cov)

def lh_prices(records):
    found={}; required={'year','current_april','previous_april','current_release_date','previous_release_date','current_source_url','previous_source_url'}
    for r in records:
        if not isinstance(r,dict) or not required.issubset(r): raise ValueError('Incomplete BLS price row')
        y=r['year']
        if y not in LH_YEARS or y in found: raise ValueError('Unexpected or duplicate BLS year')
        now=lh_date(r['current_release_date']); prior=lh_date(r['previous_release_date'])
        if not (dt.date(y,4,30)<now<=dt.date(y,5,31) and dt.date(y-1,8,31)<prior<dt.date(y,1,1)):
            raise ValueError('Price source dates unavailable by cutoff')
        lh_url(r['current_source_url'],'price'); lh_url(r['previous_source_url'],'price')
        a,b=r['current_april'],r['previous_april']
        if any(isinstance(v,bool) or not isinstance(v,(int,float)) or not np.isfinite(v) or v<=0 for v in (a,b)): raise ValueError('Invalid BLS index')
        found[y]={**r,'price_ratio':float(a)/float(b),'price_complete':True}
    return pd.DataFrame([found.get(y,{'year':y,'price_complete':False}) for y in LH_YEARS])

def lh_original_revenue(snapshot):
    raw=snapshot.get('revenue') if isinstance(snapshot,dict) else None
    if not isinstance(raw,list): raise ValueError('Missing originally reported revenue snapshot')
    rev=pd.DataFrame(raw)
    if not {'year','q2_usd','q2_filed','q2_accession'}.issubset(rev.columns) or rev.year.duplicated().any(): raise ValueError('Invalid revenue coverage')
    return rev.set_index('year')

def lh_fit(train,row,features):
    baseline=float(row.prior_q2_usd*row.price_ratio)
    if not features: return baseline
    x=train[features].to_numpy(float); mu=x.mean(axis=0); sd=x.std(axis=0,ddof=0)
    if not np.isfinite(x).all() or not np.isfinite(sd).all() or (sd<=1e-12).any(): raise ValueError('No training feature variation')
    z=(x-mu)/sd; test=(np.array([getattr(row,c) for c in features],float)-mu)/sd
    response=np.log(train.target_q2_usd.to_numpy(float)/(train.prior_q2_usd.to_numpy(float)*train.price_ratio.to_numpy(float)))
    beta=np.linalg.solve(z.T@z+np.eye(len(features)),z.T@response)
    exponent=float(test@beta)
    if not np.isfinite(exponent) or abs(exponent)>700: raise ValueError('Invalid prediction')
    return baseline*math.exp(exponent)

def long_history_research(window_results,snapshot,forecasts,fieldwork,prices):
    fa,fc=lh_forecasts(forecasts); wa,wc=lh_fieldwork(fieldwork); pa=lh_prices(prices)
    available=fa.merge(wa,on='year',validate='one_to_one').merge(pa,on='year',validate='one_to_one')
    for col in ('forecast_complete','fieldwork_complete','price_complete'): available[col]=available[col].fillna(False).astype(bool)
    rev=lh_original_revenue(snapshot); rows=[]
    for x in available.itertuples():
        if not (x.forecast_complete and x.fieldwork_complete and x.price_complete): continue
        y=int(x.year)
        if y not in rev.index or y-1 not in rev.index: continue
        actual,prior=rev.loc[y],rev.loc[y-1]
        cutoff=f'{y}-05-31'; filed=str(actual.q2_filed)[:10]; previous_filed=str(prior.q2_filed)[:10]
        if not (f'{y}-06-30'<=filed<=f'{y+1}-05-31' and previous_filed<=cutoff): raise ValueError('Revenue chronology error')
        if float(actual.q2_usd)<=0 or float(prior.q2_usd)<=0 or not str(actual.q2_accession).strip() or not str(prior.q2_accession).strip(): raise ValueError('Invalid revenue')
        rows.append({'year':y,'as_of':cutoff,'target_filed':filed,'target_q2_usd':float(actual.q2_usd),
                     'prior_q2_usd':float(prior.q2_usd),'price_ratio':float(x.price_ratio),
                     'forecast_mean_p12_pct':float(x.forecast_mean_p12_pct),'fieldwork_mean_days':float(x.fieldwork_mean_days)})
    columns=['year','as_of','target_filed','target_q2_usd','prior_q2_usd','price_ratio','forecast_mean_p12_pct','fieldwork_mean_days']
    history=pd.DataFrame(rows,columns=columns)
    original=window_results['six']['data'].copy()
    if original.year.duplicated().any() or set(original.year)&set(LH_YEARS): raise ValueError('Original cohort mismatch')
    augmented=pd.concat([history,original[columns]],ignore_index=True).sort_values('year')
    existing=window_results['six']['predictions']
    features={'price_only':[],'price_forecast':['forecast_mean_p12_pct'],
              'price_forecast_fieldwork':['forecast_mean_p12_pct','fieldwork_mean_days']}
    output=[]
    for y in range(2022,2027):
        matches=original[original.year.eq(y)]
        if len(matches)!=1: raise ValueError(f'Missing original test year {y}')
        row=next(matches.itertuples()); cutoff=str(row.as_of)[:10]
        if cutoff!=f'{y}-05-31' or str(row.target_filed)[:10]<=cutoff: raise ValueError('Test cutoff mismatch')
        for label,df in (('original',original),('extended',augmented)):
            train=df[(df.year<y)&(df.target_filed.astype(str).str[:10]<=cutoff)]
            if len(train)<6: raise ValueError('Less than six eligible training years')
            for model,cols in features.items():
                pred=lh_fit(train,row,cols); actual=float(row.target_q2_usd)
                output.append({'year':y,'training':label,'model':model,'train_years':len(train),
                               'earlier_years':','.join(map(str,train.year.astype(int))),
                               'prediction_usd':pred,'actual_usd':actual,'absolute_error_usd':abs(pred-actual)})
                if label=='original':
                    prior=existing[(existing.year==y)&(existing.model==model)]
                    if len(prior)!=1 or not np.isclose(pred,float(prior.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
                        raise AssertionError('Original-model reproduction differs from existing notebook')
    pred=pd.DataFrame(output)
    score=pred.groupby(['model','training'],as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    paired=pred.pivot(index=['year','model'],columns='training',values='absolute_error_usd').reset_index()
    paired['extended_minus_original_usd']=paired.extended-paired.original
    summary=paired.groupby('model',as_index=False).agg(years=('year','size'),extended_wins=('extended_minus_original_usd',lambda x:int((x<0).sum())))
    cov=available[['year','forecast_complete','fieldwork_complete','price_complete','weeks_present']].copy()
    cov['used']=cov.year.isin(set(history.year))
    LH_OUTPUT.mkdir(parents=True,exist_ok=True)
    for name,frame in [('forecast_coverage_private.csv',fc),('fieldwork_coverage_private.csv',wc),('price_sources_private.csv',pa),('coverage_private.csv',cov),('historical_features_private.csv',history),('predictions_private.csv',pred),('scores_private.csv',score),('paired_private.csv',paired),('wins_private.csv',summary)]: frame.to_csv(LH_OUTPUT/name,index=False)
    p22=pred[(pred.year==2022)&(pred.model=='price_forecast_fieldwork')].set_index('training')
    report=f'''# Older-training-year exploratory test

Earlier years with complete Iowa GFS MOS forecast, six-week fieldwork, originally reported revenue and BLS price sources: {', '.join(map(str,history.year.astype(int))) if len(history) else 'none'}. The original and extended models use the same 2022–2026 prediction rows, same price-only baseline, training-only standardization, penalty 1 and no intercept. The original model is required to reproduce the existing saved predictions exactly. In 2022, original six-week prediction error was ${float(p22.loc['original','absolute_error_usd'])/1e6:,.1f}m, extended error ${float(p22.loc['extended','absolute_error_usd'])/1e6:,.1f}m. Full matched scores, per-year results, coverage and source metadata are in this folder.

Older nitrogen price ratios are reconstructed from original BLS current-April initial releases and the prior April's four-month-revised value published the previous September; they are not ALFRED vintages. Only earlier target results filed by each simulated cutoff are used. Forecast aggregates are manually transcribed archive summaries and may have been revised; exact historic product-publication timing has not been independently verified. Fieldwork records may use a later report's previous-week value or attributed reproductions; record their individual dates and possible revisions. The notebook validates the metadata and arithmetic, not the accuracy of the transcription or external source contents.

The historical tests through 2026 were already inspected; this is exploratory, not a new untouched holdout. More training years do not establish causality or guarantee better future performance. No frozen 2027 prediction or protocol was altered.
'''
    (LH_OUTPUT/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(cov)
    display(score.assign(mae_musd=lambda x:x.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display(summary)
    display((paired.query("model == 'price_forecast_fieldwork'").set_index('year')[['original','extended']]/1e6).round(1))
    return {'history':history,'coverage':cov,'predictions':pred,'scores':score,'paired':paired,'summary':summary,'report':report}

LONG_HISTORY_RESULTS=long_history_research(FIELDWORK_WINDOW_RESULTS,SNAPSHOT_JSON,HISTORICAL_FORECASTS,HISTORICAL_FIELDWORK,HISTORICAL_PRICES)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd


def missing_fieldwork_research(reported, archived, output_subdir='iowa_missing_fieldwork'):
    old_history = LONG_HISTORY_RESULTS['history'].copy()
    old_predictions = LONG_HISTORY_RESULTS['predictions']
    combined = HISTORICAL_FIELDWORK + reported
    fa, _ = lh_forecasts(HISTORICAL_FORECASTS)
    wa, wc = lh_fieldwork(combined)
    expected_url = 'https://mesonet.agron.iastate.edu/plotting/auto/plot/127/state:IA::short_desc:FD::w:last::threshold:50::cmap:tab20b::_r:43::dpi:100.csv'
    seen = set()
    for r in archived:
        y=int(r['year']); end=lh_date(r['week_end']); prev=lh_date(r['previous_week_end'])
        if end not in lh_expected(y) or (end-prev).days != 7 or (y,end) in seen:
            raise ValueError('Invalid or duplicate archive week')
        seen.add((y,end))
        if r['source_url'] != expected_url or lh_date(r['retrieved_date']) <= dt.date(y,5,31):
            raise ValueError('Archive source or retrieval date invalid')
        before,after=float(r['cumulative_previous']),float(r['cumulative_current'])
        value=after-before
        if not np.isfinite(value) or not 0 <= value <= 7: raise ValueError('Invalid difference')
        mask=(wc.year==y)&(wc.week_end==end.isoformat())
        if mask.sum()!=1 or bool(wc.loc[mask,'usable'].iloc[0]): raise ValueError('Archive week conflicts with an existing observation')
        wc.loc[mask,'days_suitable']=round(value,6)
        wc.loc[mask,'source_url']=r['source_url']
        wc.loc[mask,'release_date']=None
        wc.loc[mask,'basis']='later_archive_difference_not_point_in_time'
        wc.loc[mask,'retrieved_date']=r['retrieved_date']
        wc.loc[mask,'usable']=True
    for y in LH_YEARS:
        selection=wc[(wc.year==y)&(wc.usable==True)]
        complete=len(selection)==6
        wa.loc[wa.year==y,'weeks_present']=len(selection)
        wa.loc[wa.year==y,'fieldwork_complete']=complete
        wa.loc[wa.year==y,'fieldwork_mean_days']=float(selection.days_suitable.mean()) if complete else np.nan
    pa = lh_prices(HISTORICAL_PRICES)
    avail = fa.merge(wa, on='year', validate='one_to_one').merge(pa, on='year', validate='one_to_one')
    rev = lh_original_revenue(SNAPSHOT_JSON)
    rows = []
    previously = set(old_history.year.astype(int))
    for x in avail.itertuples():
        y = int(x.year)
        if y in previously or not (x.forecast_complete and x.fieldwork_complete and x.price_complete):
            continue
        if y not in rev.index or y-1 not in rev.index: continue
        now, prior = rev.loc[y], rev.loc[y-1]
        cutoff = f'{y}-05-31'; filed = str(now.q2_filed)[:10]; prior_filed = str(prior.q2_filed)[:10]
        if not (f'{y}-06-30' <= filed <= f'{y+1}-05-31' and prior_filed <= cutoff):
            raise ValueError('Revenue chronology invalid')
        if float(now.q2_usd) <= 0 or float(prior.q2_usd) <= 0 or not str(now.q2_accession).strip() or not str(prior.q2_accession).strip():
            raise ValueError('Invalid revenue')
        rows.append({'year':y, 'as_of':cutoff, 'target_filed':filed,
                     'target_q2_usd':float(now.q2_usd), 'prior_q2_usd':float(prior.q2_usd),
                     'price_ratio':float(x.price_ratio), 'forecast_mean_p12_pct':float(x.forecast_mean_p12_pct),
                     'fieldwork_mean_days':float(x.fieldwork_mean_days)})
    extra = pd.DataFrame(rows, columns=old_history.columns)
    orig = FIELDWORK_WINDOW_RESULTS['six']['data'].copy()
    cols = list(old_history.columns)
    old = pd.concat([old_history, orig[cols]], ignore_index=True).sort_values('year')
    new = pd.concat([old, extra], ignore_index=True).sort_values('year')
    if old.year.duplicated().any() or new.year.duplicated().any(): raise ValueError('Duplicate training years')
    features = {'price_forecast': ['forecast_mean_p12_pct'],
                'price_forecast_fieldwork': ['forecast_mean_p12_pct', 'fieldwork_mean_days']}
    predictions=[]
    for y in range(2022, 2027):
        test = orig[orig.year.eq(y)]
        if len(test) != 1: raise ValueError('Missing evaluation year')
        row = next(test.itertuples()); cutoff=str(row.as_of)[:10]
        if cutoff != f'{y}-05-31' or str(row.target_filed)[:10] <= cutoff: raise ValueError('Test cutoff invalid')
        for label, data in [('prior',old),('new',new)]:
            train = data[(data.year < y) & (data.target_filed.astype(str).str[:10] <= cutoff)]
            for model, fields in features.items():
                pred=lh_fit(train,row,fields); actual=float(row.target_q2_usd)
                if label=='prior':
                    oldp=old_predictions[(old_predictions.year==y)&(old_predictions.model==model)&(old_predictions.training=='extended')]
                    if len(oldp)!=1 or not np.isclose(pred,float(oldp.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
                        raise AssertionError('Prior long-history result could not be reproduced')
                predictions.append({'year':y,'model':model,'training':label,'train_years':len(train),
                                    'earlier_years':','.join(map(str,train.year.astype(int))),
                                    'actual_usd':actual,'prediction_usd':pred,'absolute_error_usd':abs(pred-actual)})
    pred=pd.DataFrame(predictions)
    scores=pred.groupby(['model','training'],as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    paired=pred.pivot(index=['model','year'],columns='training',values='absolute_error_usd').reset_index()
    paired['new_minus_prior_usd']=paired['new']-paired['prior']
    if output_subdir not in ('iowa_missing_fieldwork','iowa_missing_fieldwork_sensitivity'): raise ValueError('Invalid output directory')
    output=Path('/kaggle/working') / output_subdir; output.mkdir(parents=True,exist_ok=True)
    for name,df in [('coverage_private.csv',wc),('additional_years_private.csv',extra),('predictions_private.csv',pred),('scores_private.csv',scores),('paired_private.csv',paired)]: df.to_csv(output/name,index=False)
    report = f'''# Additional Iowa fieldwork records: exploratory comparison

New complete earlier years: {', '.join(map(str, extra.year.astype(int))) if len(extra) else 'none'}. Same 2022–2026 evaluation years, training-only standardization, features and model as previous long-history comparison; prior predictions must be reproduced exactly. Only previous targets filed by the cutoff are used. Some missing fieldwork values are reconstructed by differencing the current IEM Quick Stats cumulative series. They may have been revised and cannot be claimed to represent data known at the original cutoff. No report date is invented for them. The 2013 May 5 report narrative and USDA state stories say 2.3 days; an Iowa report table says 2.0, so a sensitivity analysis is necessary. Weekly source links, retrieval/release dates, and caveats appear in the notebook and coverage table. Forecast archive availability and revisions, BLS price reconstructions, and previously documented source limitations continue to apply.

All test years had already been examined. This is exploratory, not a new independent test or proof of causality; the prospective 2027 protocol remains unchanged. Company-wide revenue is not Iowa-specific.
'''
    (output/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(avail[['year','weeks_present','fieldwork_complete']])
    display(scores.assign(mae_musd=lambda d:d.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display((paired.query("model == 'price_forecast_fieldwork'").set_index('year')[['prior','new']]/1e6).round(1))
    return {'additional_years':extra, 'coverage':wc, 'predictions':pred, 'scores':scores, 'paired':paired, 'report':report}

MISSING_FIELDWORK_RESULTS = missing_fieldwork_research(NEW_FIELDWORK_REPORTED, NEW_FIELDWORK_ARCHIVE)

_alternate_reported = [dict(r) for r in NEW_FIELDWORK_REPORTED]
for _r in _alternate_reported:
    if _r['year']==2013 and _r['week_end']=='2013-05-05':
        _r.update(days_suitable=2.0, release_date='2013-05-06',
                  source_url='https://publications.iowa.gov/14585/1/Vol08__05_06_13.pdf',
                  basis='conflicting_original_report_table_sensitivity')
MISSING_FIELDWORK_SENSITIVITY = missing_fieldwork_research(_alternate_reported, NEW_FIELDWORK_ARCHIVE, 'iowa_missing_fieldwork_sensitivity')
```

```python
from itertools import permutations
from pathlib import Path
import math
import numpy as np
import pandas as pd


def diagnose_extra_training_years():
    cols=list(LONG_HISTORY_RESULTS['history'].columns)
    old_years=LONG_HISTORY_RESULTS['history'].copy()
    extra=MISSING_FIELDWORK_RESULTS['additional_years'].copy()
    orig=FIELDWORK_WINDOW_RESULTS['six']['data'].copy()
    assert set(extra.year.astype(int)) == {2010,2011,2013}
    old=pd.concat([old_years,orig[cols]],ignore_index=True).sort_values('year')
    if old.year.duplicated().any(): raise ValueError('Duplicate year')
    all_data=pd.concat([old,extra[cols]],ignore_index=True).sort_values('year')
    if all_data.year.duplicated().any(): raise ValueError('Duplicate year')
    fields=['forecast_mean_p12_pct','fieldwork_mean_days']
    predictions=MISSING_FIELDWORK_RESULTS['predictions']
    effects=[]; feature_rows=[]; year_rows=[]; training_rows=[]

    def evaluate(train,row):
        x=train[fields].to_numpy(dtype=float)
        mu=x.mean(axis=0); sd=x.std(axis=0,ddof=0)
        if not np.isfinite(x).all() or not np.isfinite(sd).all() or (sd<=1e-12).any(): raise ValueError('Invalid training features')
        z=(x-mu)/sd
        target=np.log(train.target_q2_usd.to_numpy(float)/(train.prior_q2_usd.to_numpy(float)*train.price_ratio.to_numpy(float)))
        beta=np.linalg.solve(z.T@z+np.eye(2),z.T@target)
        point=np.array([getattr(row,c) for c in fields],dtype=float)
        test_z=(point-mu)/sd; terms=beta*test_z
        baseline=float(row.prior_q2_usd*row.price_ratio)
        prediction=baseline*math.exp(float(terms.sum()))
        if not np.isclose(prediction,lh_fit(train,row,fields),rtol=1e-12,atol=1e-3): raise AssertionError('Reconstruction mismatch')
        return dict(mu=mu,sd=sd,test_z=test_z,beta=beta,terms=terms,baseline=baseline,prediction=prediction,
                    correlation=float(np.corrcoef(x.T)[0,1]),log_target_mean=float(np.mean(target)))

    for year in (2023,2026):
        row=next(orig[orig.year.eq(year)].itertuples()); cutoff=str(row.as_of)[:10]
        def allowed(frame): return frame[(frame.year<year)&(frame.target_filed.astype(str).str[:10]<=cutoff)].copy()
        base=allowed(old); added=allowed(extra); full=allowed(all_data)
        if set(added.year.astype(int)) != {2010,2011,2013}: raise ValueError('Unexpected additions')
        old_fit=evaluate(base,row); new_fit=evaluate(full,row)
        for label,fit in [('prior',old_fit),('new',new_fit)]:
            ref=predictions[(predictions.year==year)&(predictions.model=='price_forecast_fieldwork')&(predictions.training==label)]
            if len(ref)!=1 or not np.isclose(fit['prediction'],float(ref.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3): raise AssertionError('Prior results differ')
            actual=float(row.target_q2_usd)
            effects.append(dict(year=year,training=label,training_years=len(base) if label=='prior' else len(full),
                actual_usd=actual,baseline_usd=fit['baseline'],prediction_usd=fit['prediction'],
                signed_error_usd=fit['prediction']-actual,absolute_error_usd=abs(fit['prediction']-actual),
                adjustment_pct=(fit['prediction']/fit['baseline']-1)*100,forecast_fieldwork_train_correlation=fit['correlation']))
            for j,feature in enumerate(fields):
                feature_rows.append(dict(year=year,training=label,feature=feature,value=float(getattr(row,feature)),
                    training_mean=fit['mu'][j],training_std=fit['sd'][j],test_z=fit['test_z'][j],
                    ridge_coefficient=fit['beta'][j],log_contribution=fit['terms'][j]))
        total={y:0. for y in (2010,2011,2013)}
        for order in permutations(total):
            selected=[]; before=old_fit['prediction']
            for addition in order:
                selected.append(addition)
                train=pd.concat([base,added[added.year.isin(selected)]],ignore_index=True)
                after=evaluate(train,row)['prediction']
                total[addition]+=(after-before)/6.
                before=after
            if not np.isclose(before,new_fit['prediction'],rtol=1e-10,atol=1e-3): raise AssertionError('Order endpoint mismatch')
        if not np.isclose(sum(total.values()),new_fit['prediction']-old_fit['prediction'],rtol=1e-10,atol=1e-3): raise AssertionError('Contributions do not sum')
        for y,value in total.items(): year_rows.append(dict(year=year,added_year=y,average_prediction_change_usd=value))
        for r in added.itertuples():
            training_rows.append(dict(prediction_year=year,training_year=int(r.year),
                forecast_mean_p12_pct=r.forecast_mean_p12_pct,fieldwork_mean_days=r.fieldwork_mean_days,
                log_revenue_over_price_baseline=math.log(r.target_q2_usd/(r.prior_q2_usd*r.price_ratio))))
    effects=pd.DataFrame(effects); feats=pd.DataFrame(feature_rows); years=pd.DataFrame(year_rows); added_profiles=pd.DataFrame(training_rows)
    out=Path('/kaggle/working/iowa_older_year_diagnostic'); out.mkdir(parents=True,exist_ok=True)
    for name,df in [('predictions_private.csv',effects),('feature_terms_private.csv',feats),('year_contributions_private.csv',years),('added_year_profiles_private.csv',added_profiles)]: df.to_csv(out/name,index=False)
    report='''# Why adding the older years changed predictions

This decomposition exactly reconstructs the historical ridge model. Adding old rows changes the fitted coefficients and training means and standard deviations; the same fixed test-year values interact with these refitted quantities. Feature terms are on a log scale, and their change is a mechanical decomposition rather than a causal statement. The contribution of each added year is the average prediction change over all six orders of adding the three years; it is allocation-dependent and not an independent causal effect. Training labels are filtered to those filed by each May 31 cutoff, but several older fieldwork observations come from a current archive with unknown historical revisions. All outcomes through 2026 have been inspected and these analyses are exploratory. CF net sales cover more than Iowa corn. Frozen 2027 protocol unchanged.
'''
    (out/'REPORT.md').write_text(report,encoding='utf-8')
    print(report)
    display(effects.assign(actual_m=lambda x:x.actual_usd/1e6,baseline_m=lambda x:x.baseline_usd/1e6,prediction_m=lambda x:x.prediction_usd/1e6,signed_error_m=lambda x:x.signed_error_usd/1e6,absolute_error_m=lambda x:x.absolute_error_usd/1e6).drop(columns=['actual_usd','baseline_usd','prediction_usd','signed_error_usd','absolute_error_usd']).round(3))
    display(feats.round(4))
    display(years.assign(average_change_m=lambda x:x.average_prediction_change_usd/1e6).drop(columns='average_prediction_change_usd').round(2))
    display(added_profiles.drop(columns='prediction_year').drop_duplicates().round(4))
    return {'predictions':effects,'features':feats,'year_contributions':years,'added_profiles':added_profiles,'report':report}

OLDER_YEAR_DIAGNOSTIC=diagnose_extra_training_years()
from pathlib import Path
import numpy as np
import pandas as pd


def recency_weighted_fit(train, row, features, old_weight=0.5):
    weights=np.where(train.year.isin([2010, 2011, 2013]),old_weight,1.0)
    weights=weights/weights.mean()
    x=train[features].to_numpy(float)
    mu=np.average(x,axis=0,weights=weights)
    sd=np.sqrt(np.average((x-mu)**2,axis=0,weights=weights))
    if not np.isfinite(x).all() or not np.isfinite(sd).all() or (sd<=1e-12).any():
        raise ValueError('Invalid training features')
    z=(x-mu)/sd
    log_resid=np.log(train.target_q2_usd.to_numpy(float)/(train.prior_q2_usd.to_numpy(float)*train.price_ratio.to_numpy(float)))
    beta=np.linalg.solve(z.T@(weights[:,None]*z)+np.eye(len(features)),z.T@(weights*log_resid))
    test=(np.asarray([getattr(row,c) for c in features],float)-mu)/sd
    return float(row.prior_q2_usd*row.price_ratio)*np.exp(float(test@beta))


def recency_sensitivity(recovered, tag):
    if tag not in ('main','2013_table'): raise ValueError('Invalid tag')
    orig=FIELDWORK_WINDOW_RESULTS['six']['data'].copy()
    old_history=LONG_HISTORY_RESULTS['history'].copy()
    added=recovered['additional_years'].copy()
    columns=list(old_history.columns)
    old=pd.concat([old_history,orig[columns]],ignore_index=True).sort_values('year')
    all_years=pd.concat([old,added],ignore_index=True).sort_values('year')
    if old.year.duplicated().any() or all_years.year.duplicated().any(): raise ValueError('Duplicate years')
    if set(added.year)!={2010,2011,2013}: raise ValueError('Missing expected years')
    features=['forecast_mean_p12_pct','fieldwork_mean_days']
    saved=recovered['predictions']
    records=[]
    for y in range(2022,2027):
        row=next(orig[orig.year.eq(y)].itertuples()); cutoff=str(row.as_of)[:10]
        if cutoff != f'{y}-05-31' or str(row.target_filed)[:10]<=cutoff: raise ValueError('Cutoff invalid')
        oldtrain=old[(old.year<y)&(old.target_filed.astype(str).str[:10]<=cutoff)]
        train=all_years[(all_years.year<y)&(all_years.target_filed.astype(str).str[:10]<=cutoff)]
        oldpred=lh_fit(oldtrain,row,features)
        full=lh_fit(train,row,features)
        price=lh_fit(train,row,[])
        forecast=lh_fit(train,row,['forecast_mean_p12_pct'])
        oldforecast=lh_fit(oldtrain,row,['forecast_mean_p12_pct'])
        halfweight=recency_weighted_fit(train,row,features)
        halfadjust=np.sqrt(full*price)
        fullcheck=recency_weighted_fit(train,row,features,old_weight=1.0)
        if not np.isclose(full,fullcheck,rtol=1e-12,atol=1e-3): raise AssertionError('Equal weights fail to reproduce original fit')
        for label,estimate in [('prior_long',oldpred),('all_full',full),('all_half_weight_added',halfweight),
                               ('all_half_adjust',halfadjust),('all_forecast_only',forecast),('prior_forecast_only',oldforecast),('price_only',price)]:
            if label in ('prior_long','all_full'):
                previous=saved[(saved.year==y)&(saved.model=='price_forecast_fieldwork')&(saved.training==('prior' if label=='prior_long' else 'new'))]
                if len(previous)!=1 or not np.isclose(estimate,float(previous.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
                    raise AssertionError('Saved prediction mismatch')
            actual=float(row.target_q2_usd)
            records.append({'year':y,'candidate':label,'train_years':len(oldtrain) if label in ('prior_long','prior_forecast_only') else len(train),
                            'actual_usd':actual,'prediction_usd':estimate,'signed_error_usd':estimate-actual,
                            'absolute_error_usd':abs(estimate-actual)})
    pred=pd.DataFrame(records)
    scores=pred.groupby('candidate',as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    base=pred[pred.candidate.eq('all_full')].set_index('year').absolute_error_usd
    scores['wins_vs_all_full']=[int((g.set_index('year').absolute_error_usd < base).sum()) for _,g in pred.groupby('candidate')]
    output=Path('/kaggle/working/iowa_recency_sensitivity')/tag
    output.mkdir(parents=True,exist_ok=True)
    pred.to_csv(output/'predictions_private.csv',index=False)
    scores.to_csv(output/'scores_private.csv',index=False)
    report='''The two fixed variants half-weight the three reconstructed years or halve the complete fitted log correction. All years use earlier filed labels and identical test rows. Scaling and coefficients are fit only on the training fold; weights are normalized to mean one before applying the original ridge penalty. Historical sources may be revised. Both choices are post-hoc exploratory sensitivities based on inspected years, not evidence of future improvement. The 2027 protocol remains unchanged.'''
    (output/'REPORT.md').write_text(report,encoding='utf8')
    print('2013 source:',tag,report)
    display(scores.assign(mae_musd=lambda d:d.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display((pred.pivot(index='year',columns='candidate',values='absolute_error_usd')/1e6).round(1))
    display((pred[pred.year.isin([2023,2026])].pivot(index='year',columns='candidate',values='signed_error_usd')/1e6).round(1))
    return {'predictions':pred,'scores':scores}

RECENCY_SENSITIVITY=recency_sensitivity(MISSING_FIELDWORK_RESULTS,'main')
RECENCY_SENSITIVITY_2013_TABLE=recency_sensitivity(MISSING_FIELDWORK_SENSITIVITY,'2013_table')
```

```python
PHOSPHATE_SEGMENT=[
 dict(year=2009,phosphate_q2_usd=236e6,filed_date='2009-08-03',source_url='https://www.sec.gov/Archives/edgar/data/1324404/000104746909007107/a2193837z10-q.htm',basis='original_q2_filing'),
 dict(year=2010,phosphate_q2_usd=185.1e6,filed_date='2010-08-06',source_url='https://www.sec.gov/Archives/edgar/data/1324404/000104746910007131/a2199646z10-q.htm',basis='original_q2_filing'),
 dict(year=2011,phosphate_q2_usd=296.6e6,filed_date='2011-08-05',source_url='https://www.sec.gov/Archives/edgar/data/1324404/000104746911007023/R53.htm',basis='original_q2_filing'),
 dict(year=2012,phosphate_q2_usd=231.5e6,filed_date='2012-08-07',source_url='https://www.sec.gov/Archives/edgar/data/1324404/000104746912007878/a2210380z10-q.htm',basis='original_q2_filing'),
 dict(year=2013,phosphate_q2_usd=189.7e6,filed_date='2013-08-07',source_url='https://www.sec.gov/Archives/edgar/data/1324404/000104746913008170/a2216170z10-q.htm',basis='original_q2_filing'),
 dict(year=2014,phosphate_q2_usd=23.3e6,filed_date='2014-08-07',source_url='https://www.sec.gov/Archives/edgar/data/1324404/000104746914006740/a2220959z10-q.htm',basis='original_q2_filing'),
 dict(year=2015,phosphate_q2_usd=0,filed_date='2015-08-05',source_url='https://www.sec.gov/Archives/edgar/data/1324404/000132440415000004/cf-08052015_exhibit991xear.htm',basis='earnings_release_zero_post_disposal')]

from pathlib import Path
from urllib.parse import urlsplit
import numpy as np
import pandas as pd


def phosphate_sensitivity(segments, result, output_subdir):
    if output_subdir not in ('iowa_phosphate_sensitivity','iowa_phosphate_alt_fieldwork'):
        raise ValueError('Invalid output directory')
    if not isinstance(segments, list): raise ValueError('Expected segment records')
    records = {}
    for r in segments:
        needed={'year','phosphate_q2_usd','filed_date','source_url','basis'}
        if not isinstance(r,dict) or not needed.issubset(r): raise ValueError('Incomplete segment provenance')
        y=r['year']; value=r['phosphate_q2_usd']; date=lh_date(r['filed_date']); url=urlsplit(r['source_url'])
        if isinstance(y,bool) or not isinstance(y,int) or y not in range(2009,2016) or y in records:
            raise ValueError('Unexpected or duplicate segment year')
        if isinstance(value,bool) or not isinstance(value,(int,float)) or not np.isfinite(value) or value<0:
            raise ValueError('Invalid phosphate sales')
        if not dt.date(y,6,30) < date <= dt.date(y+1,5,31): raise ValueError('Invalid filing date')
        if url.scheme!='https' or url.hostname!='www.sec.gov' or not url.path.startswith('/Archives/edgar/data/1324404/'):
            raise ValueError('Expected CF SEC filing')
        if not isinstance(r['basis'],str) or not r['basis'].strip(): raise ValueError('Missing segment basis')
        records[y]=dict(r)
    if set(records) != set(range(2009,2016)): raise ValueError('Missing segment records; cannot safely subtract')
    if records[2015]['phosphate_q2_usd'] != 0: raise ValueError('Expected no disposed phosphate segment in 2015')
    history=LONG_HISTORY_RESULTS['history'].copy()
    additional=result['additional_years'].copy()
    original=FIELDWORK_WINDOW_RESULTS['six']['data'].copy()
    cols=list(history.columns)
    full=pd.concat([history,additional,original[cols]],ignore_index=True).sort_values('year')
    if full.year.duplicated().any() or not set(range(2010,2016)).issubset(set(full.year)):
        raise ValueError('Expected complete, unique historical cohort')
    adjusted=full.copy()
    changes=[]
    for y in range(2010,2016):
        i=adjusted.index[adjusted.year.eq(y)]
        if len(i)!=1: raise ValueError('Missing historical year')
        i=i[0]; prior, current=records[y-1],records[y]
        row=adjusted.loc[i]; cutoff=lh_date(str(row.as_of)[:10]); target_filed=lh_date(str(row.target_filed)[:10])
        if lh_date(prior['filed_date'])>cutoff: raise ValueError('Prior segment not available at historical cutoff')
        if lh_date(current['filed_date'])>target_filed: raise ValueError('Segment source later than recorded target filing')
        new_target=float(row.target_q2_usd)-float(current['phosphate_q2_usd'])
        new_prior=float(row.prior_q2_usd)-float(prior['phosphate_q2_usd'])
        if min(new_target,new_prior)<=0: raise ValueError('Segment exceeds total sales')
        adjusted.loc[i,['target_q2_usd','prior_q2_usd']]=[new_target,new_prior]
        changes.append({'year':y,'target_segment_filed':current['filed_date'],'prior_segment_filed':prior['filed_date'],
                        'target_phosphate_usd':current['phosphate_q2_usd'],'prior_phosphate_usd':prior['phosphate_q2_usd'],
                        'target_source':current['source_url'],'prior_source':prior['source_url']})
    features=['forecast_mean_p12_pct','fieldwork_mean_days']
    predictions=[]
    for year in range(2022,2027):
        test=original[original.year.eq(year)]
        if len(test)!=1: raise ValueError('Missing test year')
        row=next(test.itertuples()); cutoff=str(row.as_of)[:10]
        if cutoff!=f'{year}-05-31' or str(row.target_filed)[:10]<=cutoff: raise ValueError('Test cutoff invalid')
        for label, data in [('unadjusted',full),('segment_adjusted',adjusted)]:
            train=data[(data.year<year)&(data.target_filed.astype(str).str[:10]<=cutoff)]
            if any(lh_date(records[int(y)]['filed_date'])>lh_date(cutoff) for y in train.year if int(y) in range(2010,2016)):
                raise ValueError('Segment target not yet known at evaluation')
            prediction=lh_fit(train,row,features)
            if label=='unadjusted':
                reference=result['predictions']
                ref=reference[(reference.year==year)&(reference.model=='price_forecast_fieldwork')&(reference.training=='new')]
                if len(ref)!=1 or not np.isclose(prediction,float(ref.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
                    raise AssertionError('Original full-history prediction differs from previous notebook result')
            predictions.append({'year':year,'variant':label,'training_years':len(train),
                                'actual_usd':float(row.target_q2_usd),'prediction_usd':prediction,
                                'absolute_error_usd':abs(prediction-float(row.target_q2_usd))})
    pred=pd.DataFrame(predictions)
    scores=pred.groupby('variant',as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    paired=pred.pivot(index='year',columns='variant',values='absolute_error_usd')
    report='''# Phosphate-segment comparability: post-hoc sensitivity

The added older years precede the 2014 sale of CF's phosphate business. For 2010–2015 training rows only, compare originally reported company-wide Q2 sales with historical total sales minus the separately reported phosphate segment, subtracting phosphate from both the target and prior-year base. Retain the same price/weather features, test cutoffs, full company-wide test targets, training-only standardization and ridge penalty. Reproduce the unadjusted full-history predictions exactly before reporting a comparison.

The old segment data are sourced from dated CF filings and checked for availability. Some 2010–2013 fieldwork records are reconstructed from a current, potentially revised IEM series. The prior forecast-vintage and price-reconstruction caveats also apply. Nitrogen-only historical sales are an approximation to present-day company-wide business, not a restatement of the test target. A phosphate exclusion was proposed after looking at past results, so this is exploratory. 2022–2026 are not an untouched test. It cannot demonstrate that phosphate caused errors or establish a predictive improvement. The frozen 2027 protocol is unchanged.
'''
    out=Path('/kaggle/working')/output_subdir; out.mkdir(parents=True,exist_ok=True)
    for name,frame in [('segment_provenance_private.csv',pd.DataFrame(records.values())),('training_adjustments_private.csv',pd.DataFrame(changes)),('predictions_private.csv',pred),('scores_private.csv',scores),('paired_private.csv',paired.reset_index())]:
        frame.to_csv(out/name,index=False)
    (out/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(scores.assign(mae_musd=lambda d:d.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display((paired/1e6).round(1))
    return {'scores':scores,'predictions':pred,'paired':paired,'report':report}

PHOSPHATE_RESULTS=phosphate_sensitivity(PHOSPHATE_SEGMENT,MISSING_FIELDWORK_RESULTS,'iowa_phosphate_sensitivity')
PHOSPHATE_ALT_FIELDWORK_RESULTS=phosphate_sensitivity(PHOSPHATE_SEGMENT,MISSING_FIELDWORK_SENSITIVITY,'iowa_phosphate_alt_fieldwork')
```

```python
from pathlib import Path
from urllib.parse import urlsplit
import datetime as dt
import numpy as np
import pandas as pd


def forecast_phosphate_comparison():
    features=['forecast_mean_p12_pct']
    oldhistory=LONG_HISTORY_RESULTS['history'].copy()
    add=MISSING_FIELDWORK_RESULTS['additional_years'].copy()
    alternate=MISSING_FIELDWORK_SENSITIVITY['additional_years'].copy()
    orig=FIELDWORK_WINDOW_RESULTS['six']['data'].copy()
    cols=list(oldhistory.columns)
    if set(add.year.astype(int)) != {2010,2011,2013}: raise ValueError('Missing added years')
    independent=[c for c in cols if c!='fieldwork_mean_days']
    pd.testing.assert_frame_equal(add[independent].sort_values('year').reset_index(drop=True),
                                  alternate[independent].sort_values('year').reset_index(drop=True))
    prev=pd.concat([oldhistory,orig[cols]],ignore_index=True).sort_values('year')
    full=pd.concat([prev,add[cols]],ignore_index=True).sort_values('year')
    if prev.year.duplicated().any() or full.year.duplicated().any(): raise ValueError('Duplicate cohort year')
    if not set(range(2010,2016)).issubset(set(full.year.astype(int))): raise ValueError('Incomplete older cohort')

    records={}
    for r in PHOSPHATE_SEGMENT:
        if not isinstance(r,dict) or not {'year','phosphate_q2_usd','filed_date','source_url','basis'}<=set(r):
            raise ValueError('Missing segment provenance')
        y=r['year']; value=r['phosphate_q2_usd']; url=urlsplit(r['source_url']); filed=lh_date(r['filed_date'])
        if isinstance(y,bool) or not isinstance(y,int) or y not in range(2009,2016) or y in records: raise ValueError('Invalid segment year')
        if isinstance(value,bool) or not isinstance(value,(int,float)) or not np.isfinite(value) or value<0: raise ValueError('Invalid phosphate amount')
        if not dt.date(y,6,30)<filed<=dt.date(y+1,5,31): raise ValueError('Invalid segment filing date')
        if url.scheme!='https' or url.hostname!='www.sec.gov' or not url.path.startswith('/Archives/edgar/data/1324404/') or not str(r['basis']).strip(): raise ValueError('Invalid segment source')
        records[y]=dict(r)
    if set(records)!=set(range(2009,2016)) or records[2015]['phosphate_q2_usd']!=0: raise ValueError('Incomplete segment series')

    def adjust(data):
        adjusted=data.copy()
        for idx,row in adjusted[adjusted.year.between(2010,2015)].iterrows():
            y=int(row.year); current=records[y]; prior=records[y-1]
            if lh_date(prior['filed_date'])>lh_date(str(row.as_of)[:10]): raise ValueError('Prior segment unavailable at row cutoff')
            if lh_date(current['filed_date'])>lh_date(str(row.target_filed)[:10]): raise ValueError('Target segment later than revenue filing')
            target=float(row.target_q2_usd)-current['phosphate_q2_usd']
            base=float(row.prior_q2_usd)-prior['phosphate_q2_usd']
            if min(target,base)<=0: raise ValueError('Segment exceeds sales')
            adjusted.loc[idx,['target_q2_usd','prior_q2_usd']]=[target,base]
        return adjusted
    prev_adjusted=adjust(prev); full_adjusted=adjust(full)
    variants={'prior_unadjusted':prev,'prior_adjusted':prev_adjusted,
              'full_unadjusted':full,'full_adjusted':full_adjusted}
    output=[]
    for y in range(2022,2027):
        match=orig[orig.year.eq(y)]
        if len(match)!=1: raise ValueError('Missing test row')
        row=next(match.itertuples()); cutoff=str(row.as_of)[:10]
        if cutoff!=f'{y}-05-31' or str(row.target_filed)[:10]<=cutoff: raise ValueError('Bad test cutoff')
        for label,data in variants.items():
            train=data[(data.year<y)&(data.target_filed.astype(str).str[:10]<=cutoff)].copy()
            if len(train)<6: raise ValueError('Too few training rows')
            if 'adjusted' in label and not label.endswith('unadjusted'):
                for trainyear in train.year.astype(int):
                    if trainyear in range(2010,2016) and lh_date(records[trainyear]['filed_date'])>lh_date(cutoff): raise ValueError('Segment target unavailable')
            prediction=lh_fit(train,row,features)
            if label.endswith('unadjusted'):
                if label=='prior_unadjusted':
                    ref=LONG_HISTORY_RESULTS['predictions']; train_label='extended'
                else:
                    ref=MISSING_FIELDWORK_RESULTS['predictions']; train_label='new'
                ref=ref[(ref.year==y)&(ref.model=='price_forecast')&(ref.training==train_label)]
                if len(ref)!=1 or not np.isclose(prediction,float(ref.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3): raise AssertionError('Prior prediction mismatch')
            actual=float(row.target_q2_usd)
            output.append({'year':y,'variant':label,'training_years':len(train),
                           'training_year_list':','.join(map(str,train.year.astype(int))),
                           'actual_usd':actual,'prediction_usd':prediction,'signed_error_usd':prediction-actual,
                           'absolute_error_usd':abs(prediction-actual)})
    pred=pd.DataFrame(output)
    scores=pred.groupby('variant',as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    annual=pred.pivot(index='year',columns='variant',values='absolute_error_usd')
    for cohort in ('prior','full'):
        if f'{cohort}_adjusted' not in annual.columns or f'{cohort}_unadjusted' not in annual.columns: raise ValueError('Missing comparison')
    report='''# Phosphate-adjusted simpler forecast model: exploratory comparison

The one-feature forecast model uses the same 2022–2026 test rows and original company-wide test targets, the same price-only benchmark, forecast feature, training-only scaling and penalty as before. The phosphate adjustment changes the training Q2 target and previous Q2 baseline for historical years 2010–2015 only, using the original dated CF segment releases. Both the shorter prior-history cohort and the full-history cohort are compared with and without that adjustment, on matched years. Existing unadjusted predictions must be reproduced exactly. The two 2013 fieldwork alternatives agree on every variable used here, so this model is invariant to that fieldwork discrepancy.

Past outcomes informed this and earlier research; these five years are not an untouched holdout. The nitrogen-only historical training figures are only an approximation to current company-wide revenue, not a restatement of test revenue. Historic forecast-vintage uncertainty and older BLS price reconstructions remain limitations. More historical years and additional model comparisons increase selection risk; results do not establish causation or reliable future improvement. The frozen 2027 protocol is unchanged.
'''
    out=Path('/kaggle/working/iowa_phosphate_forecast'); out.mkdir(parents=True,exist_ok=True)
    pred.to_csv(out/'predictions_private.csv',index=False)
    scores.to_csv(out/'scores_private.csv',index=False)
    annual.reset_index().to_csv(out/'annual_errors_private.csv',index=False)
    (out/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(scores.assign(mae_musd=lambda d:d.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display((annual/1e6).round(1))
    display(pred.assign(actual_m=lambda d:d.actual_usd/1e6,prediction_m=lambda d:d.prediction_usd/1e6,signed_error_m=lambda d:d.signed_error_usd/1e6)[['year','variant','training_years','actual_m','prediction_m','signed_error_m']].round(2))
    return {'predictions':pred,'scores':scores,'annual':annual,'report':report}

PHOSPHATE_FORECAST_RESULTS=forecast_phosphate_comparison()
```

```python
from pathlib import Path
import math
import numpy as np
import pandas as pd


def diagnose_phosphate_forecast_2025():
    orig=FIELDWORK_WINDOW_RESULTS['six']['data'].copy()
    old=LONG_HISTORY_RESULTS['history'].copy()
    cols=list(old.columns)
    data=pd.concat([old,orig[cols]],ignore_index=True).sort_values('year').reset_index(drop=True)
    if data.year.duplicated().any(): raise ValueError('Duplicate historical year')
    records={int(r['year']):r for r in PHOSPHATE_SEGMENT}
    if set(records)!=set(range(2009,2016)): raise ValueError('Incomplete phosphate sources')
    adjusted=data.copy()
    for idx, row in adjusted[adjusted.year.between(2010,2015)].iterrows():
        y=int(row.year); current=records[y]; prior=records[y-1]
        if lh_date(prior['filed_date'])>lh_date(str(row.as_of)[:10]) or lh_date(current['filed_date'])>lh_date(str(row.target_filed)[:10]):
            raise ValueError('Segment data was not available at stated cutoff')
        target=float(row.target_q2_usd)-float(current['phosphate_q2_usd'])
        base=float(row.prior_q2_usd)-float(prior['phosphate_q2_usd'])
        if min(target,base)<=0: raise ValueError('Invalid segment adjustment')
        adjusted.loc[idx,['target_q2_usd','prior_q2_usd']]=[target,base]
    rowmatches=orig[orig.year.eq(2025)]
    if len(rowmatches)!=1: raise ValueError('2025 test year missing')
    row=next(rowmatches.itertuples()); cutoff=str(row.as_of)[:10]
    if cutoff!='2025-05-31' or str(row.target_filed)[:10] <= cutoff: raise ValueError('Invalid cutoff')

    def train_from(frame):
        train=frame[(frame.year<2025)&(frame.target_filed.astype(str).str[:10]<=cutoff)].copy()
        if len(train)<6 or (train.year>=2025).any(): raise ValueError('Invalid training cohort')
        return train
    def fit(train):
        x=train.forecast_mean_p12_pct.to_numpy(float)
        target=np.log(train.target_q2_usd.to_numpy(float)/(train.prior_q2_usd.to_numpy(float)*train.price_ratio.to_numpy(float)))
        mean=float(np.mean(x)); sd=float(np.std(x,ddof=0))
        if not np.isfinite(x).all() or not np.isfinite(target).all() or not np.isfinite(sd) or sd<=1e-12:
            raise ValueError('Invalid features or targets')
        z=(x-mean)/sd; beta=float((z@target)/(z@z+1.0))
        ztest=(float(row.forecast_mean_p12_pct)-mean)/sd; term=beta*ztest
        baseline=float(row.prior_q2_usd*row.price_ratio)
        prediction=baseline*math.exp(term)
        if not np.isclose(prediction,lh_fit(train,row,['forecast_mean_p12_pct']),rtol=1e-10,atol=1e-3):
            raise AssertionError('Original fitting does not reproduce')
        return {'mu':mean,'sd':sd,'minimum':float(x.min()),'maximum':float(x.max()),'z':ztest,
                'beta':beta,'log_adjustment':term,'multiplier':math.exp(term),'baseline':baseline,
                'prediction':prediction,'log_targets':target,'features':x}
    train_unadjusted=train_from(data); train_adjusted=train_from(adjusted)
    original=fit(train_unadjusted); adj=fit(train_adjusted)
    prev=PHOSPHATE_FORECAST_RESULTS['predictions']
    def check(label,result):
        r=prev[(prev.year==2025)&(prev.variant==label)]
        if len(r)!=1 or not np.isclose(result['prediction'],float(r.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
            raise AssertionError('Saved 2025 result is not reproduced')
    check('prior_unadjusted',original);check('prior_adjusted',adj)
    actual=float(row.target_q2_usd)
    if not np.isclose(actual,float(prev[prev.year.eq(2025)].actual_usd.iloc[0])): raise AssertionError('Target mismatch')
    profiles=[]
    for label,train,f in [('unadjusted',train_unadjusted,original),('phosphate_adjusted',train_adjusted,adj)]:
        profiles.append({'variant':label,'training_years':len(train),'forecast_p12_pct':float(row.forecast_mean_p12_pct),
                         'training_mean_pct':f['mu'],'training_sd_pct':f['sd'],'training_min_pct':f['minimum'],
                         'training_max_pct':f['maximum'],'forecast_z':f['z'],'coefficient':f['beta'],
                         'forecast_log_adjustment':f['log_adjustment'],'forecast_multiplier':f['multiplier'],
                         'baseline_usd':f['baseline'],'prediction_usd':f['prediction'],'actual_usd':actual,
                         'signed_error_usd':f['prediction']-actual,'absolute_error_usd':abs(f['prediction']-actual),
                         'training_year_list':','.join(map(str,train.year.astype(int)))})
    profiles=pd.DataFrame(profiles)
    train_adjusted['adjusted_log_residual']=adj['log_targets']
    influence=[]
    for y in train_adjusted.year.astype(int):
        loo=train_adjusted[train_adjusted.year.ne(y)]
        shifted=fit(loo)['prediction']
        influence.append({'removed_year':y,'prediction_without_year_usd':shifted,
                          'change_if_removed_usd':shifted-adj['prediction']})
    influence=pd.DataFrame(influence)
    out=Path('/kaggle/working/iowa_phosphate_forecast_2025');out.mkdir(parents=True,exist_ok=True)
    profiles.to_csv(out/'diagnostic_private.csv',index=False)
    influence.to_csv(out/'leave_one_training_year_out_private.csv',index=False)
    train_adjusted.to_csv(out/'training_rows_private.csv',index=False)
    report=f'''# Diagnosing the 2025 phosphate-adjusted forecast

The fixed price-based prediction was ${adj['baseline']/1e6:,.1f}m, actual sales ${actual/1e6:,.1f}m. The precipitation probability forecast was {float(row.forecast_mean_p12_pct):.2f}%, versus earlier training mean {adj['mu']:.2f}% and range {adj['minimum']:.2f}–{adj['maximum']:.2f}% (z={adj['z']:.2f}). The fitted coefficient was {adj['beta']:+.4f}, and it changed the price-based estimate by a factor of {adj['multiplier']:.4f}, producing ${adj['prediction']/1e6:,.1f}m and an absolute error of ${abs(adj['prediction']-actual)/1e6:,.1f}m. The unadjusted model prediction was ${original['prediction']/1e6:,.1f}m. Earlier training data are restricted to targets filed by May 31, 2025; the original old and adjusted predictions must both reproduce exactly.

Leave-one-training-year-out results are retrospective influence diagnostics, not independent tests or causation. Prediction movement reflects refitting the coefficient and scaling; fitted relationships may be unstable in this small sample. The forecast is the mean probability of measurable precipitation, not total rain or purchases. CF revenue and the price baseline are company-wide/national rather than Iowa-specific. 2025 outcomes were already inspected, so no evaluated approach is an untouched test. Historical source revisions and forecast-vintage uncertainty continue to apply. The frozen 2027 protocol is unchanged.
'''
    (out/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(profiles.assign(baseline_m=lambda d:d.baseline_usd/1e6,prediction_m=lambda d:d.prediction_usd/1e6,actual_m=lambda d:d.actual_usd/1e6,signed_error_m=lambda d:d.signed_error_usd/1e6,absolute_error_m=lambda d:d.absolute_error_usd/1e6).drop(columns=['baseline_usd','prediction_usd','actual_usd','signed_error_usd','absolute_error_usd']).round(4))
    display(influence.assign(prediction_without_year_m=lambda d:d.prediction_without_year_usd/1e6,change_if_removed_m=lambda d:d.change_if_removed_usd/1e6).drop(columns=['prediction_without_year_usd','change_if_removed_usd']).sort_values('change_if_removed_m').round(2))
    display(train_adjusted[['year','forecast_mean_p12_pct','adjusted_log_residual']].round(4))
    return {'profiles':profiles,'influence':influence,'train':train_adjusted,'report':report}

PHOSPHATE_FORECAST_2025_DIAGNOSIS=diagnose_phosphate_forecast_2025()
```

```python
from pathlib import Path
from urllib.parse import urlsplit
import datetime as dt
import math
import numpy as np
import pandas as pd


def forecast_phosphate_2025_tests():
    historical = LONG_HISTORY_RESULTS['history'].copy()
    original = FIELDWORK_WINDOW_RESULTS['six']['data'].copy()
    columns = list(historical.columns)
    data = pd.concat([historical, original[columns]], ignore_index=True).sort_values('year')
    if data.year.duplicated().any() or set(historical.year.astype(int)) != {2012, 2014, 2015}:
        raise ValueError('Unexpected prior-history training cohort')
    segments = {}
    for entry in PHOSPHATE_SEGMENT:
        if not isinstance(entry, dict) or not {'year','phosphate_q2_usd','filed_date','source_url','basis'} <= set(entry):
            raise ValueError('Incomplete historical segment record')
        year = entry['year']
        value = entry['phosphate_q2_usd']
        date = lh_date(entry['filed_date'])
        url = urlsplit(entry['source_url'])
        if (isinstance(year, bool) or not isinstance(year, int) or year in segments or year not in range(2009,2016)
            or isinstance(value,bool) or not isinstance(value,(float,int)) or not np.isfinite(value) or value < 0
            or not dt.date(year,6,30) < date <= dt.date(year+1,5,31)
            or url.scheme != 'https' or url.hostname != 'www.sec.gov'
            or not url.path.startswith('/Archives/edgar/data/1324404/') or not str(entry['basis']).strip()):
            raise ValueError('Invalid historical segment record')
        segments[year] = entry
    if set(segments) != set(range(2009,2016)) or segments[2015]['phosphate_q2_usd'] != 0:
        raise ValueError('Incomplete historical segment series')

    adjusted = data.copy()
    for idx, row in adjusted[adjusted.year.between(2010,2015)].iterrows():
        year = int(row.year)
        prior, current = segments[year-1], segments[year]
        if lh_date(prior['filed_date']) > lh_date(str(row.as_of)[:10]):
            raise ValueError('Prior segment unavailable by that year cutoff')
        if lh_date(current['filed_date']) > lh_date(str(row.target_filed)[:10]):
            raise ValueError('Segment source after originally filed target')
        target = float(row.target_q2_usd) - float(current['phosphate_q2_usd'])
        baseline = float(row.prior_q2_usd) - float(prior['phosphate_q2_usd'])
        if min(target,baseline) <= 0: raise ValueError('Invalid adjusted revenue')
        adjusted.loc[idx,['target_q2_usd','prior_q2_usd']] = [target, baseline]

    volumes = {}
    q1rows = EXTRA_POINTS.get('q1_volume') if isinstance(EXTRA_POINTS,dict) else None
    if not isinstance(q1rows,list): raise ValueError('Missing Q1 volume records')
    for entry in q1rows:
        required={'year','current_q1_tons','previous_q1_tons','current_filed','previous_filed',
                  'current_accession','previous_accession','units','current_source','previous_source'}
        if not isinstance(entry,dict) or not required <= set(entry): raise ValueError('Incomplete Q1 volume record')
        y=entry['year']
        if isinstance(y,bool) or not isinstance(y,int) or y in volumes: raise ValueError('Duplicate or invalid volume year')
        for key in ('current_q1_tons','previous_q1_tons'):
            v=entry[key]
            if isinstance(v,bool) or not isinstance(v,(int,float)) or not np.isfinite(v) or v<=0:
                raise ValueError('Invalid shipment volume')
        if entry['units']!='thousand product tons': raise ValueError('Wrong shipment units')
        if not (dt.date(y,3,31)<lh_date(entry['current_filed'])<=dt.date(y,5,31)
                and dt.date(y-1,3,31)<lh_date(entry['previous_filed'])<=dt.date(y-1,5,31)):
            raise ValueError('Q1 shipment disclosure was unavailable')
        for kind in ('current','previous'):
            u=urlsplit(entry[kind+'_source'])
            if u.scheme!='https' or u.hostname!='www.sec.gov' or not u.path.startswith('/Archives/edgar/data/1324404/') or not str(entry[kind+'_accession']).strip():
                raise ValueError('Invalid Q1 volume provenance')
        volumes[y]=entry
    if not set(range(2022,2027)).issubset(volumes): raise ValueError('Incomplete Q1 volume evaluation cohort')

    predictions=[]; fold_diagnostics=[]
    for year in range(2022,2027):
        matching=original[original.year.eq(year)]
        if len(matching)!=1: raise ValueError('Missing or duplicate test row')
        test=next(matching.itertuples()); cutoff=str(test.as_of)[:10]
        if cutoff!=f'{year}-05-31' or str(test.target_filed)[:10] <= cutoff:
            raise ValueError('Invalid test cutoff')
        train=adjusted[(adjusted.year < year) & (adjusted.target_filed.astype(str).str[:10] <= cutoff)].copy()
        if len(train)<6 or any(lh_date(segments[int(y)]['filed_date']) > lh_date(cutoff)
                              for y in train.year if int(y) in range(2010,2016)):
            raise ValueError('Insufficient eligible training data')
        x=train.forecast_mean_p12_pct.to_numpy(float)
        if not np.isfinite(x).all(): raise ValueError('Invalid training forecast')
        mean=float(x.mean()); sd=float(x.std(ddof=0))
        if sd<=1e-12 or not np.isfinite(sd): raise ValueError('No forecast variation')
        z=(x-mean)/sd
        log_response=np.log(train.target_q2_usd.to_numpy(float)/(train.prior_q2_usd.to_numpy(float)*train.price_ratio.to_numpy(float)))
        beta=float((z@log_response)/(z@z+1.0))
        current=float(test.forecast_mean_p12_pct)
        baseline=float(test.prior_q2_usd*test.price_ratio)
        ztest=(current-mean)/sd
        full=baseline*math.exp(beta*ztest)
        expected=PHOSPHATE_FORECAST_RESULTS['predictions']
        expected=expected[(expected.year==year)&(expected.variant=='prior_adjusted')]
        if len(expected)!=1 or not np.isclose(full,float(expected.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
            raise AssertionError('Best phosphate-adjusted model was not reproduced')
        if not np.isclose(full,lh_fit(train,test,['forecast_mean_p12_pct']),rtol=1e-10,atol=1e-3):
            raise AssertionError('Single-feature fit disagrees with original model')
        low,high=float(x.min()),float(x.max())
        capped=float(np.clip(current,low,high))
        predictions_by_candidate={
            'adjusted_full':full,
            'adjusted_half_log_correction':baseline*math.exp(beta*ztest*0.5),
            'adjusted_training_range_clip':baseline*math.exp(beta*(capped-mean)/sd),
            'adjusted_sqrt_q1_volume':full*math.sqrt(float(volumes[year]['current_q1_tons'])/float(volumes[year]['previous_q1_tons'])),
            'price_only':baseline,
        }
        prior=PHOSPHATE_FORECAST_RESULTS['predictions']
        prior=prior[(prior.year==year)&(prior.variant=='prior_unadjusted')]
        if len(prior)!=1: raise ValueError('Missing unadjusted forecast')
        predictions_by_candidate['unadjusted_forecast']=float(prior.prediction_usd.iloc[0])
        actual=float(test.target_q2_usd)
        for label,pred in predictions_by_candidate.items():
            predictions.append({'year':year,'candidate':label,'train_years':len(train),'actual_usd':actual,
                                'prediction_usd':pred,'signed_error_usd':pred-actual,'absolute_error_usd':abs(pred-actual)})
        fold_diagnostics.append({'year':year,'train_years':len(train),'feature_pct':current,'training_mean_pct':mean,
            'training_sd_pct':sd,'training_min_pct':low,'training_max_pct':high,'standardized_value':ztest,
            'ridge_coefficient':beta,'log_adjustment':beta*ztest,'price_baseline_usd':baseline})
    pred=pd.DataFrame(predictions)
    diag=pd.DataFrame(fold_diagnostics)
    score=[]
    for omitted in (None,2025):
        group=pred if omitted is None else pred[pred.year.ne(omitted)]
        reference=group[group.candidate.eq('adjusted_full')].set_index('year').absolute_error_usd
        for name,part in group.groupby('candidate'):
            if len(part)!=len(reference): raise ValueError('Incomplete matched folds')
            score.append({'exclude_year':str(omitted) if omitted else 'none','candidate':name,
                          'years':len(part),'mae_usd':float(part.absolute_error_usd.mean()),
                          'wins_vs_adjusted_full':int((part.set_index('year').absolute_error_usd < reference).sum())})
    scores=pd.DataFrame(score)
    annual=pred.pivot(index='year',columns='candidate',values='absolute_error_usd')
    out=Path('/kaggle/working/iowa_phosphate_forecast_2025_sensitivities')
    out.mkdir(parents=True,exist_ok=True)
    for name,frame in [('predictions_private.csv',pred),('diagnostics_private.csv',diag),('scores_private.csv',scores),('annual_errors_private.csv',annual.reset_index())]:
        frame.to_csv(out/name,index=False)
    report='''# Phosphate-adjusted 2025 forecast: exploratory sensitivities

The one-feature forecast model and these fixed alternatives use identical historical May 31 prediction rows and phosphate-adjusted training rows from the prior study. Full-model predictions must reproduce that study exactly. Feature scaling, the coefficient, and the clipping bounds use only earlier training data for each simulated cutoff; test-year targets are used only for evaluation. The forecast sensitivities halve the fitted log correction or clamp the forecast feature to the training range. A third fixed candidate scales the adjusted forecast by the square root of the ratio of current to previous company Q1 product shipments; Q1 disclosures are checked against May 31, and shipment volumes represent company-wide sales rather than Iowa farmer orders. The price-only and original unadjusted forecasts are matched references.

All choices were considered after the years were inspected; no result here is an independent future test. The 2025-omitted summary is descriptive, not a fresh holdout. The full prior forecast fit lacks enough eligible earlier training rows for a like-for-like 2016–2021 comparison, so these tests do not establish performance in the earlier period. Older forecast archives, fieldwork-selected cohorts, and reconstructed price/segment history retain their prior comparability and vintage limitations. Weather probability is a local proxy, while CF revenue is company-wide. No 2027 protocol or prediction is changed.
'''
    (out/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(scores.assign(mae_musd=lambda d:d.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display((annual/1e6).round(2))
    display(diag.round(4))
    display(pred[pred.year.eq(2025)].assign(actual_m=lambda d:d.actual_usd/1e6,prediction_m=lambda d:d.prediction_usd/1e6,signed_error_m=lambda d:d.signed_error_usd/1e6)[['candidate','actual_m','prediction_m','signed_error_m']].round(2))
    return {'predictions':pred,'diagnostics':diag,'scores':scores,'annual':annual,'report':report}

PHOSPHATE_FORECAST_2025_TESTS = forecast_phosphate_2025_tests()
```

```python
from pathlib import Path
from urllib.parse import urlsplit
import datetime as dt
import math
import numpy as np
import pandas as pd


def further_2025_sensitivities():
    v2=V2_RESULTS['predictions'].copy()
    prior=PHOSPHATE_FORECAST_2025_TESTS['predictions'].copy()
    vols={}
    for v in EXTRA_POINTS['q1_volume']:
        y=int(v['year'])
        if y in vols: raise ValueError('Duplicate volume year')
        for k in ('current_q1_tons','previous_q1_tons'):
            if not np.isfinite(float(v[k])) or float(v[k]) <= 0: raise ValueError('Invalid Q1 volume')
        if v['units'] != 'thousand product tons': raise ValueError('Volume unit mismatch')
        if not (dt.date(y,3,31)<lh_date(v['current_filed'])<=dt.date(y,5,31)
                and dt.date(y-1,3,31)<lh_date(v['previous_filed'])<=dt.date(y-1,5,31)):
            raise ValueError('Q1 volume release not available before cutoff')
        for k in ('current_source','previous_source'):
            u=urlsplit(v[k])
            if u.scheme!='https' or u.hostname!='www.sec.gov' or not u.path.startswith('/Archives/edgar/data/1324404/'):
                raise ValueError('Q1 source mismatch')
        vols[y]=v
    if not set(range(2016,2027)).issubset(vols): raise ValueError('Missing volume observations')

    def pick_v2(y,model):
        t=v2[(v2.year==y)&(v2.cohort=='price_ratio')&(v2.model==model)]
        if len(t)!=1: raise ValueError(f'Missing/duplicate V2 {model} for {y}')
        r=t.iloc[0]
        if not all(np.isfinite(float(r[k])) and float(r[k])>0 for k in ('target_usd','prediction_usd')):
            raise ValueError('Invalid V2 result')
        return r
    def pick_old(y,model):
        t=prior[(prior.year==y)&(prior.candidate==model)]
        if len(t)!=1: raise ValueError(f'Missing/duplicate previous candidate {model} for {y}')
        return t.iloc[0]
    rows=[]
    for y in range(2016,2027):
        price=pick_v2(y,'price_ratio'); q1=pick_v2(y,'q1_ratio')
        if not np.isclose(price.target_usd,q1.target_usd,rtol=1e-10,atol=1): raise ValueError('Target mismatch')
        ratio=float(vols[y]['current_q1_tons'])/float(vols[y]['previous_q1_tons'])
        p=float(price.prediction_usd); q=float(q1.prediction_usd); actual=float(price.target_usd)
        predictions={'price_only':p, 'price_sqrt_volume':p*math.sqrt(ratio),
                     'q1_revenue_ratio':q, 'geometric_price_q1_revenue':math.sqrt(p*q)}
        if y>=2022:
            old=pick_old(y,'adjusted_sqrt_q1_volume')
            full=pick_old(y,'adjusted_full')
            half=pick_old(y,'adjusted_half_log_correction')
            oldprice=pick_old(y,'price_only')
            if not np.isclose(old.actual_usd,actual,rtol=1e-10,atol=1): raise AssertionError('Actual mismatch')
            if not np.isclose(oldprice.prediction_usd,p,rtol=1e-10,atol=1): raise AssertionError('Price baseline mismatch')
            rebuilt=float(full.prediction_usd)*math.sqrt(ratio)
            if not np.isclose(old.prediction_usd,rebuilt,rtol=1e-10,atol=1): raise AssertionError('Existing adjusted model mismatch')
            predictions['previous_weather_sqrt_volume']=rebuilt
            predictions['half_weather_sqrt_volume']=float(half.prediction_usd)*math.sqrt(ratio)
        for name,forecast in predictions.items():
            rows.append(dict(year=y,candidate=name,actual_usd=actual,prediction_usd=forecast,
                             signed_error_usd=forecast-actual,absolute_error_usd=abs(forecast-actual)))
    pred=pd.DataFrame(rows)
    groups=[('2016-2021',range(2016,2022)),('2022-2026',range(2022,2027)),
            ('2022-2026 excluding 2025',[2022,2023,2024,2026])]
    scores=[]
    for group,years in groups:
        part=pred[pred.year.isin(years)]
        required=len(years)
        for name,d in part.groupby('candidate'):
            if len(d)!=required: raise ValueError('Incomplete comparison cohort')
            scores.append(dict(period=group,candidate=name,years=required,
                               mae_musd=float(d.absolute_error_usd.mean()/1e6)))
    scores=pd.DataFrame(scores)
    annual=pred.pivot(index='year',columns='candidate',values='absolute_error_usd')/1e6
    current=pred[pred.year.eq(2025)].copy()
    r=vols[2025]
    row=FIELDWORK_WINDOW_RESULTS['six']['data']; row=row[row.year.eq(2025)]
    if len(row)!=1: raise ValueError('Missing 2025 model row')
    prior_q2=float(row.iloc[0].prior_q2_usd)
    print('2025 previous Q2 revenue (USD m):',round(prior_q2/1e6,2),
          '| Q1 shipment ratio:',round(float(r['current_q1_tons'])/float(r['previous_q1_tons']),4))
    print('2025 predicted revenue and signed errors (USD millions):')
    display(current.assign(prediction_m=current.prediction_usd/1e6,
                           signed_error_m=current.signed_error_usd/1e6)
            [['candidate','prediction_m','signed_error_m']].round(2))
    print('Matched mean absolute errors (USD millions):')
    display(scores.round(2))
    print('Year-by-year absolute errors (USD millions; missing older forecast variants are intentional):')
    display(annual.round(2))
    report='''# Further 2025 exploratory sensitivity checks

This reproduces the prior phosphate-adjusted weather plus square-root Q1-volume prediction, and compares fixed variants for 2022–2026. A reduced weather correction and no-weather sensitivity each retain the same square-root Q1-volume ratio. A separate geometric average combines the price-only and Q1-revenue-ratio forecasts: Q1 revenue already reflects prices and volumes and is not an independent volume estimate. Q1-volume disclosures must be filed by each historical May 31 cutoff. The V2 price and Q1-revenue forecasts reuse the previously validated as-of data and their tests; exact target and baseline agreement is enforced.

The simple price-plus-volume and geometric price-plus-Q1 comparisons also run in 2016–2021 to probe earlier performance. The weather-plus-volume model does not have enough comparable historical training rows for that earlier period. All years have already been inspected and candidates were suggested after looking at those outcomes; the small matched evaluation is not an untouched test and candidate selection may exaggerate improvements. A year excluded from a summary is not a holdout when it informed model choices. Q1 company sales reflect both prices and volume and comparisons can be distorted by changes in the previous year. Company-wide shipments and sales are not Iowa corn purchases. Forecast probability is not observed precipitation. Historical price and forecast vintage, geographical mismatch, and changing business conditions limit conclusions. These diagnostics do not establish causation or alter the frozen 2027 test.
'''
    out=Path('/kaggle/working/iowa_forecast_further_2025'); out.mkdir(parents=True,exist_ok=True)
    pred.to_csv(out/'predictions_private.csv',index=False)
    scores.to_csv(out/'scores_private.csv',index=False)
    annual.reset_index().to_csv(out/'annual_errors_private.csv',index=False)
    (out/'REPORT.md').write_text(report,encoding='utf-8')
    print(report)
    return {'predictions':pred,'scores':scores,'annual':annual,'report':report}

FURTHER_2025_RESULTS=further_2025_sensitivities()
```

```python
from pathlib import Path
import datetime as dt
import numpy as np
import pandas as pd

DECOUPLED_COLUMNS = ['year','as_of','target_filed','target_q2_usd',
                     'prior_q2_usd','price_ratio','forecast_mean_p12_pct']


def forecast_only_rows(frame):
    if not set(DECOUPLED_COLUMNS).issubset(frame.columns):
        raise ValueError('Missing forecast, price, revenue, or date field')
    data = frame[DECOUPLED_COLUMNS].copy()
    if data.year.duplicated().any():
        raise ValueError('Duplicate year')
    for name in ('target_q2_usd','prior_q2_usd','price_ratio','forecast_mean_p12_pct'):
        data[name] = pd.to_numeric(data[name], errors='raise')
        if not np.isfinite(data[name]).all(): raise ValueError(f'Non-finite {name}')
    if (data[['target_q2_usd','prior_q2_usd','price_ratio']] <= 0).any().any():
        raise ValueError('Invalid nonpositive revenue or price ratio')
    if not data.forecast_mean_p12_pct.between(0,100).all():
        raise ValueError('Invalid precipitation probability')
    for row in data.itertuples():
        y = int(row.year)
        if y != row.year or str(row.as_of)[:10] != f'{y}-05-31': raise ValueError('Bad cutoff')
        if lh_date(str(row.target_filed)[:10]) <= dt.date(y,6,30): raise ValueError('Bad target filing')
    return data.sort_values('year').reset_index(drop=True)


def build_older_forecast_only(forecasts, prices, snapshot):
    fa, forecast_coverage = lh_forecasts(forecasts)
    pa = lh_prices(prices)
    available = fa.merge(pa,on='year',how='outer',validate='one_to_one')
    for col in ('forecast_complete','price_complete'):
        available[col] = available[col].fillna(False).astype(bool)
    rev = lh_original_revenue(snapshot)
    rows, statuses = [], []
    for row in available.sort_values('year').itertuples():
        y = int(row.year)
        revenue_ok = y in rev.index and y-1 in rev.index
        complete = bool(row.forecast_complete and row.price_complete and revenue_ok)
        statuses.append(dict(year=y,forecast_complete=bool(row.forecast_complete),
                             price_complete=bool(row.price_complete),revenue_available=revenue_ok,
                             forecast_only_eligible=complete))
        if not complete: continue
        actual, prior = rev.loc[y], rev.loc[y-1]
        cutoff = dt.date(y,5,31)
        filed = lh_date(str(actual.q2_filed)[:10])
        previous_filed = lh_date(str(prior.q2_filed)[:10])
        if not (dt.date(y,6,30)<filed<=dt.date(y+1,5,31) and previous_filed<=cutoff):
            raise ValueError('Revenue filing unavailable at stated cutoff')
        if not str(actual.q2_accession).strip() or not str(prior.q2_accession).strip():
            raise ValueError('Missing source accession')
        rows.append(dict(year=y,as_of=cutoff.isoformat(),target_filed=filed.isoformat(),
                         target_q2_usd=float(actual.q2_usd),prior_q2_usd=float(prior.q2_usd),
                         price_ratio=float(row.price_ratio),forecast_mean_p12_pct=float(row.forecast_mean_p12_pct)))
    frame = pd.DataFrame(rows, columns=DECOUPLED_COLUMNS)
    return forecast_only_rows(frame), pd.DataFrame(statuses), forecast_coverage


def predict_forecast_only(training_rows, test_row):
    data=forecast_only_rows(training_rows)
    cutoff=str(test_row.as_of)[:10]
    train=data[(data.year<int(test_row.year)) & (data.target_filed.astype(str).str[:10]<=cutoff)].copy()
    if len(train)<6: raise ValueError('Insufficient earlier training observations')
    return lh_fit(train,test_row,['forecast_mean_p12_pct']),train


def _check_no_fieldwork_dependency(frame):
    base=forecast_only_rows(frame)
    changed=frame.copy()
    changed['fieldwork_complete']=False
    changed['fieldwork_mean_days']=np.nan
    pd.testing.assert_frame_equal(base,forecast_only_rows(changed))
    if len(base)>=7:
        row=next(base.tail(1).itertuples())
        p1,t1=predict_forecast_only(frame,row)
        p2,t2=predict_forecast_only(changed,row)
        pd.testing.assert_frame_equal(t1,t2)
        if not np.isclose(p1,p2,rtol=1e-12,atol=1e-3):
            raise AssertionError('Forecast changed when fieldwork was removed')
    return True


def decoupled_forecast_research():
    old, old_coverage, _ = build_older_forecast_only(HISTORICAL_FORECASTS,HISTORICAL_PRICES,SNAPSHOT_JSON)
    recent_raw = FORECAST_RESULTS['data'].copy()
    if 'price_as_of' not in recent_raw: raise ValueError('Missing recent price vintage')
    for row in recent_raw.itertuples():
        if str(row.price_as_of)[:10] != f'{int(row.year)}-05-31':
            raise ValueError('Recent price vintage mismatch')
    recent = forecast_only_rows(recent_raw)
    _check_no_fieldwork_dependency(recent_raw)
    if set(old.year) & set(recent.year): raise ValueError('Duplicate old/recent years')
    standalone = forecast_only_rows(pd.concat([old,recent],ignore_index=True))

    old_matched = forecast_only_rows(LONG_HISTORY_RESULTS['history'])
    prior_recent = forecast_only_rows(FIELDWORK_WINDOW_RESULTS['six']['data'])
    recovered = forecast_only_rows(MISSING_FIELDWORK_RESULTS['additional_years'])
    matched_prior = forecast_only_rows(pd.concat([old_matched,prior_recent],ignore_index=True))
    matched_full = forecast_only_rows(pd.concat([matched_prior,recovered],ignore_index=True))

    segments = {int(item['year']):item for item in PHOSPHATE_SEGMENT}
    if len(segments)!=len(PHOSPHATE_SEGMENT) or set(segments)!=set(range(2009,2016)):
        raise ValueError('Missing or duplicate phosphate segments')
    def adjust(frame):
        result = frame.copy()
        for idx, row in result[result.year.between(2010,2015)].iterrows():
            y=int(row.year); current=segments[y]; prior=segments[y-1]
            if lh_date(prior['filed_date'])>lh_date(str(row.as_of)[:10]) or lh_date(current['filed_date'])>lh_date(str(row.target_filed)[:10]):
                raise ValueError('Phosphate segment unavailable at cutoff')
            values=[float(row.target_q2_usd)-float(current['phosphate_q2_usd']),
                    float(row.prior_q2_usd)-float(prior['phosphate_q2_usd'])]
            if not all(np.isfinite(v) and v>0 for v in values): raise ValueError('Invalid phosphate adjustment')
            result.loc[idx,['target_q2_usd','prior_q2_usd']]=values
        return result
    cohorts = {'prior_unadjusted':matched_prior,'prior_adjusted':adjust(matched_prior),
               'full_unadjusted':matched_full,'full_adjusted':adjust(matched_full),
               'decoupled_unadjusted':standalone,'decoupled_adjusted':adjust(standalone)}
    reference = PHOSPHATE_FORECAST_RESULTS['predictions']
    outrows=[]
    for y in range(2022,2027):
        test = prior_recent[prior_recent.year.eq(y)]
        if len(test)!=1: raise ValueError('Missing matched test year')
        row=next(test.itertuples()); cutoff=str(row.as_of)[:10]
        if cutoff != f'{y}-05-31' or str(row.target_filed)[:10] <= cutoff: raise ValueError('Test cutoff mismatch')
        for label,data in cohorts.items():
            prediction,train=predict_forecast_only(data,row)
            if label.endswith('_adjusted'):
                for ty in train.year.astype(int):
                    if ty in range(2010,2016) and lh_date(segments[ty]['filed_date'])>lh_date(cutoff):
                        raise ValueError('Training segment filed after forecast cutoff')
            if label in ('prior_unadjusted','prior_adjusted','full_unadjusted','full_adjusted'):
                check=reference[(reference.year.eq(y)) & (reference.variant.eq(label))]
                if len(check)!=1 or not np.isclose(prediction,float(check.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
                    raise AssertionError('Saved matched benchmark failed to reproduce')
            actual=float(row.target_q2_usd)
            outrows.append(dict(year=y,variant=label,training_years=len(train),
                                training_year_list=','.join(map(str,train.year.astype(int))),
                                prediction_usd=prediction,actual_usd=actual,
                                signed_error_usd=prediction-actual,absolute_error_usd=abs(prediction-actual)))
    pred=pd.DataFrame(outrows)
    scores=pred.groupby('variant',as_index=False).agg(years=('year','size'),mae_usd=('absolute_error_usd','mean'))
    annual=pred.pivot(index='year',columns='variant',values='absolute_error_usd')
    prior_years=set(matched_prior.year.astype(int)); full_years=set(matched_full.year.astype(int)); free_years=set(standalone.year.astype(int))
    common=matched_full.merge(standalone,on='year',suffixes=('_full','_free'),validate='one_to_one')
    for col in ('target_q2_usd','prior_q2_usd','price_ratio','forecast_mean_p12_pct'):
        if not np.allclose(common[col+'_full'],common[col+'_free'],rtol=1e-10,atol=1e-3):
            raise AssertionError(f'Decoupled feature differs on shared rows: {col}')
    for col in ('as_of','target_filed'):
        if not common[col+'_full'].astype(str).str[:10].equals(common[col+'_free'].astype(str).str[:10]):
            raise AssertionError(f'Decoupled date differs on shared rows: {col}')
    newly_prior=sorted(free_years-prior_years); newly_full=sorted(free_years-full_years)
    absent_full=sorted(full_years-free_years)
    if not newly_full and not absent_full:
        for variant in ('adjusted','unadjusted'):
            a=pred[pred.variant.eq('full_'+variant)].sort_values('year').prediction_usd.to_numpy()
            b=pred[pred.variant.eq('decoupled_'+variant)].sort_values('year').prediction_usd.to_numpy()
            if not np.allclose(a,b,rtol=1e-10,atol=1e-3): raise AssertionError('Identical cohorts must reproduce predictions')
    report=f'''# Forecast-only fieldwork decoupling (exploratory)

The standalone training cohort is built from complete forecast, price and revenue records without requiring fieldwork, including both older 2010–2015 and newer 2016–2026 records. Test rows remain the same matched 2022–2026 rows, and fits use only earlier filed targets with training-only scaling. Existing prior/full, adjusted/unadjusted predictions must reproduce before comparing. Years eligible in the decoupled cohort relative to the earlier prior cohort: {newly_prior}; additional years relative to the previously recovered full cohort: {newly_full}; years in full but absent from the decoupled cohort: {absent_full}. Some formerly missing fieldwork years were already restored to the full comparison, so decoupling does not necessarily add new data or improve scores. Full and independent cohorts with identical year and feature sets must produce identical predictions.

This is retrospective, not a fresh holdout; all 2022–2026 results were inspected. Dated sources are validated as in the earlier notebook, but manually recorded historic forecast values, forecast issuance times, older price reconstructions, and revisions remain unverified. The forecast feature is a probability rather than rainfall, fieldwork or farmer purchases. Results concern total company revenue. The frozen 2027 protocol is unchanged. Coverage, per-year errors and results stay on Kaggle.
'''
    output=Path('/kaggle/working/iowa_decoupled_forecast'); output.mkdir(parents=True,exist_ok=True)
    for filename,frame in [('historical_coverage_private.csv',old_coverage),('decoupled_rows_private.csv',standalone),('predictions_private.csv',pred),('scores_private.csv',scores),('annual_errors_private.csv',annual.reset_index())]:
        frame.to_csv(output/filename,index=False)
    (output/'REPORT.md').write_text(report,encoding='utf8')
    print(report)
    display(old_coverage)
    display(scores.assign(mae_musd=lambda d:d.mae_usd/1e6).drop(columns='mae_usd').round(2))
    display((annual/1e6).round(2))
    return {'historical_coverage':old_coverage,'data':standalone,'predictions':pred,'scores':scores,'annual':annual,'report':report}


DECOUPLED_FORECAST_RESULTS=decoupled_forecast_research()
```

```python
import json, gzip, base64
_browser_forecast_rows=json.loads(gzip.decompress(base64.b64decode("")))
_browser_forecasts={(r["year"],r["station"]):{"source_url":r["source_url"],"retrieved_at_utc":r["retrieved_utc"],"records_scope":"selection_only","records":r["rows"]} for r in _browser_forecast_rows}

from pathlib import Path
from urllib.parse import urlencode
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError
from datetime import datetime, timezone
import gzip
import hashlib
import json
import math
import time

import pandas as pd

AUDIT_FORECAST_ENDPOINT = 'https://mesonet.agron.iastate.edu/cgi-bin/request/mos.py'
AUDIT_FORECAST_STATIONS = ('KAMW','KCID','KSUX')
AUDIT_FORECAST_YEARS = tuple(range(2010,2027))
AUDIT_FORECAST_FIELDS = ('station','model','runtime','ftime','p12')
AUDIT_FORECAST_OUT = Path('/kaggle/working/iowa_forecast_input_audit')

def _audit_canonical(value):
    return json.dumps(value,sort_keys=True,ensure_ascii=False,separators=(',',':'),allow_nan=False)

def _audit_hash(value):
    return hashlib.sha256(_audit_canonical(value).encode('utf-8')).hexdigest()

def _audit_utc(value):
    try:
        t=pd.Timestamp(value)
        if pd.isna(t): return None
        return t.tz_localize('UTC') if t.tzinfo is None else t.tz_convert('UTC')
    except (TypeError, ValueError, OverflowError):
        return None

def _audit_forecast_url(station,year):
    if station not in AUDIT_FORECAST_STATIONS or year not in AUDIT_FORECAST_YEARS:
        raise ValueError('Unknown station or year')
    params={'station':station,'model':'GFS','sts':f'{year}-04-01T00:00Z',
            'ets':f'{year}-05-21T00:00Z','format':'json'}
    return AUDIT_FORECAST_ENDPOINT+'?'+urlencode(params)

def audit_forecast_rows(records, station, year):
    if not isinstance(records,list): raise ValueError('Expected JSON row array')
    expected=set(pd.date_range(f'{year}-04-01',f'{year}-05-20',freq='D',tz='UTC'))
    daily=[]; values={}; invalid=0; duplicates=0
    for row in records:
        if not isinstance(row,dict):
            daily.append({'invalid_non_dict':row,'invalid_non_dict_sha256':_audit_hash(row)})
            invalid+=1
            continue
        runtime=_audit_utc(row.get('runtime')); valid=_audit_utc(row.get('ftime'))
        if runtime not in expected or valid is None or valid != runtime+pd.Timedelta(hours=24):
            continue
        daily.append({'original_source_record':row,'original_record_sha256':_audit_hash(row)})
        if row.get('station') != station or row.get('model') != 'GFS':
            invalid+=1
            continue
        value=row.get('p12')
        if value is None or value in ('','M','null'): continue
        if isinstance(value,bool): invalid+=1; continue
        try: value=float(value)
        except (ValueError,TypeError): invalid+=1; continue
        if not math.isfinite(value) or not 0<=value<=100: invalid+=1; continue
        if runtime in values:
            if value in values[runtime]: duplicates+=1
            values[runtime].add(value)
        else: values[runtime]={value}
    conflict=sum(len(x)>1 for x in values.values())
    clean=[next(iter(x)) for x in values.values() if len(x)==1]
    return daily,{'expected_days':50,'valid_days':len(clean),'duplicate_rows':duplicates,
                  'conflicting_days':conflict,'invalid_rows':invalid,
                  'station_mean_p12_pct':sum(clean)/len(clean) if clean else None,
                  'usable':len(clean)>=40 and invalid==0}

def verify_saved_forecast_ledger(path, manifests):
    groups={}
    with gzip.open(path,'rt',encoding='utf-8') as f:
        for line in f:
            row=json.loads(line)
            if 'original_source_record' in row:
                raw=row['original_source_record']
                if _audit_hash(raw)!=row['original_record_sha256']:
                    raise AssertionError('Saved original-row checksum mismatch')
            else:
                raw=row['invalid_non_dict']
                if _audit_hash(raw)!=row['invalid_non_dict_sha256']:
                    raise AssertionError('Saved invalid-row checksum mismatch')
            groups.setdefault((row['year'],row['station']),[]).append(raw)
    retrieved={(m['year'],m['station']) for m in manifests if m['status']=='retrieved'}
    if not set(groups).issubset(retrieved): raise AssertionError('Unattributed saved row')
    for m in manifests:
        if m['status']!='retrieved': continue
        _, rebuilt=audit_forecast_rows(groups.get((m['year'],m['station']),[]),m['station'],m['year'])
        for key in ('expected_days','valid_days','duplicate_rows','conflicting_days','invalid_rows','usable'):
            if rebuilt[key]!=m[key]: raise AssertionError('Saved ledger failed to rebuild '+key)
        a,b=rebuilt['station_mean_p12_pct'],m['station_mean_p12_pct']
        if (a is None)!=(b is None) or (a is not None and not math.isclose(a,b,rel_tol=1e-12,abs_tol=1e-12)):
            raise AssertionError('Saved ledger failed to rebuild mean')
    return len(retrieved)

def audit_forecast_history(historical_snapshot, newer_snapshot, *, output_dir=AUDIT_FORECAST_OUT,
                           timeout_seconds=8, max_requests=51, max_seconds=100, fetch=None,
                           collected=None):
    output=Path(output_dir); output.mkdir(parents=True,exist_ok=True)
    if timeout_seconds<=0 or not 0<=max_requests<=51 or max_seconds<=0:
        raise ValueError('Invalid network limits')
    snapshots={}
    for item in list(historical_snapshot)+list(newer_snapshot):
        key=(int(item['year']),str(item['station']))
        if key in snapshots: raise ValueError('Duplicate snapshot station/year')
        snapshots[key]=item
    expected={(y,s) for y in AUDIT_FORECAST_YEARS for s in AUDIT_FORECAST_STATIONS}
    if set(snapshots)!=expected:
        raise ValueError('Historical snapshot coverage differs from 2010–2026 station set')
    if collected is not None and (not isinstance(collected,dict) or not set(collected).issubset(expected)):
        raise ValueError('Invalid browser-collected station/year mapping')
    def http_fetch(url,timeout):
        req=Request(url,headers={'User-Agent':'IowaForecastResearch/1.0 public research','Accept':'application/json'})
        with urlopen(req,timeout=timeout) as response:
            return response.read()
    fetch=fetch or http_fetch
    ledger=[]; manifests=[]; start=time.monotonic(); attempts=0; stop_reason=None
    last_attempt=None
    for year in AUDIT_FORECAST_YEARS:
        for station in AUDIT_FORECAST_STATIONS:
            url=_audit_forecast_url(station,year); snapshot=snapshots[(year,station)]
            m={'year':year,'station':station,'source_url':url,'source_version':'unknown',
               'publication_time_utc':None,'availability_time_utc':None,
               'historical_availability':'unknown','retrieved_at_utc':None,
               'response_sha256':None,'parsed_response_canonical_sha256':None,
               'selected_payload_canonical_sha256':None,'records_scope':None,
               'selection_completeness':'unknown','retrieval_method':None,
               'status':'not_attempted','details':None,
               'prior_valid_days':snapshot.get('valid_days'),
               'prior_mean_p12_pct':snapshot.get('station_mean_p12_pct')}
            if collected is not None and (year,station) not in collected:
                m['details']='not present in browser-collected mapping'; manifests.append(m); continue
            if stop_reason is not None or attempts>=max_requests or time.monotonic()-start>=max_seconds:
                m['details']=stop_reason or 'request/time budget exhausted'; manifests.append(m); continue
            if collected is None and last_attempt is not None:
                pause=1.1-(time.monotonic()-last_attempt)
                if pause>0: time.sleep(pause)
            if time.monotonic()-start>=max_seconds:
                m['details']='request/time budget exhausted'; manifests.append(m); continue
            last_attempt=time.monotonic()
            attempts+=1
            try:
                if collected is None:
                    payload=fetch(url,timeout_seconds)
                    if not isinstance(payload,bytes): raise ValueError('Fetch did not return bytes')
                    if len(payload)>15_000_000: raise ValueError('Response exceeded byte limit')
                    records=json.loads(payload.decode('utf-8'))
                    retrieved=datetime.now(timezone.utc).isoformat()
                    m['response_sha256']=hashlib.sha256(payload).hexdigest()
                    m['retrieval_method']='direct_http'; m['records_scope']='complete_json_array'
                else:
                    item=collected[(year,station)]
                    if not isinstance(item,dict) or item.get('source_url')!=url:
                        raise ValueError('Collected source URL missing or does not match')
                    retrieved=str(item.get('retrieved_at_utc',''))
                    timestamp=_audit_utc(retrieved)
                    if timestamp is None or not retrieved.endswith(('Z','+00:00')):
                        raise ValueError('Collected retrieval time must be an explicit UTC timestamp')
                    if timestamp>pd.Timestamp.now(tz='UTC')+pd.Timedelta(minutes=1):
                        raise ValueError('Collected retrieval time is in the future')
                    records=item.get('records')
                    if not isinstance(records,list): raise ValueError('Collected response is not an array')
                    scope=item.get('records_scope')
                    if scope not in ('complete_json_array','selection_only'):
                        raise ValueError('Browser-collected record scope must be explicit')
                    if scope=='selection_only':
                        expected_times=set(pd.date_range(f'{year}-04-01',f'{year}-05-20',freq='D',tz='UTC'))
                        for source_row in records:
                            if not isinstance(source_row,dict): raise ValueError('Selected source row must be a dict')
                            rt=_audit_utc(source_row.get('runtime')); vt=_audit_utc(source_row.get('ftime'))
                            if rt not in expected_times or vt is None or vt!=rt+pd.Timedelta(hours=24):
                                raise ValueError('Selected source row lies outside fixed selection')
                        m['selected_payload_canonical_sha256']=_audit_hash(records)
                        m['details']='Browser preselection; completeness and full-response invalid counts are not independently verified'
                    else:
                        m['parsed_response_canonical_sha256']=_audit_hash(records)
                    m['records_scope']=scope; m['retrieval_method']='browser_parsed_array'
                kept,summary=audit_forecast_rows(records,station,year)
                m.update(summary); m['retrieved_at_utc']=retrieved
                m['status']='retrieved'
                previous=snapshot.get('station_mean_p12_pct')
                m['mean_difference_pct_points']=(summary['station_mean_p12_pct']-float(previous)
                    if summary['station_mean_p12_pct'] is not None and previous is not None else None)
                m['same_count_and_mean_within_1e5_pct_points']=(summary['valid_days']==snapshot.get('valid_days') and
                    (m['mean_difference_pct_points'] is not None and abs(m['mean_difference_pct_points'])<=1e-5))
                for item in kept:
                    ledger.append({'year':year,'station':station,'source_url':url,
                        'retrieved_at_utc':retrieved,'publication_time_utc':None,
                        'availability_time_utc':None,'source_version':'unknown',
                        'retrieval_method':m['retrieval_method'],'records_scope':m['records_scope'],**item})
            except (HTTPError,URLError,TimeoutError,OSError,ValueError,UnicodeError,json.JSONDecodeError) as exc:
                m['status']='fetch_or_parse_error'; m['details']=f'{type(exc).__name__}: {str(exc)[:200]}'
                if isinstance(exc,HTTPError):
                    if exc.code in (401,403,429) or exc.code>=500:
                        stop_reason='source access/throttle/server error; remaining requests were not attempted'
                elif collected is None and (isinstance(exc,(URLError,TimeoutError,OSError)) or attempts==1):
                    stop_reason='source network/first response failed; remaining requests were not attempted'
            manifests.append(m)
    for r in ledger:
        if 'original_source_record' in r:
            if _audit_hash(r['original_source_record'])!=r['original_record_sha256']:
                raise AssertionError('Original-row checksum mismatch')
        elif _audit_hash(r['invalid_non_dict'])!=r['invalid_non_dict_sha256']:
            raise AssertionError('Invalid-row checksum mismatch')
    with gzip.open(output/'forecast_daily_private.jsonl.gz','wt',encoding='utf-8') as f:
        for row in ledger: f.write(_audit_canonical(row)+'\n')
    verified=verify_saved_forecast_ledger(output/'forecast_daily_private.jsonl.gz',manifests)
    pd.DataFrame(manifests).to_csv(output/'forecast_sources_private.csv',index=False)
    successes=sum(x['status']=='retrieved' for x in manifests)
    report=(f'Forecast station-years retrieved: {successes}/{len(manifests)}; '
        f'selected source rows preserved: {len(ledger)}; rebuilt from saved ledger: {verified}. '
        'See status and snapshot comparison per station-year. '
        'Source version, actual publication and availability times are unknown. '
        'This is the current archive response, possibly revised; it does not prove historical availability. '
        'Model initialization is not publication time. Row hashes cover canonical JSON of each saved full original daily record; '
        'Direct response hashes identify response bytes but those full bytes are not retained. '
        'Browser-collected parsed arrays or selected-row payloads have a canonical JSON hash, not a response-byte hash. '
        'Browser selection completeness and invalid counts outside selected rows cannot be verified by this code. '
        'Snapshot comparison uses an explicit 1e-5 percentage-point tolerance; differences can reflect original rounding or revisions. '
        'The GFS query can return legacy AVN records; they are retained and counted invalid rather than silently accepted. '
        'Requests not attempted and failures are not missing weather observations. Frozen 2027 test unchanged.\n')
    (output/'REPORT.txt').write_text(report,encoding='utf-8')
    print(report)
    return {'ledger':ledger,'manifest':pd.DataFrame(manifests),'report':report}

FORECAST_INPUT_AUDIT=audit_forecast_history(HISTORICAL_FORECASTS,FORECAST_SNAPSHOT,collected=_browser_forecasts)

from pathlib import Path
from urllib.parse import urlencode, urlsplit, parse_qs
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError
import csv
import datetime as dt
import hashlib
import io
import json
import math
import time

PRICE_AUDIT_OUT = Path('/kaggle/working/iowa_price_audit')
PRICE_AUDIT_SERIES = 'WPU065201'
PRICE_AUDIT_ENDPOINT = 'https://alfred.stlouisfed.org/graph/alfredgraph.csv'

class PriceObservationUnavailable(ValueError):
    pass

def _price_audit_now():
    return dt.datetime.now(dt.timezone.utc).isoformat(timespec='seconds')

def _price_audit_date(value):
    if not isinstance(value,str): raise ValueError('Date must be an ISO string')
    value2=dt.date.fromisoformat(value)
    if value2.isoformat()!=value: raise ValueError('Date must be canonical ISO')
    return value

def _price_audit_record(value, year, observation, source, basis,
                        recorded_as_of=None, recorded_release_date=None):
    year=int(year); observation=_price_audit_date(str(observation))
    value=float(value)
    if not math.isfinite(value) or value<=0: raise ValueError('Invalid price')
    u=urlsplit(str(source))
    if u.scheme!='https' or not (u.hostname=='alfred.stlouisfed.org' or
                                  u.hostname=='bls.gov' or (u.hostname or '').endswith('.bls.gov')):
        raise ValueError('Unexpected price source')
    if recorded_as_of is not None: _price_audit_date(recorded_as_of)
    if recorded_release_date is not None: _price_audit_date(recorded_release_date)
    return {'forecast_year':year,'series_id':PRICE_AUDIT_SERIES,
            'observation_date':observation,'value':value,'source_url':str(source),
            'recorded_as_of_date':recorded_as_of,
            'source_release_date_from_snapshot':recorded_release_date,
            'publication_or_available_at_utc':None,
            'historical_retrieved_at_utc':None,
            'source_version':('ALFRED as-of date '+recorded_as_of if recorded_as_of else
                              'BLS reconstructed from dated releases; per-value version unverified'),
            'evidence_status':basis,
            'note':'Snapshot value; source content and time-of-day were not independently verified.'}

def price_audit_snapshot(snapshot, older_prices):
    if not isinstance(snapshot,dict) or not isinstance(snapshot.get('price_index'),list):
        raise ValueError('Missing snapshot price_index')
    if not isinstance(older_prices,list): raise ValueError('Missing historical price list')
    observations=[]; seen=set()
    for item in snapshot['price_index']:
        y=int(item['year']); cutoff=f'{y}-05-31'
        if y in seen: raise ValueError('Duplicate snapshot year')
        seen.add(y)
        if item['as_of']!=cutoff or item['vintage_label']!=cutoff:
            raise ValueError('Snapshot as-of label mismatch')
        for prefix, obs_year in [('previous',y-1),('current',y)]:
            obs=item[prefix+'_observation']
            if obs!=f'{obs_year}-04-01': raise ValueError('Wrong snapshot observation date')
            observations.append(_price_audit_record(item[prefix+'_april'],y,obs,item['source'],
                                'manual_alfred_asof_snapshot_not_independently_retrieved',cutoff))
    seen=set()
    for item in older_prices:
        y=int(item['year'])
        if y in seen: raise ValueError('Duplicate historical year')
        seen.add(y)
        for prefix, obs_year in [('previous',y-1),('current',y)]:
            release=_price_audit_date(item[prefix+'_release_date'])
            if dt.date.fromisoformat(release)>dt.date(y,5,31):
                raise ValueError('Historical release later than cutoff')
            observations.append(_price_audit_record(item[prefix+'_april'],y,f'{obs_year}-04-01',
                item[prefix+'_source_url'],'manual_bls_reconstruction_not_independently_retrieved',
                recorded_release_date=release))
    return observations

def _price_audit_url(year):
    y=int(year); cutoff=f'{y}-05-31'
    return PRICE_AUDIT_ENDPOINT+'?'+urlencode({'id':PRICE_AUDIT_SERIES,
        'vintage_date':cutoff,'cosd':f'{y-1}-04-01','coed':f'{y}-04-01'})

def parse_price_alfred_csv(raw,year):
    if len(raw)>500_000: raise ValueError('Response exceeds small audit limit')
    text=raw.decode('utf-8-sig',errors='strict')
    lines=list(csv.reader(io.StringIO(text)))
    expected=f'{PRICE_AUDIT_SERIES}_{int(year)}0531'
    if not lines or len(lines[0])!=2 or lines[0][0] not in ('observation_date','DATE','date') or lines[0][1]!=expected:
        raise ValueError('ALFRED response has unexpected column/vintage header')
    found={}
    for row in lines[1:]:
        if len(row)!=2: raise ValueError('Unexpected ALFRED CSV shape')
        date=_price_audit_date(row[0])
        if date in found: raise ValueError('Duplicate observation date')
        if row[1] in ('','.'): continue
        v=float(row[1])
        if not math.isfinite(v) or v<=0: raise ValueError('Invalid source value')
        found[date]=v
    dates=(f'{int(year)-1}-04-01',f'{int(year)}-04-01')
    if any(d not in found for d in dates): raise PriceObservationUnavailable('As-of source missing an April observation')
    return {d:found[d] for d in dates}

def run_price_audit(snapshot,older_prices,fetch_alfred=False,timeout=8):
    observations=price_audit_snapshot(snapshot,older_prices)
    now=_price_audit_now(); PRICE_AUDIT_OUT.mkdir(parents=True,exist_ok=True)
    reconciled=[]; records=[]; attempted=False; network_stopped=False
    years=sorted({r['forecast_year'] for r in observations})
    for y in years:
        url=_price_audit_url(y); recorded=[r for r in observations if r['forecast_year']==y]
        result={'forecast_year':y,'source_url':url,'requested_as_of_date':f'{y}-05-31',
                'retrieved_at_utc':None,'response_sha256':None,'response_file':None,
                'status':'not_requested','comparison':None,'error':None}
        if fetch_alfred and not network_stopped:
            attempted=True
            try:
                req=Request(url,headers={'User-Agent':'IowaFertilizerResearch/1.0 (source audit)',
                                         'Accept':'text/csv'})
                with urlopen(req,timeout=timeout) as response:
                    raw=response.read(500_001)
                retrieved=_price_audit_now(); parsed=parse_price_alfred_csv(raw,y)
                digest=hashlib.sha256(raw).hexdigest()
                dest=PRICE_AUDIT_OUT/f'alfred_{y}_asof_may31.csv'
                dest.write_bytes(raw)
                discrepancies=[]
                for original in recorded:
                    source_value=parsed.get(original['observation_date'])
                    if source_value is None or not math.isclose(original['value'],source_value,rel_tol=1e-9,abs_tol=1e-7):
                        discrepancies.append({'observation_date':original['observation_date'],
                            'snapshot_value':original['value'],'source_value':source_value})
                    records.append({'forecast_year':y,'series_id':PRICE_AUDIT_SERIES,
                        'observation_date':original['observation_date'],'value':source_value,
                        'source_url':url,'source_version':f'ALFRED as-of {y}-05-31',
                        'publication_or_available_at_utc':None,'retrieved_at_utc':retrieved,
                        'response_sha256':digest,'response_file':dest.name,
                        'evidence_status':'fresh_source_response_asof_header_checked',
                        'note':'Header identifies requested as-of date; individual release and intraday availability not supplied.'})
                result.update(status='retrieved',retrieved_at_utc=retrieved,response_sha256=digest,
                              response_file=dest.name,comparison='match' if not discrepancies else 'mismatch',
                              discrepancies=discrepancies)
            except HTTPError as exc:
                if exc.code==404:
                    result.update(status='source_vintage_unavailable',error=f'HTTP {exc.code}')
                else:
                    result.update(status='fetch_failed',error=f'HTTP {exc.code}')
                    network_stopped=True
            except PriceObservationUnavailable as exc:
                result.update(status='observation_unavailable',error=str(exc))
            except (OSError,URLError,TimeoutError) as exc:
                result.update(status='fetch_failed',error=f'{type(exc).__name__}: {str(exc)[:180]}')
                network_stopped=True
            except (UnicodeError,ValueError,csv.Error) as exc:
                result.update(status='unverifiable_response',error=f'{type(exc).__name__}: {str(exc)[:180]}')
                network_stopped=True
            time.sleep(0.2)
        elif fetch_alfred and network_stopped:
            result.update(status='not_attempted_after_failure')
        reconciled.append(result)
    def encode_record(record):
        canonical=json.dumps(record,sort_keys=True,separators=(',',':'),ensure_ascii=False,allow_nan=False)
        return {'record':record,'canonical_sha256':hashlib.sha256(canonical.encode('utf-8')).hexdigest()}
    ledger=[encode_record(r) for r in observations+records]
    (PRICE_AUDIT_OUT/'price_observations_private.jsonl').write_text(
        ''.join(json.dumps(r,sort_keys=True,ensure_ascii=False,allow_nan=False)+'\n' for r in ledger),encoding='utf8')
    (PRICE_AUDIT_OUT/'reconciliation_private.json').write_text(json.dumps(reconciled,indent=2),encoding='utf8')
    totals={'audit_created_at_utc':now,'series_id':PRICE_AUDIT_SERIES,
            'snapshot_observations':len(observations),'fresh_observations':len(records),
            'source_vintages_retrieved':sum(r['status']=='retrieved' for r in reconciled),
            'matching_vintages':sum(r['comparison']=='match' for r in reconciled),
            'mismatching_vintages':sum(r['comparison']=='mismatch' for r in reconciled),
            'fetch_requested':bool(fetch_alfred),'fetch_attempted':attempted,
            'fetch_failure':next((r for r in reconciled if r['status'] in ('fetch_failed','unverifiable_response')),None),
            'warning':'Checksums prove byte integrity only, not publication time, archive completeness or historical availability. Earlier BLS reconstructions are not an independent ALFRED source; they may legitimately differ. No existing predictions, datasets or 2027 protocol were changed.'}
    (PRICE_AUDIT_OUT/'manifest_private.json').write_text(json.dumps(totals,indent=2),encoding='utf8')
    print('Price audit:',json.dumps(totals,indent=2))
    return {'manifest':totals,'observations':observations,'fresh_observations':records,'reconciliation':reconciled}

def ingest_browser_price_evidence(pairs,snapshot):
    if not isinstance(pairs,list) or not isinstance(snapshot,dict):
        raise ValueError('Expected browser evidence list and existing snapshot')
    existing={int(r['year']):r for r in snapshot['price_index']}
    if len(existing)!=len(snapshot['price_index']): raise ValueError('Duplicate model snapshot year')
    seen=set(); rows=[]
    for p in pairs:
        if not isinstance(p,dict): raise ValueError('Invalid browser evidence pair')
        y=int(p['year']); cutoff=f'{y}-05-31'
        if y in seen: raise ValueError('Duplicate browser evidence year')
        seen.add(y)
        if y not in existing: raise ValueError('Browser vintage has no saved model counterpart')
        if existing[y]['as_of']!=cutoff or existing[y]['vintage_label']!=cutoff:
            raise ValueError('Model snapshot vintage mismatch')
        u=urlsplit(p['source_url']); q=parse_qs(u.query)
        expected={'id':[PRICE_AUDIT_SERIES],'vintage_date':[cutoff],
                  'cosd':[f'{y-1}-04-01'],'coed':[f'{y}-04-01']}
        if u.scheme!='https' or u.hostname!='alfred.stlouisfed.org' or u.path!='/graph/' or any(q.get(k)!=v for k,v in expected.items()):
            raise ValueError('Graph source/version does not match requested year')
        start=dt.datetime.fromisoformat(p['retrieval_start_utc'].replace('Z','+00:00'))
        end=dt.datetime.fromisoformat(p['retrieval_end_utc'].replace('Z','+00:00'))
        if start.tzinfo is None or end.tzinfo is None or start.utcoffset()!=dt.timedelta(0) or end.utcoffset()!=dt.timedelta(0) or end<start:
            raise ValueError('Retrieval interval must be ordered and in UTC')
        for label,obs_year in [('previous',y-1),('current',y)]:
            observed=float(p[label+'_april']); saved=float(existing[y][label+'_april'])
            if not all(math.isfinite(x) and x>0 for x in (observed,saved)):
                raise ValueError('Invalid observation')
            if existing[y][label+'_observation']!=f'{obs_year}-04-01':
                raise ValueError('Snapshot observation period mismatch')
            row={'forecast_year':y,'series_id':PRICE_AUDIT_SERIES,
                 'observation_date':f'{obs_year}-04-01','value':observed,
                 'source_url':p['source_url'],'source_version':'ALFRED as-of '+cutoff,
                 'historical_publication_or_available_at_utc':None,
                 'retrieval_start_utc':p['retrieval_start_utc'],'retrieval_end_utc':p['retrieval_end_utc'],
                 'displayed_series_updated_text':p.get('displayed_series_updated_optional'),
                 'source_byte_sha256':None,'snapshot_value':saved,
                 'matches_snapshot':math.isclose(observed,saved,rel_tol=1e-9,abs_tol=1e-7),
                 'evidence_status':'browser_transcribed_asof_graph_not_source_bytes',
                 'note':'Requested as-of date and visible graph value; retrieval interval is current audit, not historical availability. Original source bytes and individual release times not retained.'}
            canonical=json.dumps(row,sort_keys=True,separators=(',',':'),ensure_ascii=False,allow_nan=False)
            rows.append({'record':row,'canonical_sha256':hashlib.sha256(canonical.encode('utf8')).hexdigest()})
    PRICE_AUDIT_OUT.mkdir(parents=True,exist_ok=True)
    (PRICE_AUDIT_OUT/'browser_price_evidence_private.jsonl').write_text(
        ''.join(json.dumps(r,sort_keys=True,ensure_ascii=False)+'\n' for r in rows),encoding='utf8')
    summary={'browser_vintage_pairs':len(pairs),'browser_observations':len(rows),
             'matching_snapshot_observations':sum(r['record']['matches_snapshot'] for r in rows),
             'mismatches':[r['record'] for r in rows if not r['record']['matches_snapshot']],
             'missing_model_years':sorted(set(existing)-seen),
             'note':'Browser-transcribed ALFRED as-of graphs. Record hashes verify saved transcription only; they do not prove raw source bytes or exact historical availability.'}
    (PRICE_AUDIT_OUT/'browser_price_reconciliation_private.json').write_text(json.dumps(summary,indent=2),encoding='utf8')
    print('Browser price evidence:',json.dumps(summary,indent=2))
    return {'summary':summary,'observations':rows}

PRICE_AUDIT_RESULTS=run_price_audit(SNAPSHOT_JSON,HISTORICAL_PRICES,
    fetch_alfred=bool(globals().get('PRICE_AUDIT_FETCH_ALFRED',False)))


_browser_price_pairs = [[2016,325.1,275.6,"May 13, 2016 12:54 PM CDT"],[2017,275.6,260.5,"May 11, 2017 8:46 AM CDT"],[2018,258.8,259.4,"May 9, 2018 8:43 AM CDT"],[2019,259.7,261,"May 9, 2019 7:52 AM CDT"],[2020,276.2,236.7,"May 13, 2020 7:55 AM CDT"],[2021,247.8,302.7,"May 13, 2021 8:05 AM CDT"],[2022,298.1,643.171,"May 12, 2022 7:58 AM CDT"],[2023,665.277,390.037,"May 11, 2023 7:56 AM CDT"],[2024,394.114,349.732,"May 14, 2024 7:58 AM CDT"],[2025,344.408,370.604,"May 15, 2025 7:55 AM CDT"],[2026,362.254,480.39,"May 13, 2026 9:28 AM CDT"]]
_browser_prices=[{'year':y,'previous_april':old,'current_april':new,'source_url':f'https://alfred.stlouisfed.org/graph/?id=WPU065201&vintage_date={y}-05-31&cosd={y-1}-04-01&coed={y}-04-01','retrieval_start_utc':'2026-11-05T14:30:00Z','retrieval_end_utc':'2026-11-05T14:35:00Z','displayed_series_updated_optional':updated} for y,old,new,updated in _browser_price_pairs]
BROWSER_PRICE_RESULTS=ingest_browser_price_evidence(_browser_prices,SNAPSHOT_JSON)
print('Audit complete; source availability remains unverified.')
```

```python
from pathlib import Path
from collections import defaultdict
import math
import numpy as np
import pandas as pd

MD_STATIONS=('KAMW','KCID','KSUX')
MD_YEARS=tuple(range(2010,2027))
MD_OUT=Path('/kaggle/working/iowa_missing_forecast_days')

def md_day_map(records, station, year):
    expected=list(pd.date_range(f'{year}-04-01',f'{year}-05-20',freq='D',tz='UTC'))
    vals=defaultdict(set); seen=defaultdict(int); nulls=defaultdict(int); invalid=defaultdict(int)
    for row in records:
        if not isinstance(row,dict): raise ValueError('Unexpected non-dict in selected ledger')
        runtime=_audit_utc(row.get('runtime')); valid=_audit_utc(row.get('ftime'))
        if runtime not in expected or valid is None or valid!=runtime+pd.Timedelta(hours=24):
            raise ValueError('Selected row outside configured window')
        seen[runtime]+=1
        if row.get('station')!=station or row.get('model')!='GFS':
            invalid[runtime]+=1; continue
        value=row.get('p12')
        if value is None or value in ('','M','null'):
            nulls[runtime]+=1; continue
        if isinstance(value,bool): invalid[runtime]+=1; continue
        try: v=float(value)
        except (TypeError,ValueError): invalid[runtime]+=1; continue
        if not math.isfinite(v) or not 0<=v<=100: invalid[runtime]+=1; continue
        vals[runtime].add(v)
    result=[]
    for day in expected:
        choices=vals.get(day,set())
        if len(choices)>1: state='conflicting_values'; v=None
        elif len(choices)==1: state='valid'; v=next(iter(choices))
        elif invalid.get(day,0): state='invalid_only'; v=None
        elif nulls.get(day,0): state='null_only'; v=None
        else: state='absent_in_selected_ledger'; v=None
        result.append({'year':year,'station':station,'runtime_utc':day.isoformat(),
                       'date':day.date().isoformat(),'state':state,'p12_pct':v,
                       'selected_rows':seen.get(day,0),'null_rows':nulls.get(day,0),
                       'invalid_rows':invalid.get(day,0),'distinct_values':len(choices)})
    return result

def md_build_daily(audit):
    if not isinstance(audit,dict) or not {'ledger','manifest'}<=set(audit):
        raise ValueError('Missing input audit ledger and manifest')
    man=audit['manifest'].copy()
    if man.duplicated(['year','station']).any(): raise ValueError('Duplicate audit manifest')
    observed={}
    for row in audit['ledger']:
        key=(int(row['year']),str(row['station']))
        if 'original_source_record' not in row: raise ValueError('Unexpected non-dict selected row')
        if _audit_hash(row['original_source_record'])!=row['original_record_sha256']:
            raise AssertionError('Selected source record checksum mismatch')
        observed.setdefault(key,[]).append(row['original_source_record'])
    daily=[]; stats=[]
    for y in MD_YEARS:
        for station in MD_STATIONS:
            item=man[(man.year.eq(y)) & man.station.eq(station)]
            if len(item)!=1: raise ValueError('Expected one manifest per station-year')
            m=item.iloc[0]
            if m.status!='retrieved':
                stats.append({'year':y,'station':station,'source_status':str(m.status),'available':False})
                continue
            raw=observed.get((y,station),[])
            _, replay=audit_forecast_rows(raw,station,y)
            for k in ('valid_days','duplicate_rows','conflicting_days','invalid_rows'):
                if int(replay[k])!=int(m[k]): raise AssertionError('Saved manifest failed replay: '+k)
            if replay['station_mean_p12_pct'] is not None and not np.isclose(replay['station_mean_p12_pct'],float(m.station_mean_p12_pct),rtol=0,atol=1e-10):
                raise AssertionError('Station mean failed replay')
            rows=md_day_map(raw,station,y); daily.extend(rows)
            stats.append({'year':y,'station':station,'source_status':'retrieved','available':True,
                'valid_days':replay['valid_days'],'missing_or_unusable_days':50-replay['valid_days'],
                'duplicate_rows':replay['duplicate_rows'],'conflicting_days':replay['conflicting_days'],
                'invalid_rows':replay['invalid_rows'],'usable_original':replay['usable'],
                'station_mean_p12_pct':replay['station_mean_p12_pct'],
                'missing_dates':','.join(r['date'] for r in rows if r['state']!='valid')})
    return pd.DataFrame(daily),pd.DataFrame(stats)

def md_annual_features(daily, stats):
    rows=[]
    for y in MD_YEARS:
        g=stats[stats.year.eq(y)]
        if len(g)!=3 or not g.available.fillna(False).all():
            rows.append({'year':y,'source_complete':False,'common_days':np.nan,'strict_50':False,
                         'original_pct':np.nan,'common_pct':np.nan}); continue
        d=daily[daily.year.eq(y)]
        wide=d.pivot(index='date',columns='station',values='p12_pct').reindex(columns=MD_STATIONS)
        common=wide.dropna()
        original=float(g.station_mean_p12_pct.mean()) if g.usable_original.all() else np.nan
        common_pct=float(common.to_numpy().mean()) if g.usable_original.all() and len(common)>=40 else np.nan
        strict=bool(g.usable_original.all() and len(common)==50)
        rows.append({'year':y,'source_complete':True,'common_days':len(common),'strict_50':strict,
                     'original_pct':original,'common_pct':common_pct,
                     'common_minus_original_pct_points':common_pct-original if np.isfinite(original) and np.isfinite(common_pct) else np.nan})
    return pd.DataFrame(rows)

def md_forecast_comparisons(annual, dataset, segments, saved):
    data=dataset.copy()
    if data.year.duplicated().any(): raise ValueError('Duplicate forecast dataset year')
    data=data.merge(annual,on='year',how='left',validate='one_to_one')
    if data.original_pct.isna().any(): raise ValueError('Missing original replay on existing data')
    if not np.allclose(data.original_pct,data.forecast_mean_p12_pct,atol=1e-5,rtol=0):
        raise AssertionError('Rebuilt annual feature differs from original')
    segments={int(x['year']):x for x in segments}
    if set(segments)!=set(range(2009,2016)): raise ValueError('Incomplete phosphate records')
    for idx,row in data[data.year.between(2010,2015)].iterrows():
        y=int(row.year)
        if lh_date(segments[y-1]['filed_date'])>lh_date(str(row.as_of)[:10]) or lh_date(segments[y]['filed_date'])>lh_date(str(row.target_filed)[:10]):
            raise ValueError('Phosphate record chronology error')
        data.loc[idx,'target_q2_usd']-=float(segments[y]['phosphate_q2_usd'])
        data.loc[idx,'prior_q2_usd']-=float(segments[y-1]['phosphate_q2_usd'])
    records=[]; omissions=[]
    for mode in ('common','strict_50'):
        eligibility=data.common_pct.notna() if mode=='common' else data.strict_50.fillna(False)
        eligible=data[eligibility].copy()
        for y in range(2022,2027):
            match=eligible[eligible.year.eq(y)]
            if len(match)!=1:
                omissions.append({'mode':mode,'year':y,'reason':'test year does not meet coverage'}); continue
            row=next(match.itertuples()); cutoff=str(row.as_of)[:10]
            train=eligible[(eligible.year<y)&(eligible.target_filed.astype(str).str[:10]<=cutoff)].copy()
            if len(train)<6:
                omissions.append({'mode':mode,'year':y,'reason':'fewer than six matched training years'}); continue
            actual=float(row.target_q2_usd)
            for variant,feature in [('original','original_pct'),('common_day','common_pct')]:
                prediction=lh_fit(train,row,[feature])
                records.append({'cohort':mode,'year':y,'variant':variant,'training_years':len(train),
                    'training_year_list':','.join(map(str,train.year.astype(int))),
                    'prediction_usd':prediction,'actual_usd':actual,
                    'signed_error_usd':prediction-actual,'absolute_error_usd':abs(prediction-actual)})
    pred=pd.DataFrame(records,columns=['cohort','year','variant','training_years','training_year_list','prediction_usd','actual_usd','signed_error_usd','absolute_error_usd'])
    if pred.empty: scores=pd.DataFrame(columns=['cohort','variant','years','mae_musd'])
    else: scores=pred.groupby(['cohort','variant'],as_index=False).agg(years=('year','size'),mae_musd=('absolute_error_usd',lambda x:x.mean()/1e6))
    original=pred[(pred.cohort=='common')&(pred.variant=='original')]
    for row in original.itertuples():
        savedrow=saved[(saved.year.eq(row.year))&(saved.variant.eq('full_adjusted'))]
        if len(savedrow)!=1: raise ValueError('Saved full adjusted reference missing')
        prior=dataset[(dataset.year<row.year)&(dataset.target_filed.astype(str).str[:10]<=f'{row.year}-05-31')]
        if row.training_year_list==','.join(map(str,prior.sort_values('year').year.astype(int))) and not np.isclose(row.prediction_usd,float(savedrow.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
            raise AssertionError('Saved adjusted reference mismatch')
    return pred,scores,pd.DataFrame(omissions,columns=['mode','year','reason'])

def missing_forecast_days_research(audit, dataset, segments, saved):
    daily,stats=md_build_daily(audit)
    annual=md_annual_features(daily,stats)
    pred,scores,omitted=md_forecast_comparisons(annual,dataset,segments,saved)
    MD_OUT.mkdir(parents=True,exist_ok=True)
    for name,frame in [('daily_status_private.csv',daily),('station_coverage_private.csv',stats),
                       ('annual_features_private.csv',annual),('predictions_private.csv',pred),
                       ('scores_private.csv',scores),('omitted_private.csv',omitted)]: frame.to_csv(MD_OUT/name,index=False)
    print('Captured station-years:',int(stats.available.fillna(False).sum()),'/',len(stats))
    print('Station-years with fewer than 50 usable forecast days:',int((stats.valid_days.fillna(0)<50).sum()))
    print('Days with a non-valid state in retrieved selected records:',int((daily.state!='valid').sum()))
    print('Years with all 50 days valid at all three stations:',int(annual.strict_50.fillna(False).sum()),'/',len(annual))
    print('Annual original vs within-year common days (percentage points):')
    display(annual.round(6))
    print('Station-years with unavailable/short/conflicting/invalid days:')
    display(stats[(~stats.available.fillna(False)) | (stats.valid_days.fillna(0)<50) | (stats.conflicting_days.fillna(0)>0) | (stats.invalid_rows.fillna(0)>0)])
    print('Same-cohort model mean absolute errors (USD millions):'); display(scores.round(3))
    print('Omitted model test years:'); display(omitted)
    report='''This is a retrospective audit of saved selected forecast rows, not a fresh holdout. For each year common days are intersected among all three stations within that year; common-day means use the same dates/stations. Strict coverage requires 50 valid days at every station. The original rule accepts >=40 days per station and disallows invalid rows; conflicting values are excluded, duplicates counted once, and nulls omitted. Model variants in each comparison use exactly the same test and training rows. The precipitation values are forecasts of probability, not rainfall amounts. Browser preselection cannot prove source completeness, and archive revisions and historical forecast publication times remain unverified. No 2027 protocol changes. All outputs.
'''
    (MD_OUT/'REPORT.txt').write_text(report,encoding='utf-8'); print(report)
    return {'daily':daily,'station_coverage':stats,'annual':annual,'predictions':pred,'scores':scores,'omitted':omitted,'report':report}

MISSING_DAYS_RESULTS=missing_forecast_days_research(FORECAST_INPUT_AUDIT,DECOUPLED_FORECAST_RESULTS['data'],PHOSPHATE_SEGMENT,PHOSPHATE_FORECAST_RESULTS['predictions'])
```

```python
from pathlib import Path
import datetime as dt
import math
import numpy as np
import pandas as pd


def correction_predict(train, row, shrinkage=0.0):
    if shrinkage not in (0.0, 0.5, 1.0):
        raise ValueError('Unexpected intercept shrinkage')
    x=train['forecast_mean_p12_pct'].to_numpy(dtype=float)
    means=x.mean(); sd=x.std(ddof=0)
    if len(x)<6 or not np.isfinite(x).all() or not np.isfinite(sd) or sd<=1e-12:
        raise ValueError('Insufficient usable training variation')
    target=train['target_q2_usd'].to_numpy(dtype=float)
    baseline=train['prior_q2_usd'].to_numpy(dtype=float)*train['price_ratio'].to_numpy(dtype=float)
    row_baseline=float(row.prior_q2_usd)*float(row.price_ratio)
    feature=float(row.forecast_mean_p12_pct)
    if (not np.isfinite(target).all() or not np.isfinite(baseline).all() or
        (target<=0).any() or (baseline<=0).any() or not np.isfinite(row_baseline) or
        row_baseline<=0 or not np.isfinite(feature)):
        raise ValueError('Invalid revenue, price or forecast')
    residual=np.log(target/baseline)
    z=(x-means)/sd
    beta=float(z@residual/(z@z+1.0))
    intercept=float(shrinkage*residual.mean())
    exponent=intercept+(feature-means)/sd*beta
    if not np.isfinite(exponent) or abs(exponent)>700:
        raise ValueError('Invalid prediction exponent')
    return row_baseline*math.exp(exponent),intercept,beta


def adjusted_training_inputs(data, segments):
    required={'year','as_of','target_filed','target_q2_usd','prior_q2_usd',
              'price_ratio','forecast_mean_p12_pct'}
    if not required.issubset(data.columns) or data.year.duplicated().any():
        raise ValueError('Incomplete dataset or duplicate year')
    lookup={}
    for segment in segments:
        y=int(segment['year'])
        if y in lookup: raise ValueError('Duplicate phosphate year')
        amount=float(segment['phosphate_q2_usd'])
        if not np.isfinite(amount) or amount<0: raise ValueError('Invalid segment')
        lookup[y]=segment
    if set(lookup)!=set(range(2009,2016)): raise ValueError('Missing segment year')
    result=data.copy().sort_values('year').reset_index(drop=True)
    for idx,row in result.iterrows():
        y=int(row.year)
        if str(row.as_of)[:10]!=f'{y}-05-31' or str(row.target_filed)[:10]<=f'{y}-06-30':
            raise ValueError('Invalid prediction or target filing date')
        if y in range(2010,2016):
            prior,current=lookup[y-1],lookup[y]
            if str(prior['filed_date'])[:10]>str(row.as_of)[:10] or str(current['filed_date'])[:10]>str(row.target_filed)[:10]:
                raise ValueError('Segment chronology mismatch')
            result.loc[idx,'target_q2_usd']-=float(current['phosphate_q2_usd'])
            result.loc[idx,'prior_q2_usd']-=float(prior['phosphate_q2_usd'])
    if (result[['target_q2_usd','prior_q2_usd','price_ratio']]<=0).any().any():
        raise ValueError('Nonpositive adjusted revenue or price')
    return result


def intercept_comparison(data, segments, older_prior_years, saved_predictions=None):
    adjusted=adjusted_training_inputs(data,segments)
    prior_set=set(int(y) for y in older_prior_years)
    if not prior_set.issubset(set(range(2010,2016))): raise ValueError('Unexpected older cohort')
    result=[]; omissions=[]
    for cohort in ('full','prior'):
        rows=adjusted if cohort=='full' else adjusted[(adjusted.year>=2016)|adjusted.year.isin(prior_set)]
        for year in range(2016,2027):
            test=rows[rows.year.eq(year)]
            if len(test)!=1:
                omissions.append({'cohort':cohort,'year':year,'reason':'test year unavailable'}); continue
            row=next(test.itertuples()); cutoff=f'{year}-05-31'
            train=rows[(rows.year<year)&(rows.target_filed.astype(str).str[:10]<=cutoff)].sort_values('year')
            if len(train)<6:
                omissions.append({'cohort':cohort,'year':year,'reason':'fewer than six eligible training years'}); continue
            try:
                preds={f'alpha_{w:g}':correction_predict(train,row,w) for w in (0.0,0.5,1.0)}
            except ValueError as exc:
                omissions.append({'cohort':cohort,'year':year,'reason':str(exc)}); continue
            actual=float(row.target_q2_usd)
            if year in range(2022,2027) and saved_predictions is not None:
                ref=saved_predictions[(saved_predictions.year.eq(year)) & (saved_predictions.variant.eq(f'{cohort}_adjusted'))]
                if len(ref)!=1 or not np.isclose(preds['alpha_0'][0],float(ref.prediction_usd.iloc[0]),rtol=1e-10,atol=1e-3):
                    raise AssertionError('Existing adjusted prediction not reproduced')
            preds['price_only']=(float(row.prior_q2_usd*row.price_ratio),0.0,0.0)
            baseline=float(row.prior_q2_usd*row.price_ratio)
            preds['price_correction_only']=(baseline*math.exp(preds['alpha_0.5'][1]),preds['alpha_0.5'][1],0.0)
            for variant,(prediction,intercept,beta) in preds.items():
                result.append({'cohort':cohort,'year':year,'period':'2016-2021' if year<=2021 else '2022-2026',
                    'variant':variant,'training_years':len(train),'training_year_list':','.join(map(str,train.year.astype(int))),
                    'actual_usd':actual,'prediction_usd':prediction,'signed_error_usd':prediction-actual,
                    'absolute_error_usd':abs(prediction-actual),'intercept_log':intercept,'weather_coefficient':beta})
    pred=pd.DataFrame(result)
    if pred.empty: raise ValueError('No valid comparison years')
    scores=pred.groupby(['cohort','period','variant'],as_index=False).agg(years=('year','size'),mae_musd=('absolute_error_usd',lambda x:x.mean()/1e6))
    for key,group in pred.groupby(['cohort','year']):
        if len(group)!=5 or group.training_year_list.nunique()!=1 or group.actual_usd.nunique()!=1:
            raise AssertionError(f'Unmatched comparison: {key}')
    return {'predictions':pred,'scores':scores,'omitted':pd.DataFrame(omissions,columns=['cohort','year','reason'])}


def save_intercept_comparison(result, output_dir):
    out=Path(output_dir); out.mkdir(parents=True,exist_ok=True)
    for k in ('predictions','scores','omitted'):
        result[k].to_csv(out/(k+'_private.csv'),index=False)
    report=('Exploratory expanding-window comparison. Alpha_0 is the original zero-intercept fit; '
        'alpha_0.5 adds half the earlier training mean log ratio of revenue to the price baseline; '
        'alpha_1 adds its full mean, as a sensitivity. The weather slope uses the same centered '
        'training features and ridge penalty 1 in all fits. Price_only and price_correction_only '
        'are controls. The same '
        'training and test years are used within each comparison. Prior and full history show '
        'training-cohort sensitivity. Full-history 2016-2021 and both cohorts in 2022-2026 are '
        'reported, with any omitted years. Historical data and model choices were already '
        'examined, so these are not new holdouts. Historical weather publication and vintage '
        'remain unverified; older prices are reconstructed and phosphate-adjusted revenue is '
        'approximate. A correction is not automatically adopted based on this small sample. '
        'The frozen 2027 protocol remains unchanged.\n')
    (out/'REPORT.txt').write_text(report,encoding='utf-8')
    return report


INTERCEPT_RESULTS=intercept_comparison(
    DECOUPLED_FORECAST_RESULTS['data'],PHOSPHATE_SEGMENT,
    LONG_HISTORY_RESULTS['history']['year'].astype(int),
    PHOSPHATE_FORECAST_RESULTS['predictions'])
print(save_intercept_comparison(INTERCEPT_RESULTS,'/kaggle/working/iowa_intercept_research'))
print('Exploratory matched-fold mean absolute error (USD millions):')
display(INTERCEPT_RESULTS['scores'].round(3))
print('Omitted years:')
display(INTERCEPT_RESULTS['omitted'])
print('Year-by-year full cohort:')
display(INTERCEPT_RESULTS['predictions'].query("cohort == 'full'").pivot(index='year',columns='variant',values='absolute_error_usd').div(1e6).round(2))

adjusted_training_inputs(DECOUPLED_FORECAST_RESULTS['data'],PHOSPHATE_SEGMENT).to_csv('/kaggle/working/iowa_intercept_research/adjusted_annual_inputs_private.csv',index=False)
print('Saved matching adjusted annual input for the standalone runner.')
```

```python
from __future__ import annotations
import datetime as dt
import math
import numpy as np
import pandas as pd

INTERVAL_REQUIRED = ('year', 'decision_date', 'target_filed', 'model', 'prediction_usd',
            'actual_usd', 'training_year_list')


def _interval_date(value, label):
    if not isinstance(value, str):
        raise ValueError(f'{label} must be an ISO date')
    try:
        d = dt.date.fromisoformat(value)
    except ValueError as exc:
        raise ValueError(f'Invalid {label}') from exc
    if d.isoformat() != value:
        raise ValueError(f'{label} must be YYYY-MM-DD')
    return d


def _interval_training_years(value, prediction_year):
    if value is None or (isinstance(value, float) and np.isnan(value)) or value == '':
        return ()
    if not isinstance(value, str):
        raise ValueError('training_year_list must be a comma-separated string')
    try:
        parts = [int(x) for x in value.split(',')]
    except (TypeError, ValueError) as exc:
        raise ValueError('Invalid training year list') from exc
    if any(str(x) != raw.strip() for x, raw in zip(parts, value.split(','))):
        raise ValueError('Invalid training year list')
    if len(parts) != len(set(parts)) or any(y >= prediction_year for y in parts):
        raise ValueError('Training years must be unique and precede prediction year')
    return tuple(parts)


def validate_prediction_rows(predictions, expected_models=None):
    missing = set(INTERVAL_REQUIRED) - set(predictions.columns)
    if missing:
        raise ValueError(f'Missing prediction columns: {sorted(missing)}')
    p = predictions.copy()
    if p.empty:
        raise ValueError('No predictions supplied')
    years = pd.to_numeric(p.year, errors='raise')
    if years.isna().any() or not np.isfinite(years).all() or (years != np.floor(years)).any():
        raise ValueError('Invalid prediction year')
    p['year'] = years.astype(int)
    if p.model.isna().any() or any(not isinstance(x, str) or not x.strip() for x in p.model):
        raise ValueError('Missing model name')
    if p.duplicated(['year', 'model']).any():
        raise ValueError('Duplicate model-year prediction')
    p['prediction_usd'] = pd.to_numeric(p.prediction_usd, errors='raise')
    p['actual_usd'] = pd.to_numeric(p.actual_usd, errors='raise')
    if not np.isfinite(p.prediction_usd).all() or (p.prediction_usd <= 0).any():
        raise ValueError('Invalid prediction')
    if (p.actual_usd.dropna() <= 0).any() or not np.isfinite(p.actual_usd.dropna()).all():
        raise ValueError('Invalid actual')
    for row in p.itertuples():
        cutoff = _interval_date(row.decision_date, 'decision_date')
        if cutoff.year != row.year or cutoff != dt.date(row.year, 5, 31):
            raise ValueError('Decision date must be the May 31 prediction date')
        _interval_training_years(row.training_year_list, row.year)
        if pd.isna(row.actual_usd):
            if not pd.isna(row.target_filed) and row.target_filed != '':
                raise ValueError('Filing date supplied without an outcome')
        else:
            filed = _interval_date(row.target_filed, 'target_filed')
            if filed <= dt.date(row.year, 6, 30):
                raise ValueError('Target filing must follow quarter end')
    models = set(expected_models) if expected_models is not None else set(p.model)
    if not models:
        raise ValueError('No expected models')
    for year, group in p.groupby('year'):
        if set(group.model) != models:
            raise ValueError(f'Unmatched model years: {year}')
        if group.decision_date.nunique() != 1:
            raise ValueError('Decision dates differ within a year')
        if group.actual_usd.isna().nunique() != 1 or group.target_filed.fillna('').nunique() != 1:
            raise ValueError('Targets or filing dates differ within a year')
        if group.actual_usd.notna().all() and not np.allclose(group.actual_usd, group.actual_usd.iloc[0], rtol=0, atol=0.001):
            raise ValueError('Targets differ within a year')
        if group.training_year_list.fillna('').nunique() != 1:
            raise ValueError('Training years are not matched')
    actual_dates = p.drop_duplicates('year').set_index('year').target_filed.to_dict()
    for row in p.itertuples():
        for train_year in _interval_training_years(row.training_year_list, row.year):
            if train_year in actual_dates:
                filing = actual_dates[train_year]
                if pd.isna(filing) or not filing or _interval_date(filing, 'target_filed') > _interval_date(row.decision_date, 'decision_date'):
                    raise ValueError('Training target not filed by prediction date')
    return p.sort_values(['year', 'model']).reset_index(drop=True)


def sequential_ranges(predictions, expected_models=None, coverage=0.8):
    if not isinstance(coverage, (float, int)) or not math.isfinite(coverage) or not 0 < coverage < 1:
        raise ValueError('coverage must be strictly between 0 and 1')
    p = validate_prediction_rows(predictions, expected_models)
    result = []
    for row in p.itertuples(index=False):
        earlier = p[(p.model == row.model) & (p.year < row.year) & p.actual_usd.notna()].copy()
        earlier = earlier[earlier.target_filed <= row.decision_date].sort_values('year')
        errs = (earlier.prediction_usd - earlier.actual_usd).abs().to_numpy(dtype=float)
        n = len(errs)
        k = math.ceil((n + 1) * float(coverage))
        width = float(np.sort(errs)[k - 1]) if 1 <= k <= n else float('nan')
        lower = max(0.0, float(row.prediction_usd) - width) if np.isfinite(width) else float('nan')
        upper = float(row.prediction_usd) + width if np.isfinite(width) else float('nan')
        included = (bool(lower <= float(row.actual_usd) <= upper)
                    if np.isfinite(width) and pd.notna(row.actual_usd) else None)
        result.append({'year': int(row.year), 'decision_date': row.decision_date, 'model': row.model,
                       'prediction_usd': float(row.prediction_usd), 'actual_usd': row.actual_usd,
                       'target_filed': row.target_filed, 'calibration_years': ','.join(map(str, earlier.year)),
                       'calibration_count': n, 'nominal_coverage': float(coverage), 'rank': k,
                       'absolute_error_radius_usd': width, 'lower_usd': lower, 'upper_usd': upper,
                       'contains_actual': included})
    out = pd.DataFrame(result)
    summaries = []
    for model, g in out.groupby('model', sort=True):
        ev = g[g.contains_actual.notna()]
        inside = int(ev.contains_actual.sum()) if len(ev) else 0
        summaries.append({'model': model, 'total_prediction_years': len(g),
                          'ranges_available': int(g.lower_usd.notna().sum()),
                          'evaluated_ranges': len(ev), 'covered': inside,
                          'empirical_coverage': inside / len(ev) if len(ev) else float('nan'),
                          'first_evaluation_year': int(ev.year.min()) if len(ev) else None,
                          'last_evaluation_year': int(ev.year.max()) if len(ev) else None})
    return out, pd.DataFrame(summaries)


def summarize_range_periods(ranges, periods):
    needed = {'year', 'model', 'lower_usd', 'contains_actual'}
    if not needed.issubset(ranges.columns) or ranges.empty or not periods:
        raise ValueError('Missing ranges or periods')
    rows = []
    for label, bounds in periods.items():
        if not isinstance(label, str) or not label or len(bounds) != 2 or bounds[0] > bounds[1]:
            raise ValueError('Invalid period')
        part = ranges[ranges.year.between(bounds[0], bounds[1])]
        for model in sorted(ranges.model.unique()):
            g = part[part.model == model]
            observed = g[g.contains_actual.notna()]
            wins = int(observed.contains_actual.sum()) if len(observed) else 0
            rows.append({'period': label, 'model': model, 'prediction_years': len(g),
                         'ranges_available': int(g.lower_usd.notna().sum()),
                         'evaluated_ranges': len(observed), 'covered': wins,
                         'empirical_coverage': wins / len(observed) if len(observed) else float('nan')})
    return pd.DataFrame(rows)


from pathlib import Path

RANGE_MODELS = ('price_only', 'price_weather', 'price_shipments', 'price_weather_shipments')
RANGE_PERIODS = {'2016-2021': (2016, 2021), '2022-2026': (2022, 2026)}
RANGE_RESULTS, RANGE_SUMMARY = sequential_ranges(INCREMENTAL_RESULTS['predictions'], RANGE_MODELS, coverage=0.8)
RANGE_PERIOD_SUMMARY = summarize_range_periods(RANGE_RESULTS, RANGE_PERIODS)
RANGE_OUT = Path('/kaggle/working/iowa_exploratory_ranges')
RANGE_OUT.mkdir(parents=True, exist_ok=True)
RANGE_RESULTS.to_csv(RANGE_OUT/'prediction_ranges_private.csv', index=False)
RANGE_SUMMARY.to_csv(RANGE_OUT/'overall_coverage_private.csv', index=False)
RANGE_PERIOD_SUMMARY.to_csv(RANGE_OUT/'period_coverage_private.csv', index=False)
RANGE_REPORT = '''Sequential exploratory ranges around each prediction use that model's earlier absolute prediction errors, only when the target was filed by the new decision date. The 80% rank is ceil((n+1)*0.8); when the rank exceeds the number of errors the range is unavailable. Lower bounds are clipped at zero. Actual coverage is counted only when a range and outcome are both available. The period summaries include denominators. Prior model predictions are assumed to come from legitimate, consistent expanding-window folds and require independent verification. Model selection used already-examined years; forecast publication times, weather archive revision history, and some price vintage facts remain uncertain. The annual series is tiny and potentially nonstationary; nominal 80% should not be read as a calibrated guarantee. These ranges do not change the frozen 2027 protocol.'''
(RANGE_OUT/'REPORT.txt').write_text(RANGE_REPORT, encoding='utf8')
print(RANGE_REPORT)
print('Empirical range coverage by period (small samples):')
display(RANGE_PERIOD_SUMMARY.round(3))
print('Annual price-only prediction ranges, USD millions:')
_view=RANGE_RESULTS[RANGE_RESULTS.model.eq('price_only')].copy()
for _col in ('prediction_usd','actual_usd','lower_usd','upper_usd'):
    _view[_col]=_view[_col]/1e6
display(_view[['year','prediction_usd','actual_usd','lower_usd','upper_usd','calibration_count','contains_actual']].round(2))
```

```python
import base64, gzip, importlib, json, sys
from pathlib import Path
_source_tools = Path('/kaggle/working/iowa_research_tools')
_source_tools.mkdir(exist_ok=True)
for _name, _source in json.loads(gzip.decompress(base64.b64decode(''))).items():
    (_source_tools / _name).write_text(_source, encoding='utf8')
if str(_source_tools) not in sys.path: sys.path.insert(0, str(_source_tools))
importlib.invalidate_caches()
import alfred_chart, bls_capture, sec_capture, sec_index_capture, shipment_capture, source_link_state
alfred_chart = importlib.reload(alfred_chart)
bls_capture = importlib.reload(bls_capture)
sec_index_capture = importlib.reload(sec_index_capture)
sec_capture = importlib.reload(sec_capture)
shipment_capture = importlib.reload(shipment_capture)
source_link_state = importlib.reload(source_link_state)
ALFRED_EXPECTED_YEARS=list(range(2016,2027))
ALFRED_CAPTURES=json.loads(gzip.decompress(base64.b64decode("H4sIAAAAAAAAE+2aTYvbOBiA/4rQqQsTWd+yTSbLMr30UubQdmE3ZXBtzdjgxEFS2gml/71kssyEtAf5xawPo2PADo8tPQ/2i//9jg+2crjklOkrXFe7sHcWl9+xG775uzZselziZXCrZWhR3VfeX69x2z20dVu54BfBPoY1Rr4edvZ6fTxrjVecMrWgckEZYiWlJaXLLLSrZWh+9xfb/eaLdWu8ElwRtsxCM+bI7IltNJ8G8HGjiI7iOzvyyIevsB/2rrZ3e3e8n20IO19mWdXfO9sQH/ph3/l725DBPWTeus76P73tmuu/bz9SrThl+AqH1lbNy5Icf4248nro13j1tgr2Q7ex/11v/Im3bmj2tXXo1nW1Re+2jX1EXw7oZthshqYLB3Q/OHTT2k1XV71H1bZBf/V9Zxt0OjX4Er3vghse7LYK1qNP3TZUD7ZEp9VQC8GmpyqhSFwvGF0wdUJ62mTZ6ZYfV6ILvX1eiZYhHw79karp/K6vDiXqtn23tZMTLv2u2j7fHb+p+v5pv6E3zxvlj2V2PGi1zFp2ZHU2uM5+tc1dFXCJT1fGFlR94LxkuuQ5MUr/g3/8uDprgZmyBf+rawA+A+HTlKg4PpWTPLUgugUmtWCuFihBdK4uW5BP2QKQay8GjXMNwJeD+AoiI/kKYlILoluQpxbM0wJTUkEKJi5bUEzZAqhrBuIagK8APRfEvcFwowlPJYguQZFKMFsJcqLEZQk4nbIEINNeDBrn2ng+TiF8QseWSpr0hjBCPJpaMFcLGCe5+qUFbMIWwFx7MWicawA+BplsUh7bgiI/n2y+hhbArWMpBLOFICdG08sQ8ClDABHtTJ9xogH4OIBPS0GYiSPUWhFuXtm4AG4eTzGYKwack5xdzg25mDIGINnOBRorG4BRQJ4MCkqoiGMUhSSMyRSEOPtECsJsQdDE/PqaIKcMAky2M4HGygZglBBGWRAj4gYbQkoi6SsbHcDtkykIcwVBcGIKcxkENWUQYLKdCTRWNgAj6KsoQ4mmkdHSnHCVnhAi7VMpCLMFQRMlissgTPkJIky2c4HGygZghHwaJXNKRBGFqKgkOuUg2r2Ug5lyICmR4mmc+PknNMK2n5MsAAA=")).decode('utf8'))
for _item in ALFRED_CAPTURES: _item['capture']['capture_sha256']=alfred_chart.capture_digest(_item['capture'])
import json, os, subprocess, sys
from pathlib import Path
import alfred_chart, bls_capture, sec_capture, sec_index_capture, shipment_capture, source_link_state
_source_tools = Path('/kaggle/working/iowa_research_tools')
_prior_link_state = globals().get('_SOURCE_LINK_STATE')
_base_ledger = source_link_state.prepare(REPLAY_LEDGER, _prior_link_state)
_original_report = REPLAY_MODULE['project_replay_report'](REPLAY_LEDGER, range(2022, 2027))
_bls = json.loads(Path('/kaggle/working/iowa_price_audit/bls_dom_captures_private.json').read_text())
_sec = json.loads(Path('/kaggle/working/iowa_sec_audit/sec_q2_dom_captures_private.json').read_text())
_idx = json.loads(Path('/kaggle/working/iowa_sec_audit/sec_index_dom_captures_private.json').read_text())
_q1 = json.loads(Path('/kaggle/working/iowa_sec_audit/sec_q1_shipments_dom_captures_private.json').read_text())
_linked = bls_capture.compare_captures_to_ledger(_base_ledger, _bls,
             [(year, role) for year in range(2010, 2016) for role in ('current', 'previous')])
_linked = sec_capture.compare_captures_to_ledger(_linked, _sec, list(range(2016, 2027)))
_linked = sec_capture.attach_index_captures(_linked, _idx, [2017, 2018, *range(2020, 2027)])
_linked = shipment_capture.compare_captures_to_ledger(_linked, _q1,
             [('q1_shipments_' + role, year) for year in range(2016, 2027)
              for role in ('current', 'previous')])
if 'ALFRED_CAPTURES' in globals():
    if 'ALFRED_EXPECTED_YEARS' not in globals():
        raise ValueError('Declare expected ALFRED capture years')
    _alfred_captures = ALFRED_CAPTURES
    _alfred_years = ALFRED_EXPECTED_YEARS
elif 'ALFRED_CAPTURE' in globals():
    _alfred_captures = [{'year': 2025, 'capture': ALFRED_CAPTURE}]
    _alfred_years = [2025]
else:
    raise ValueError('Supply ALFRED chart evidence')
_alfred = alfred_chart.compare_captures_to_ledger(_linked, _alfred_captures, _alfred_years)
_linked = _alfred['ledger']
_alfred_digest = alfred_chart.batch_digest(_alfred_captures, _alfred_years)
if _prior_link_state is not None:
    alfred_chart.require_same_batch(_alfred_captures, _alfred_years,
                                    globals().get('_ALFRED_AUDIT_DIGEST'))
_alfred_matched = sum(d['status'] == 'matched' for d in _alfred['diagnostics'])
_counts = {key: sum(key in r for r in _linked) for key in
           ('bls_comparison', 'sec_10q_comparison', 'sec_10q_index_comparison',
            'sec_q1_shipment_comparison', 'alfred_chart_comparison')}
assert _counts == {'bls_comparison': 12, 'sec_10q_comparison': 11,
                   'sec_10q_index_comparison': 9, 'sec_q1_shipment_comparison': 22,
                   'alfred_chart_comparison': _alfred_matched}
_next_link_state = source_link_state.checkpoint(_base_ledger, _linked)
source_link_state.require_same_links(_linked, _prior_link_state)
_new_report = REPLAY_MODULE['project_replay_report'](_linked, range(2022, 2027))
assert _new_report == _original_report
_test_env = dict(os.environ, OPENBLAS_NUM_THREADS='1')
_test = subprocess.run([sys.executable, '-m', 'unittest', '-q', 'test_alfred_chart', 'test_bls_capture',
                        'test_sec_capture', 'test_sec_index_capture', 'test_shipment_capture',
                        'test_source_link_state'], cwd=_source_tools,
                       env=_test_env, text=True, capture_output=True, check=True)
REPLAY_LEDGER, REPLAY_REPORT = _linked, _new_report
_output = REPLAY_MODULE['save_private_ledger'](REPLAY_LEDGER, REPLAY_REPORT,
                                                '/kaggle/working/iowa_replay')
(_output / 'alfred_chart_captures_private.json').write_text(
    json.dumps(_alfred_captures, indent=2, ensure_ascii=False) + '\n', encoding='utf8')
(_output / 'alfred_chart_diagnostics_private.json').write_text(
    json.dumps(_alfred['diagnostics'], indent=2, ensure_ascii=False) + '\n', encoding='utf8')
if len(_alfred_captures) == 1 and _alfred_years == [2025]:
    (_output / 'alfred_chart_capture_private.json').write_text(
        json.dumps(_alfred_captures[0]['capture'], indent=2, ensure_ascii=False) + '\n', encoding='utf8')
(_output / 'source_link_summary_private.json').write_text(json.dumps({
    'counts': _counts, 'replay_eligibility_unchanged': True,
    'alfred_expected_years': _alfred['expected_years'],
    'alfred_captured_years': _alfred['captured_years'],
    'alfred_diagnostic_counts': {status: sum(d['status'] == status for d in _alfred['diagnostics'])
                                 for status in ('matched', 'mismatch', 'capture_missing',
                                                'ledger_missing', 'ledger_duplicate')},
    'limitation': 'Later selected DOM excerpts corroborate values, links, and an archived '
                  'ALFRED chart vintage; they do not establish historical availability, '
                  'original server bytes, or absence of later archive changes.'
}, indent=2) + '\n', encoding='utf8')
_SOURCE_LINK_STATE = _next_link_state
_ALFRED_AUDIT_DIGEST = _alfred_digest
print('Source links:', _counts)
print('ALFRED chart comparison statuses:', {status: sum(d['status'] == status for d in _alfred['diagnostics'])
      for status in ('matched', 'mismatch', 'capture_missing', 'ledger_missing', 'ledger_duplicate')})
print('Replay eligibility unchanged; historical availability and archive immutability remain unverified.')
print(_test.stderr)
```
